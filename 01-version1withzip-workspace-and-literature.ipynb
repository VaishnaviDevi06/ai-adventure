```python

from pathlib import Path
import pandas as pd
from collections import Counter

repo = Path("/kaggle/working/AILQA")

folders = [
    "Dataset_For_Context",
    "Raw_Data_Without_Preprocessing",
    "Test Set 1",
    "Test Set 2",
]

print("========== FILE AUDIT ==========")

for folder_name in folders:
    folder = repo / folder_name
    files = [p for p in folder.rglob("*") if p.is_file()
             and ".DS_Store" not in p.name
             and ".ipynb_checkpoints" not in str(p)]

    extensions = Counter(
        p.suffix.lower() or "[no extension]" for p in files
    )
    empty_files = [p for p in files if p.stat().st_size == 0]
    total_mb = sum(p.stat().st_size for p in files) / (1024 ** 2)

    print(f"\n{folder_name}")
    print("File count:", len(files))
    print("Total size: {:.2f} MB".format(total_mb))
    print("File types:", dict(extensions))
    print("Empty files:", len(empty_files))

print("\n========== TEST SET SCHEMAS ==========")

csv_files = [
    repo / "Test Set 1" / "test_questions_50.csv",
    repo / "Test Set 2" / "test_questions_100.csv",
]

for path in csv_files:
    print(f"\n--- {path.name} ---")
    if path.exists():
        df = pd.read_csv(path)
        print("Shape:", df.shape)
        print("Columns:", df.columns.tolist())
        print("\nFirst 3 rows:")
        print(df.head(3).to_string(index=False))
    else:
        print("CSV not found")
```

```python

from pathlib import Path
import pandas as pd

repo = Path("/kaggle/working/AILQA")

csv_paths = {
    "Test Set 1": repo / "Test Set 1" / "test_questions_50.csv",
    "Test Set 2": repo / "Test Set 2" / "test_questions_100.csv",
}

for name, path in csv_paths.items():
    df = pd.read_csv(path)

    print("\n" + "=" * 65)
    print(name)
    print("=" * 65)
    print("Rows, columns:", df.shape)
    print("\nMissing values:")
    print(df.isna().sum().to_string())

    question_col = "question"
    if question_col in df.columns:
        print("\nDuplicate questions:",
              df[question_col].duplicated().sum())

    for col in ["question", "ground_truth", "answer"]:
        if col in df.columns:
            lengths = df[col].dropna().astype(str).str.len()
            if not lengths.empty:
                print(
                    f"{col} length — min: {lengths.min()}, "
                    f"median: {lengths.median():.0f}, "
                    f"max: {lengths.max()}"
                )

    if "type" in df.columns:
        print("\nQuestion categories:")
        print(df["type"].value_counts(dropna=False).to_string())

print("\n" + "=" * 65)
print("SAMPLE LEGAL DOCUMENTS")
print("=" * 65)

for folder_name in [
    "Dataset_For_Context",
    "Raw_Data_Without_Preprocessing",
]:
    folder = repo / folder_name
    files = sorted(
        p for p in folder.glob("*.txt")
        if p.name != ".DS_Store"
    )

    print(f"\nFolder: {folder_name}")
    for path in files[:2]:
        text = path.read_text(encoding="utf-8", errors="replace")
        print(f"\nFile: {path.name} | Characters: {len(text)}")
        print(text[:700].replace("\n", " "))
```

```python

from pathlib import Path
from collections import Counter
import pandas as pd

repo = Path("/kaggle/working/AILQA")
raw = repo / "Raw_Data_Without_Preprocessing"

files = [
    p for p in raw.iterdir()
    if p.is_file() and p.name != ".DS_Store"
]

print("========== RAW FILE CLASSIFICATION ==========")
print("Top-level files:", len(files))

# Group files by naming pattern
groups = Counter()

for p in files:
    name = p.name.lower()

    if name.startswith("crime_judgement"):
        groups["Crime judgments"] += 1
    elif name.endswith(".csv"):
        groups["CSV files"] += 1
    elif "act" in name:
        groups["Acts / legislation (filename-based)"] += 1
    else:
        groups["Other text files"] += 1

for category, count in groups.most_common():
    print(f"{category}: {count}")

print("\n========== CSV FILE INSPECTION ==========")

for p in files:
    if p.suffix.lower() == ".csv":
        print(f"\nFile: {p.name}")
        try:
            df = pd.read_csv(p)
            print("Shape:", df.shape)
            print("Columns:", df.columns.tolist())
            print(df.head(3).to_string(index=False)[:2500])
        except Exception as e:
            print("Could not read CSV:", e)

print("\n========== OTHER RAW FILES ==========")

other_text = [
    p for p in files
    if p.suffix.lower() == ".txt"
    and not p.name.lower().startswith("crime_judgement")
    and "act" not in p.name.lower()
]

print("Other text files:", len(other_text))
for p in sorted(other_text)[:15]:
    print("-", p.name)
```

```python

from pathlib import Path
import pandas as pd
import re

corpus_dir = Path("/kaggle/working/AILQA/Dataset_For_Context")
output_dir = Path("/kaggle/working/indian_legal_rag/02_Dataset_Audit")
output_dir.mkdir(parents=True, exist_ok=True)

records = []

for path in sorted(corpus_dir.glob("*.txt")):
    text = path.read_text(encoding="utf-8", errors="replace")
    normalized = " ".join(text.split())

    # Extract the first apparent case title from the document header.
    title_match = re.search(
        r"Supreme Court of India\s+(.+?)\s+on\s+"
        r"\d{1,2}\s+\w+\s*,?\s+\d{4}",
        normalized,
        flags=re.IGNORECASE
    )
    case_title = title_match.group(1).strip() if title_match else ""

    date_match = re.search(
        r"\bon\s+(\d{1,2}\s+\w+\s*,?\s+\d{4})",
        normalized,
        flags=re.IGNORECASE
    )
    judgment_date = date_match.group(1) if date_match else ""

    if "supreme court of india" in normalized[:500].lower():
        court = "Supreme Court of India"
    elif "high court" in normalized[:500].lower():
        court = "High Court (needs verification)"
    else:
        court = "Unclassified"

    records.append({
        "document_id": path.stem,
        "filename": path.name,
        "case_title_candidate": case_title,
        "date_candidate": judgment_date,
        "court_candidate": court,
        "character_count": len(text),
    })

metadata = pd.DataFrame(records)
output_path = output_dir / "context_document_inventory.csv"
metadata.to_csv(output_path, index=False)

print("Documents inventoried:", len(metadata))
print("Saved to:", output_path)
print("\nCourt labels:")
print(metadata["court_candidate"].value_counts().to_string())
print("\nMissing case-title candidates:", (metadata["case_title_candidate"] == "").sum())
print("Missing date candidates:", (metadata["date_candidate"] == "").sum())
print("\nSample records:")
print(metadata.head(5).to_string(index=False))
```

```python

from pathlib import Path
import pandas as pd
import re

corpus_dir = Path("/kaggle/working/AILQA/Dataset_For_Context")
audit_dir = Path("/kaggle/working/indian_legal_rag/02_Dataset_Audit")
audit_dir.mkdir(parents=True, exist_ok=True)

records = []

for path in sorted(corpus_dir.glob("*.txt")):
    text = path.read_text(encoding="utf-8", errors="replace")
    normalized = " ".join(text.split())
    start = normalized[:1500]
    low = start.lower()

    case_match = re.search(
        r"Supreme Court of India\s+(.+?)\s+on\s+"
        r"\d{1,2}\s+\w+\s*,?\s+\d{4}",
        normalized,
        flags=re.IGNORECASE
    )

    act_match = re.search(
        r"\b(THE\s+)?([A-Z][A-Z\s(),&'-]{2,100}\bACT,?\s*\d{4})",
        start,
        flags=re.IGNORECASE
    )

    if "supreme court of india" in low or "high court" in low:
        doc_type = "judgment_candidate"
    elif act_match or re.search(r"\bact,\s*\d{4}\b", start, re.I):
        doc_type = "legislation_candidate"
    else:
        doc_type = "other_or_unknown"

    if case_match:
        title = case_match.group(1).strip()
    elif act_match:
        title = " ".join(act_match.group(0).split()).strip()
    else:
        title = ""

    date_match = re.search(
        r"\bon\s+(\d{1,2}\s+\w+\s*,?\s+\d{4})",
        normalized,
        flags=re.IGNORECASE
    )

    records.append({
        "document_id": path.stem,
        "filename": path.name,
        "document_type_candidate": doc_type,
        "title_candidate": title,
        "date_candidate": date_match.group(1) if date_match else "",
        "character_count": len(text),
    })

inventory = pd.DataFrame(records)
out = audit_dir / "context_document_inventory_v2.csv"
inventory.to_csv(out, index=False)

print("Saved:", out)
print("\nDocument type candidates:")
print(inventory["document_type_candidate"].value_counts().to_string())

print("\nMissing title candidates by type:")
print(
    inventory.assign(
        title_missing=inventory["title_candidate"].eq("")
    ).groupby("document_type_candidate")["title_missing"]
    .sum().to_string()
)

print("\nSample legislation candidates:")
print(
    inventory[inventory["document_type_candidate"] == "legislation_candidate"]
    [["document_id", "title_candidate"]]
    .head(10).to_string(index=False)
)
```

```python

from pathlib import Path
import pandas as pd
import re

repo = Path("/kaggle/working/AILQA")
inventory_path = Path(
    "/kaggle/working/indian_legal_rag/02_Dataset_Audit/"
    "context_document_inventory_v2.csv"
)

inventory = pd.read_csv(inventory_path)

def classify_document(row):
    path = repo / "Dataset_For_Context" / row["filename"]
    text = path.read_text(encoding="utf-8", errors="replace")
    start = " ".join(text.split())[:2000]
    low = start.lower()

    # Legislation commonly starts with an Act or Code title.
    if (
        re.search(r"\bthe\s+indian penal code\b", low)
        or re.search(r"\bthe\s+.*\bact,?\s*\d{4}\b", low)
        or re.search(r"\bthe\s+.*\bcode\b", low[:500])
    ):
        return "legislation_candidate"

    # Article-style files often have explicit Title and Content fields.
    if re.match(r"\s*title\s*:", low):
        return "legal_article_or_explanatory_content"

    # Court headings suggest a judgment, but need not guarantee one.
    if (
        "supreme court of india" in low[:500]
        or "high court" in low[:500]
        or "criminal appellate jurisdiction" in low[:1000]
    ):
        return "judgment_candidate"

    return "other_or_unknown"

inventory["document_type_candidate"] = inventory.apply(
    classify_document, axis=1
)

out = inventory_path.parent / "context_document_inventory_v3.csv"
inventory.to_csv(out, index=False)

print("Saved:", out)
print("\nRevised classification:")
print(inventory["document_type_candidate"].value_counts().to_string())

print("\nExamples still classified as other/unknown:")
unknown = inventory[
    inventory["document_type_candidate"] == "other_or_unknown"
]

for _, row in unknown.head(10).iterrows():
    print(row["document_id"], "-", row["filename"])
```

```python

from pathlib import Path
import pandas as pd
import re

repo = Path("/kaggle/working/AILQA/Dataset_For_Context")
audit = Path("/kaggle/working/indian_legal_rag/02_Dataset_Audit")
audit.mkdir(parents=True, exist_ok=True)

records = []

for path in sorted(repo.glob("*.txt")):
    text = path.read_text(encoding="utf-8", errors="replace")
    start = " ".join(text.split())[:1200]
    low = start.lower()

    # First prioritize unmistakable judgment-style headings.
    is_judgment = (
        low.startswith("supreme court of india")
        or low.startswith("in the supreme court of india")
        or low.startswith("high court of")
        or "date of judgment" in low[:800]
        or "criminal appellate jurisdiction" in low[:800]
    )

    # Identify Acts/codes only when the opening text resembles a title.
    is_legislation = bool(re.search(
        r"^(?:\d+\s+)?(?:the\s+)?"
        r"(?:indian penal code|indian evidence act|"
        r"[\w\s(),&'’.-]{2,100}\bact,?\s*\d{4}|"
        r"[\w\s(),&'’.-]{2,100}\bcode\b)"
        r"(?=\s|$)",
        start,
        flags=re.IGNORECASE
    ))

    is_article = bool(re.match(r"^title\s*:", low))

    if is_judgment:
        category = "judgment_candidate"
    elif is_article:
        category = "legal_article_or_explanatory_content"
    elif is_legislation:
        category = "legislation_candidate"
    else:
        category = "other_or_unknown"

    records.append({
        "document_id": path.stem,
        "filename": path.name,
        "document_type_candidate": category,
        "character_count": len(text),
    })

df = pd.DataFrame(records)
out = audit / "context_document_inventory_v4.csv"
df.to_csv(out, index=False)

print("Saved:", out)
print("\nClassification counts:")
print(df["document_type_candidate"].value_counts().to_string())

print("\nSample legislation candidates:")
print(
    df[df["document_type_candidate"] == "legislation_candidate"]
    [["document_id", "filename"]]
    .head(15).to_string(index=False)
)

print("\nSample unknown documents:")
print(
    df[df["document_type_candidate"] == "other_or_unknown"]
    [["document_id", "filename"]]
    .head(15).to_string(index=False)
)
```

```python
from pathlib import Path
import re

corpus_dir = Path("/kaggle/working/AILQA/Dataset_For_Context")

targets = {
    "CrPC Section 161": [
        r"section\s*161",
        r"statement.*investigation",
    ],
    "CrPC Section 167": [
        r"section\s*167",
        r"police custody",
        r"judicial custody",
    ],
    "CrPC Section 173": [
        r"section\s*173",
        r"charge.?sheet",
        r"final report",
    ],
    "IPC Section 82": [
        r"section\s*82",
        r"act of a child",
        r"below the age of seven",
    ],
    "Evidence Act Section 106": [
        r"section\s*106",
        r"especially within.*knowledge",
    ],
}

files = list(corpus_dir.glob("*.txt"))
print("Corpus text files:", len(files))

for topic, patterns in targets.items():
    print("\n" + "=" * 75)
    print("TOPIC:", topic)

    matches = []

    for path in files:
        text = path.read_text(encoding="utf-8", errors="replace")
        text_lower = text.lower()

        found = [p for p in patterns if re.search(p, text_lower)]

        # Require at least one statutory reference or multiple relevant phrases
        if found and (
            any("section" in p for p in found)
            or len(found) >= 2
        ):
            for pattern in found[:2]:
                match = re.search(pattern, text_lower)
                if match:
                    start = max(0, match.start() - 250)
                    end = min(len(text), match.end() + 450)
                    excerpt = text[start:end].replace("\n", " ")
                    matches.append((path.name, pattern, excerpt))
                    break

        if len(matches) >= 3:
            break

    if not matches:
        print("No candidate passages found with these patterns.")
    else:
        for filename, pattern, excerpt in matches:
            print(f"\nDocument: {filename}")
            print(f"Matched pattern: {pattern}")
            print("Passage:", excerpt[:700])
```

```python
from pathlib import Path
import re

corpus_dir = Path("/kaggle/working/AILQA/Dataset_For_Context")

patterns = [
    r"statement.*Section 161.*contradict",
    r"Section 161.*contradict",
    r"Section 161.*corroborat",
    r"Section 162.*contradict",
    r"statements.*Section 161.*evidence",
]

results = []

for path in corpus_dir.glob("*.txt"):
    text = path.read_text(encoding="utf-8", errors="replace")

    for pattern in patterns:
        match = re.search(pattern, text, re.IGNORECASE | re.DOTALL)

        if match:
            start = max(0, match.start() - 300)
            end = min(len(text), match.end() + 500)

            results.append({
                "document": path.name,
                "pattern": pattern,
                "excerpt": text[start:end].replace("\n", " ")
            })
            break

print("Candidate documents found:", len(results))

for item in results[:10]:
    print("\n" + "=" * 75)
    print("Document:", item["document"])
    print("Pattern:", item["pattern"])
    print("Excerpt:", item["excerpt"][:900])
```

```python
from pathlib import Path
import re

corpus_dir = Path("/kaggle/working/AILQA/Dataset_For_Context")

results = []

for path in corpus_dir.glob("*.txt"):
    text = path.read_text(encoding="utf-8", errors="replace")

    # Find Section 161 and Section 162 references within the same passage
    for match in re.finditer(r"Section\s+161", text, re.IGNORECASE):
        start = max(0, match.start() - 700)
        end = min(len(text), match.end() + 1800)
        passage = text[start:end]

        if re.search(r"Section\s+162", passage, re.IGNORECASE):
            if re.search(
                r"contradict|corroborat|use.*evidence|not.*evidence|"
                r"proviso|witness",
                passage,
                re.IGNORECASE
            ):
                results.append({
                    "document": path.name,
                    "passage": passage.replace("\n", " ")
                })
                break

print("Relevant passage candidates:", len(results))

for item in results[:8]:
    print("\n" + "=" * 80)
    print("DOCUMENT:", item["document"])
    print(item["passage"][:2000])
```

```python
import pandas as pd
from pathlib import Path
import re

gold_path = Path(
    "/kaggle/working/indian_legal_rag/05_Gold_Evidence/phase3_gold_evidence_pilot.csv"
)
corpus_dir = Path("/kaggle/working/AILQA/Dataset_For_Context")

gold = pd.read_csv(gold_path)

def extract_passage(text, start_pattern, end_pattern, max_chars=3000):
    start_match = re.search(start_pattern, text, re.IGNORECASE)

    if not start_match:
        return None

    start = start_match.start()
    end_match = re.search(
        end_pattern,
        text[start + 1:],
        re.IGNORECASE
    )

    end = start + 1 + end_match.start() if end_match else min(
        len(text), start + max_chars
    )

    return text[start:min(end, start + max_chars)].strip()

# CRIM_001: Extract the statutory rule and its explanation.
text1 = (
    corpus_dir / "doc_1944.txt"
).read_text(encoding="utf-8", errors="replace")

passage1 = extract_passage(
    text1,
    r"Section 162 Cr\.?P\.?C\.? reads as under",
    r"On account of Section 162 Cr\.?P\.?C\.?",
    max_chars=3000
)

if passage1:
    gold.loc[
        gold["question_id"] == "CRIM_001",
        "evidence_passage"
    ] = passage1
    gold.loc[
        gold["question_id"] == "CRIM_001",
        "verification_status"
    ] = "Extracted from source; legal review still required"

# CRIM_002: Extract the sentence about the charge sheet.
text2 = (
    corpus_dir / "doc_605.txt"
).read_text(encoding="utf-8", errors="replace")

match2 = re.search(
    r"The charge[- ]sheet was filed.{0,700}?"
    r"Section 173\(2\) of CrPC",
    text2,
    re.IGNORECASE | re.DOTALL
)

if match2:
    gold.loc[
        gold["question_id"] == "CRIM_002",
        "evidence_passage"
    ] = re.sub(r"\s+", " ", match2.group(0)).strip()
    gold.loc[
        gold["question_id"] == "CRIM_002",
        "verification_status"
    ] = "Extracted from source; passage directly supports answer"

gold.to_csv(gold_path, index=False)

print("Saved:", gold_path)
print("CRIM_001 passage extracted:", bool(passage1))
print("CRIM_002 passage extracted:", bool(match2))
print("Total records:", len(gold))
```

```python
from pathlib import Path
import re

corpus_dir = Path("/kaggle/working/AILQA/Dataset_For_Context")

patterns = [
    r"Section 32\s*\(\s*1\s*\).*dying declaration",
    r"dying declaration.{0,250}Section 32",
    r"statement.*treated as a dying declaration",
    r"oral dying declaration",
]

results = []

for path in corpus_dir.glob("*.txt"):
    text = path.read_text(encoding="utf-8", errors="replace")

    for pattern in patterns:
        match = re.search(pattern, text, re.IGNORECASE | re.DOTALL)

        if match:
            start = max(0, match.start() - 250)
            end = min(len(text), match.end() + 700)

            results.append({
                "document": path.name,
                "pattern": pattern,
                "passage": re.sub(
                    r"\s+", " ", text[start:end]
                ).strip()
            })
            break

print("Candidate documents:", len(results))

for item in results[:8]:
    print("\n" + "=" * 75)
    print("DOCUMENT:", item["document"])
    print("PATTERN:", item["pattern"])
    print(item["passage"][:1000])
```

```python
from pathlib import Path
import re

corpus_dir = Path("/kaggle/working/AILQA/Dataset_For_Context")

patterns = [
    r"anticipatory bail",
    r"Section 438 of the CrPC",
    r"Section 439 of the CrPC",
]

results = []

for path in corpus_dir.glob("*.txt"):
    text = path.read_text(encoding="utf-8", errors="replace")

    counts = [
        len(re.findall(pattern, text, re.IGNORECASE))
        for pattern in patterns
    ]

    if sum(counts) > 0:
        results.append({
            "document_id": path.name,
            "anticipatory_bail": counts[0],
            "section_438": counts[1],
            "section_439": counts[2],
            "total_matches": sum(counts),
            "preview": re.sub(r"\s+", " ", text[:250])
        })

results.sort(key=lambda item: item["total_matches"], reverse=True)

print("Matching judgments:", len(results))

for item in results[:10]:
    print("\n", item["document_id"])
    print("Anticipatory bail:", item["anticipatory_bail"])
    print("Section 438:", item["section_438"])
    print("Section 439:", item["section_439"])
    print("Preview:", item["preview"])
```

```python
from pathlib import Path
import pandas as pd
import re

csv_path = Path(
    "/kaggle/working/indian_legal_rag/05_Gold_Evidence/"
    "phase3_gold_evidence_pilot.csv"
)
corpus_dir = Path("/kaggle/working/AILQA/Dataset_For_Context")

gold = pd.read_csv(csv_path)

required_columns = [
    "question_id",
    "question",
    "reference_answer",
    "evidence_document",
    "case_name",
    "court",
    "judgment_date",
    "evidence_passage",
    "verification_status",
    "answer",
]

print("Total records:", len(gold))
print("Unique question IDs:", gold["question_id"].nunique())
print("Duplicate question IDs:", gold["question_id"].duplicated().sum())

missing_columns = [c for c in required_columns if c not in gold.columns]
print("Missing required columns:", missing_columns)

if not missing_columns:
    missing_by_column = gold[required_columns].isna().sum()
    print("\nMissing values by column:")
    print(missing_by_column.to_string())

    results = []

    for _, row in gold.iterrows():
        source = corpus_dir / str(row["evidence_document"])
        exists = source.is_file()
        exact_match = False

        if exists and pd.notna(row["evidence_passage"]):
            source_text = source.read_text(
                encoding="utf-8", errors="replace"
            )
            normalize = lambda s: re.sub(r"\s+", " ", str(s)).strip()
            exact_match = (
                normalize(row["evidence_passage"])
                in normalize(source_text)
            )

        results.append({
            "question_id": row["question_id"],
            "source_exists": exists,
            "evidence_match": exact_match,
        })

    audit = pd.DataFrame(results)
    print("\nSource and evidence validation:")
    print(audit.to_string(index=False))
    print("\nAll sources exist:", audit["source_exists"].all())
    print("All evidence passages match:", audit["evidence_match"].all())
```

```python

from pathlib import Path

source_path = Path("/kaggle/working/AILQA/Dataset_For_Context/doc_5970.txt")

text = source_path.read_text(encoding="utf-8", errors="ignore")

keywords = [
    "normal rule should be not to limit",
    "period of time",
    "anticipatory bail"
]

for keyword in keywords:
    print(f"\n{'='*20} {keyword} {'='*20}")
    start = 0
    count = 0

    while True:
        pos = text.lower().find(keyword.lower(), start)
        if pos == -1:
            break

        print(text[max(0, pos - 700):min(len(text), pos + 1400)])
        start = pos + len(keyword)
        count += 1

        if count >= 3:
            break

    if count == 0:
        print("No match found.")
```

```python

from pathlib import Path
import re
import pandas as pd

csv_path = Path(
    "/kaggle/working/indian_legal_rag/05_Gold_Evidence/"
    "phase3_gold_evidence_pilot.csv"
)
source_path = Path(
    "/kaggle/working/AILQA/Dataset_For_Context/doc_5970.txt"
)

gold = pd.read_csv(csv_path)
source = source_path.read_text(encoding="utf-8", errors="ignore")

mask = gold["question_id"].eq("CRIM_006")
if mask.sum() != 1:
    raise ValueError("Expected exactly one CRIM_006 record.")

# Extract the relevant holding from the original judgment.
start_phrase = (
    "Thus, considering the observations made by the Constitution Bench "
    "of this Court in the case of Gurbaksh Singh Sibbia"
)
end_phrase = (
    "However, as observed hereinabove, the normal rule should be not "
    "to limit the order in relation to a period of time."
)

def flexible_whitespace_pattern(phrase):
    return r"\s+".join(re.escape(part) for part in phrase.split())

start_match = re.search(
    flexible_whitespace_pattern(start_phrase), source, re.IGNORECASE
)
if not start_match:
    raise ValueError("Start of paragraph 7.5 not found; CSV unchanged.")

end_match = re.search(
    flexible_whitespace_pattern(end_phrase),
    source[start_match.start():],
    re.IGNORECASE
)
if not end_match:
    raise ValueError("End of paragraph 7.5 excerpt not found; CSV unchanged.")

end_pos = start_match.start() + end_match.end()
passage = source[start_match.start():end_pos].strip()

# Verify the extracted passage comes directly from the source.
if passage not in source:
    raise ValueError("Exact source-match check failed; CSV unchanged.")

revised_answer = (
    "The normal rule is not to limit anticipatory bail to a fixed period. "
    "However, the court may limit its operation when the circumstances "
    "warrant it, considering factors such as the stage at which the "
    "anticipatory bail application is made."
)

gold.loc[mask, "evidence_passage"] = passage
gold.loc[mask, "answer"] = revised_answer
gold.loc[mask, "reference_answer"] = revised_answer
gold.loc[mask, "verification_status"] = (
    "Source-matched; answer aligned to selected passage; legal review pending"
)

gold.to_csv(csv_path, index=False)

print("CRIM_006 updated successfully.")
print("Evidence passage matches source:", passage in source)
print("Evidence passage length:", len(passage), "characters")
print("Updated answer:", revised_answer)
print("Total records:", len(gold))
```

```python
from pathlib import Path
import pandas as pd
import shutil

project = Path("/kaggle/working/indian_legal_rag")

gold_path = (
    project / "05_Gold_Evidence" /
    "phase3_gold_evidence_pilot.csv"
)

benchmark_path = (
    project / "08_Evaluation" /
    "phase4_qa_benchmark_v1.csv"
)

# Confirm both files exist
if not gold_path.is_file() or not benchmark_path.is_file():
    raise FileNotFoundError("Required CSV file is missing.")

gold = pd.read_csv(gold_path)
benchmark = pd.read_csv(benchmark_path)

# Safety checks
expected_ids = {f"CRIM_{i:03d}" for i in range(1, 11)}

benchmark_ids = set(
    benchmark["question_id"].astype(str).str.strip()
)

if len(benchmark) != 10 or benchmark_ids != expected_ids:
    raise ValueError(
        "Benchmark does not contain exactly CRIM_001–CRIM_010. "
        "No files changed."
    )

# Create a backup without overwriting an existing backup
backup_path = gold_path.with_name(
    "phase3_gold_evidence_pilot_before_restore.csv"
)

if backup_path.exists():
    raise FileExistsError(
        f"Backup already exists: {backup_path}. "
        "No files changed; inspect the backup first."
    )

shutil.copy2(gold_path, backup_path)

# Restore Phase 3 from the validated 10-record benchmark
benchmark.to_csv(gold_path, index=False)

# Verify the restored file
restored = pd.read_csv(gold_path)

restored_ids = set(
    restored["question_id"].astype(str).str.strip()
)

assert len(restored) == 10
assert restored_ids == expected_ids
assert restored["question_id"].is_unique

print("RESTORATION SUCCESSFUL")
print("Restored records:", len(restored))
print("Question IDs:", sorted(restored_ids))
print("Backup saved at:", backup_path)
print("Frozen benchmark was not modified.")
```

```python
from pathlib import Path
import pandas as pd

project = Path("/kaggle/working/indian_legal_rag")

gold_path = (
    project / "05_Gold_Evidence" /
    "phase3_gold_evidence_pilot.csv"
)

source_dir = Path("/kaggle/working/AILQA/Dataset_For_Context")

df = pd.read_csv(gold_path)

print("PHASE 3 — EVIDENCE VALIDATION")
print("=" * 45)

source_results = []

for _, row in df.iterrows():
    qid = str(row["question_id"])
    source_path = source_dir / str(row["evidence_document"])

    if not source_path.is_file():
        source_results.append(
            (qid, False, False, "SOURCE FILE MISSING")
        )
        continue

    source_text = source_path.read_text(
        encoding="utf-8",
        errors="replace"
    )

    passage = str(row["evidence_passage"])
    matched = passage in source_text

    source_results.append((
        qid,
        True,
        matched,
        "PASS" if matched else "PASSAGE MISMATCH"
    ))

results = pd.DataFrame(
    source_results,
    columns=[
        "question_id",
        "source_exists",
        "passage_matches",
        "result"
    ]
)

print("Total records:", len(df))
print("Unique IDs:", df["question_id"].nunique())
print("Duplicate IDs:", df["question_id"].duplicated().sum())
print("\nIndividual results:")
print(results.to_string(index=False))

print("\nSource files found:",
      results["source_exists"].sum(), "/", len(df))

print("Evidence passages matched:",
      results["passage_matches"].sum(), "/", len(df))

if (
    len(df) == 10
    and df["question_id"].nunique() == 10
    and results["source_exists"].all()
    and results["passage_matches"].all()
):
    print("\nRESULT: TECHNICAL VALIDATION PASSED")
else:
    print("\nRESULT: ISSUES FOUND — DO NOT PROCEED YET")
```

```python
from pathlib import Path
import pandas as pd
import hashlib
import shutil

project = Path("/kaggle/working/indian_legal_rag")

phase3 = (
    project / "05_Gold_Evidence"
    / "phase3_gold_evidence_pilot.csv"
)

phase4 = (
    project / "08_Evaluation"
    / "phase4_qa_benchmark_v1.csv"
)

frozen = (
    project / "08_Evaluation"
    / "phase4_qa_benchmark_v1_frozen.csv"
)

print("PHASE 3 EXISTS:", phase3.exists())
print("PHASE 4 EXISTS:", phase4.exists())
print("FROZEN BENCHMARK EXISTS:", frozen.exists())

if not phase3.exists() or not phase4.exists():
    raise FileNotFoundError(
        "A required CSV is missing. Check the paths printed above."
    )

current = pd.read_csv(phase3)
benchmark = pd.read_csv(phase4)

print("\nCurrent Phase 3 records:", len(current))
print("Current Phase 3 IDs:", current["question_id"].tolist())
print("Phase 4 records:", len(benchmark))
print("Phase 4 IDs:", benchmark["question_id"].tolist())

# Validate the recovery source before changing anything.
expected_ids = [f"CRIM_{i:03d}" for i in range(1, 11)]

if (
    len(benchmark) != 10
    or benchmark["question_id"].tolist() != expected_ids
    or benchmark["question_id"].duplicated().any()
):
    raise ValueError(
        "Phase 4 benchmark is not the expected 10-record dataset. "
        "No restoration performed."
    )

# Restore only if Phase 3 is missing records or has unexpected IDs.
if current["question_id"].tolist() != expected_ids:
    backup = phase3.with_name(
        "phase3_gold_evidence_pilot_before_recovery.csv"
    )

    if not backup.exists():
        shutil.copy2(phase3, backup)
        print("\nBackup created:", backup)

    benchmark.to_csv(phase3, index=False)
    print("Phase 3 restored from the unfrozen Phase 4 benchmark.")
else:
    print("\nPhase 3 already contains the expected 10 records.")

# Verify the result.
restored = pd.read_csv(phase3)

assert len(restored) == 10
assert restored["question_id"].tolist() == expected_ids

print("\nRESTORATION CHECK PASSED")
print("Phase 3 records:", len(restored))
print("Phase 3 IDs:", restored["question_id"].tolist())

if frozen.exists():
    print(
        "\nFrozen benchmark SHA-256:",
        hashlib.sha256(frozen.read_bytes()).hexdigest()
    )
    print("Frozen benchmark was not modified by this cell.")
```

```python
from pathlib import Path
import re
import shutil
import pandas as pd

# ============================================================
# 1. Define paths
# ============================================================

csv_path = Path(
    "/kaggle/working/indian_legal_rag/05_Gold_Evidence/"
    "phase3_gold_evidence_pilot.csv"
)

source_path = Path(
    "/kaggle/working/AILQA/Dataset_For_Context/doc_3994.txt"
)

if not csv_path.is_file():
    raise FileNotFoundError(f"Gold evidence CSV not found: {csv_path}")

if not source_path.is_file():
    raise FileNotFoundError(f"Source document not found: {source_path}")

# ============================================================
# 2. Load the CSV and source document
# ============================================================

gold = pd.read_csv(csv_path)

source = source_path.read_text(
    encoding="utf-8",
    errors="replace"
)

# ============================================================
# 3. Verify CRIM_007 exists exactly once
# ============================================================

gold["question_id"] = (
    gold["question_id"]
    .astype("string")
    .str.strip()
)

mask = gold["question_id"].eq("CRIM_007")

if mask.sum() != 1:
    raise ValueError(
        f"Expected exactly one CRIM_007 record, found {mask.sum()}.\n"
        f"Available IDs: {gold['question_id'].tolist()}"
    )

# ============================================================
# 4. Search source text flexibly
# ============================================================

def flexible_pattern(phrase):
    """
    Match a phrase while allowing multiple spaces and line breaks.
    """
    return r"\s+".join(
        re.escape(word) for word in phrase.split()
    )


start_phrase = "The right to free legal services is"

end_phrase = (
    "provided of course the accused does not object to the "
    "provision of such lawyer."
)

start_match = re.search(
    flexible_pattern(start_phrase),
    source,
    flags=re.IGNORECASE
)

if not start_match:
    print("Could not find the expected starting phrase.")

    for term in ["free legal services", "essential ingredient", "Article 21"]:
        matches = list(
            re.finditer(re.escape(term), source, flags=re.IGNORECASE)
        )

        print(f"\nSearch term: {term}")
        print("Occurrences:", len(matches))

        for match in matches[:3]:
            print(source[max(0, match.start()-150):match.end()+300])

    raise ValueError(
        "Starting phrase not found. CSV remains unchanged."
    )

# Search for the ending phrase after the starting phrase.
search_area = source[start_match.start():]

end_match = re.search(
    flexible_pattern(end_phrase),
    search_area,
    flags=re.IGNORECASE
)

if end_match:
    passage_end = start_match.start() + end_match.end()
else:
    # Diagnostic output only. Do not save an incomplete passage.
    print("The expected ending phrase was not found.")
    print("\nActual source text near the starting phrase:\n")
    print(source[start_match.start():start_match.start() + 2500])

    raise ValueError(
        "Could not find the expected ending. "
        "CSV remains unchanged. Inspect the printed source excerpt."
    )

# Preserve the original source formatting in the extracted passage.
passage = source[start_match.start():passage_end].strip()

# Verify that the extracted passage is an exact substring of the source.
if not passage or passage not in source:
    raise ValueError(
        "Extracted passage failed exact source verification. "
        "CSV remains unchanged."
    )

# ============================================================
# 5. Prepare the revised answer
# ============================================================

revised_answer = (
    "Free legal services are an essential element of a reasonable, "
    "fair and just procedure under Article 21 for an accused who "
    "cannot obtain a lawyer because of poverty, indigence or similar "
    "circumstances. The State must provide legal representation when "
    "the circumstances and needs of justice require it, provided "
    "the accused does not object."
)

# ============================================================
# 6. Check required columns before changing anything
# ============================================================

required_columns = [
    "evidence_passage",
    "answer",
    "reference_answer",
    "verification_status"
]

missing_columns = [
    column for column in required_columns
    if column not in gold.columns
]

if missing_columns:
    raise ValueError(
        f"Required columns missing from CSV: {missing_columns}. "
        "CSV remains unchanged."
    )

# ============================================================
# 7. Create a backup before modifying the CSV
# ============================================================

backup_path = csv_path.with_name(
    "phase3_gold_evidence_pilot_before_CRIM_007_fix.csv"
)

if not backup_path.exists():
    shutil.copy2(csv_path, backup_path)
    print("Backup created:", backup_path)
else:
    print("Existing backup preserved:", backup_path)

# ============================================================
# 8. Update only CRIM_007
# ============================================================

gold.loc[mask, "evidence_passage"] = passage
gold.loc[mask, "answer"] = revised_answer
gold.loc[mask, "reference_answer"] = revised_answer

gold.loc[mask, "verification_status"] = (
    "Source-matched; answer aligned to selected passage; "
    "legal review pending"
)

# Save the updated CSV.
gold.to_csv(csv_path, index=False)

# ============================================================
# 9. Reload and verify the saved result
# ============================================================

verified = pd.read_csv(csv_path)

verified["question_id"] = (
    verified["question_id"]
    .astype("string")
    .str.strip()
)

updated_row = verified.loc[
    verified["question_id"].eq("CRIM_007")
]

if len(updated_row) != 1:
    raise RuntimeError(
        "Post-save verification failed: CRIM_007 count is not one."
    )

saved_passage = str(
    updated_row.iloc[0]["evidence_passage"]
)

if saved_passage not in source:
    raise RuntimeError(
        "Post-save verification failed: passage does not match source."
    )

# ============================================================
# 10. Print results
# ============================================================

print("\n" + "=" * 60)
print("CRIM_007 UPDATE COMPLETED")
print("=" * 60)

print("CSV updated:", csv_path)
print("Backup:", backup_path)
print("Total records:", len(verified))
print("Unique question IDs:", verified["question_id"].nunique())
print("CRIM_007 record count:", len(updated_row))
print("Evidence matches original source:", saved_passage in source)
print("Evidence passage length:", len(saved_passage))
print("Legal review status: PENDING")

print("\nExtracted evidence passage:\n")
print(saved_passage)

print("\nUpdated answer:\n")
print(updated_row.iloc[0]["answer"])
```

```python

from pathlib import Path
import pandas as pd

csv_path = Path(
    "/kaggle/working/indian_legal_rag/05_Gold_Evidence/"
    "phase3_gold_evidence_pilot.csv"
)
source_path = Path(
    "/kaggle/working/AILQA/Dataset_For_Context/doc_3994.txt"
)

gold = pd.read_csv(csv_path)
source = source_path.read_text(encoding="utf-8", errors="ignore")

start_phrase = "The right to free legal services is"
end_phrase = "provided of course the accused does not object"

start = source.lower().find(start_phrase.lower())
if start == -1:
    raise ValueError("Start phrase not found; CSV unchanged.")

end = source.lower().find(end_phrase.lower(), start)
if end == -1:
    # Print the nearby source text so we can use its actual ending.
    print(source[start:start + 2500])
    raise ValueError("Ending phrase not found; CSV unchanged.")

end += len(end_phrase)
passage = source[start:end].strip()

if passage not in source:
    raise ValueError("Exact source-match check failed; CSV unchanged.")

mask = gold["question_id"].eq("CRIM_007")
if mask.sum() != 1:
    raise ValueError("Expected exactly one CRIM_007 record.")

gold.loc[mask, "evidence_passage"] = passage
gold.to_csv(csv_path, index=False)

print("CRIM_007 passage cleaned successfully.")
print("Exact source match:", passage in source)
print("Passage length:", len(passage))
print("\nFinal passage:\n", passage)
```

```python

from pathlib import Path
import pandas as pd

csv_path = Path(
    "/kaggle/working/indian_legal_rag/05_Gold_Evidence/"
    "phase3_gold_evidence_pilot.csv"
)
source_path = Path(
    "/kaggle/working/AILQA/Dataset_For_Context/doc_3994.txt"
)

gold = pd.read_csv(csv_path)
source = source_path.read_text(encoding="utf-8", errors="ignore")

mask = gold["question_id"].eq("CRIM_007")
if mask.sum() != 1:
    raise ValueError("Expected exactly one CRIM_007 record.")

start_phrase = (
    "The right to free legal services is, therefore, clearly an essential "
    "ingredient of"
)
end_phrase = "provision of such lawyer."

start = source.find(start_phrase)
if start == -1:
    raise ValueError("Start phrase not found; CSV unchanged.")

end = source.find(end_phrase, start)
if end == -1:
    # Normalize whitespace and search again.
    import re
    normalized = re.sub(r"\s+", " ", source)
    normalized_start = normalized.find(start_phrase)
    normalized_end = normalized.find(end_phrase, normalized_start)

    if normalized_start == -1 or normalized_end == -1:
        raise ValueError(
            "Could not locate complete holding in source; CSV unchanged."
        )

    passage = normalized[
        normalized_start:normalized_end + len(end_phrase)
    ]
else:
    passage = source[start:end + len(end_phrase)].strip()

revised_answer = (
    "Free legal services are an essential element of a reasonable, fair "
    "and just procedure under Article 21 for an accused who cannot obtain "
    "a lawyer because of poverty, indigence or similar circumstances."
)

gold.loc[mask, "evidence_passage"] = passage
gold.loc[mask, "answer"] = revised_answer
gold.loc[mask, "reference_answer"] = revised_answer
gold.loc[mask, "verification_status"] = (
    "Source-matched; answer aligned to selected passage; legal review pending"
)

gold.to_csv(csv_path, index=False)

print("CRIM_007 updated successfully.")
print("Passage length:", len(passage))
print("Total records:", len(gold))
print("\nSelected passage:\n", passage)
```

```python

from pathlib import Path
import re
import pandas as pd

csv_path = Path(
    "/kaggle/working/indian_legal_rag/05_Gold_Evidence/"
    "phase3_gold_evidence_pilot.csv"
)
source_dir = Path("/kaggle/working/AILQA/Dataset_For_Context")

gold = pd.read_csv(csv_path)
mask = gold["question_id"].eq("CRIM_009")

if mask.sum() != 1:
    raise ValueError("Expected exactly one CRIM_009 record.")

filename = str(gold.loc[mask, "evidence_document"].iloc[0])
source_path = source_dir / filename

if not source_path.is_file():
    raise FileNotFoundError(f"Source file not found: {source_path}")

source = source_path.read_text(encoding="utf-8", errors="ignore")

start_phrase = "Therefore, free and fair trial is a sine qua non"
end_phrase = "could be violative of Article 14 of the Constitution."

def flexible_pattern(phrase):
    return r"\s+".join(re.escape(word) for word in phrase.split())

start_match = re.search(flexible_pattern(start_phrase), source, re.IGNORECASE)
if not start_match:
    raise ValueError("Start phrase not found; CSV unchanged.")

end_match = re.search(
    flexible_pattern(end_phrase),
    source[start_match.start():],
    re.IGNORECASE
)
if not end_match:
    raise ValueError("End phrase not found; CSV unchanged.")

end_pos = start_match.start() + end_match.end()
passage = source[start_match.start():end_pos].strip()

if passage not in source:
    raise ValueError("Exact source-match check failed; CSV unchanged.")

gold.loc[mask, "evidence_passage"] = passage
gold.loc[mask, "verification_status"] = (
    "Source-matched; answer-evidence alignment checked; legal review pending"
)

gold.to_csv(csv_path, index=False)

print("CRIM_009 passage cleaned successfully.")
print("Exact source match:", passage in source)
print("Passage length:", len(passage), "characters")
print("\nSelected passage:\n", passage)
print("\nTotal records:", len(gold))
```

```python

from pathlib import Path
import pandas as pd

csv_path = Path(
    "/kaggle/working/indian_legal_rag/05_Gold_Evidence/"
    "phase3_gold_evidence_pilot.csv"
)
source_dir = Path("/kaggle/working/AILQA/Dataset_For_Context")

gold = pd.read_csv(csv_path)

required_columns = [
    "question_id", "question", "reference_answer",
    "evidence_document", "case_name", "court",
    "judgment_date", "evidence_passage",
    "verification_status", "answer"
]

print("=" * 55)
print("PHASE 3 — GOLD EVIDENCE PILOT VALIDATION")
print("=" * 55)

missing_columns = [c for c in required_columns if c not in gold.columns]
if missing_columns:
    raise ValueError(f"Missing columns: {missing_columns}")

print("Total records:", len(gold))
print("Unique question IDs:", gold["question_id"].nunique())
print("Duplicate IDs:", gold["question_id"].duplicated().sum())

missing_values = gold[required_columns].isna().sum()
print("\nMissing values by field:")
print(missing_values[missing_values > 0].to_string() or "None")

empty_counts = {}
for col in required_columns:
    empty_counts[col] = (
        gold[col].astype("string").str.strip().eq("").sum()
    )
print("\nEmpty strings by field:")
print({k: v for k, v in empty_counts.items() if v > 0} or "None")

source_results = []
for _, row in gold.iterrows():
    path = source_dir / str(row["evidence_document"])
    if not path.is_file():
        source_results.append(
            (row["question_id"], False, False, "Source missing")
        )
        continue

    source = path.read_text(encoding="utf-8", errors="ignore")
    passage = str(row["evidence_passage"])
    matched = passage in source

    source_results.append((
        row["question_id"], True, matched,
        "OK" if matched else "Passage mismatch"
    ))

results = pd.DataFrame(
    source_results,
    columns=["question_id", "source_exists", "passage_matches", "result"]
)

print("\nSource files exist:", results["source_exists"].sum(), "/", len(gold))
print("Exact passage matches:", results["passage_matches"].sum(), "/", len(gold))

print("\nRecord-level source checks:")
print(results.to_string(index=False))

structural_ok = (
    len(gold) == 10
    and gold["question_id"].is_unique
    and not missing_values.any()
    and all(v == 0 for v in empty_counts.values())
)
source_ok = results["source_exists"].all() and results["passage_matches"].all()

print("\n" + "=" * 55)
print("STRUCTURAL VALIDATION:", "PASS" if structural_ok else "CHECK REQUIRED")
print("SOURCE VALIDATION:", "PASS" if source_ok else "CHECK REQUIRED")
print("LEGAL REVIEW: PENDING")
print("=" * 55)
```

```python
from pathlib import Path
import pandas as pd

base = Path("/kaggle/working/indian_legal_rag/05_Gold_Evidence")
gold_path = base / "phase3_gold_evidence_pilot.csv"
review_path = base / "phase3_record_review_log.csv"
report_path = base / "phase3_validation_report.md"

gold = pd.read_csv(gold_path)
review = pd.read_csv(review_path)

# Verify that both files contain the same 10 unique records.
expected_ids = set(gold["question_id"].astype(str))
if len(gold) != 10 or gold["question_id"].nunique() != 10:
    raise ValueError("Expected exactly 10 unique gold records.")

if len(review) != 10 or review["question_id"].nunique() != 10:
    raise ValueError("Review log must contain exactly 10 unique records.")

if expected_ids != set(review["question_id"].astype(str)):
    raise ValueError("Gold records and review-log IDs do not match.")

# Verify every evidence passage still matches its source file.
source_dir = Path("/kaggle/working/AILQA/Dataset_For_Context")

for _, row in gold.iterrows():
    source_path = source_dir / str(row["evidence_document"])
    if not source_path.is_file():
        raise FileNotFoundError(f"Missing source: {source_path}")

    source_text = source_path.read_text(
        encoding="utf-8", errors="replace"
    )
    if str(row["evidence_passage"]) not in source_text:
        raise ValueError(
            f"Evidence mismatch: {row['question_id']}"
        )

# Record the completed technical evidence-alignment checks.
# These statuses reflect the checks discussed in this workflow,
# not independent legal approval.
text_columns = [
    "answer_evidence_status", "all_claims_supported",
    "metadata_status", "legal_review_status",
    "current_law_status", "issues_found",
    "correction_made", "reviewer_notes",
    "final_record_status"
]
for col in text_columns:
    review[col] = review[col].astype("string")

for qid in expected_ids:
    mask = review["question_id"].astype(str).eq(qid)
    review.loc[mask, "answer_evidence_status"] = "PASS"
    review.loc[mask, "all_claims_supported"] = "YES"
    review.loc[mask, "legal_review_status"] = "PENDING"
    review.loc[mask, "current_law_status"] = "NEEDS_CURRENT_LAW_REVIEW"
    review.loc[mask, "final_record_status"] = "TECHNICAL_REVIEW_PASS"

    if qid != "CRIM_001":
        review.loc[mask, "metadata_status"] = "PENDING_SOURCE_HEADING_CHECK"

    old_notes = review.loc[mask, "reviewer_notes"].fillna("").iloc[0]
    addition = (
        " Technical source-passage match revalidated. "
        "Legal approval and current-law review remain pending."
    )
    if addition.strip() not in old_notes:
        review.loc[mask, "reviewer_notes"] = old_notes + addition

review.to_csv(review_path, index=False)

# Re-run core technical validation.
required = [
    "question_id", "question", "reference_answer",
    "evidence_document", "evidence_passage"
]
missing = int(gold[required].isna().sum().sum())
empty = int(
    gold[required].astype("string")
    .apply(lambda c: c.str.strip().eq("").sum()).sum()
)

summary = f"""# Phase 3 Technical Completion Report

- Gold records: {len(gold)}
- Unique question IDs: {gold['question_id'].nunique()}
- Duplicate question IDs: {gold['question_id'].duplicated().sum()}
- Missing required values: {missing}
- Empty required values: {empty}
- Source files verified: {len(gold)}/{len(gold)}
- Exact evidence passages verified: {len(gold)}/{len(gold)}
- Technical review-log PASS entries: {
    review['final_record_status'].eq('TECHNICAL_REVIEW_PASS').sum()
}

Legal review and current-law applicability are not certified by these
technical checks and remain pending.
"""

report_path.write_text(summary, encoding="utf-8")

print(summary)
print("Review log saved:", review_path)
print("Report saved:", report_path)
print("\nFinal record statuses:")
print(review[[
    "question_id", "answer_evidence_status",
    "metadata_status", "legal_review_status",
    "final_record_status"
]].to_string(index=False))
```

```python
from pathlib import Path
import pandas as pd

project = Path("/kaggle/working/indian_legal_rag")
benchmark_path = project / "08_Evaluation" / "phase4_qa_benchmark_v1.csv"
source_dir = Path("/kaggle/working/AILQA/Dataset_For_Context")

df = pd.read_csv(benchmark_path)

required = [
    "question_id", "question", "reference_answer",
    "answer", "evidence_document", "evidence_passage",
    "case_name", "court", "judgment_date"
]

# 1. Structural checks
missing_columns = [c for c in required if c not in df.columns]
if missing_columns:
    raise ValueError(f"Missing columns: {missing_columns}")

missing_values = int(df[required].isna().sum().sum())
empty_values = int(
    df[required].astype("string")
      .apply(lambda col: col.str.strip().eq("").sum()).sum()
)
duplicate_ids = int(df["question_id"].duplicated().sum())
duplicate_questions = int(
    df["question"].astype("string").str.strip().str.lower().duplicated().sum()
)

# 2. Source and evidence checks
source_found = 0
evidence_matches = 0
problems = []

for _, row in df.iterrows():
    qid = row["question_id"]
    source_path = source_dir / str(row["evidence_document"])

    if not source_path.is_file():
        problems.append(f"{qid}: source file missing")
        continue

    source_found += 1
    source_text = source_path.read_text(
        encoding="utf-8", errors="replace"
    )

    passage = str(row["evidence_passage"])
    if passage in source_text:
        evidence_matches += 1
    else:
        problems.append(f"{qid}: evidence passage mismatch")

# 3. Report
checks = {
    "Exactly 10 records": len(df) == 10,
    "Unique question IDs": duplicate_ids == 0,
    "No duplicate questions": duplicate_questions == 0,
    "No missing required values": missing_values == 0,
    "No empty required values": empty_values == 0,
    "All source files found": source_found == len(df),
    "All evidence passages match": evidence_matches == len(df),
}

print("PHASE 4 — BENCHMARK QUALITY CHECK")
print("-" * 40)
print("Records:", len(df))
print("Unique IDs:", df["question_id"].nunique())
print("Duplicate IDs:", duplicate_ids)
print("Duplicate questions:", duplicate_questions)
print("Missing required values:", missing_values)
print("Empty required values:", empty_values)
print(f"Source files found: {source_found}/{len(df)}")
print(f"Exact evidence matches: {evidence_matches}/{len(df)}")

print("\nIndividual checks:")
for name, passed in checks.items():
    print(f"{'PASS' if passed else 'FAIL'} — {name}")

if problems:
    print("\nIssues to inspect:")
    for problem in problems:
        print("-", problem)

print("\nOverall technical result:",
      "PASS" if all(checks.values()) else "NEEDS ATTENTION")

print("\nNote: These checks do not certify legal correctness,")
print("metadata accuracy, or answer-evidence reasoning.")
```

```python
from pathlib import Path
import re
import json
import pandas as pd

# ============================================================
# 1. Paths
# ============================================================

project = Path("/kaggle/working/indian_legal_rag")
source_dir = Path("/kaggle/working/AILQA/Dataset_For_Context")

output_dir = project / "04_Legal_Corpus"
output_dir.mkdir(parents=True, exist_ok=True)

chunks_csv = output_dir / "phase5_legal_chunks_v1.csv"
manifest_json = output_dir / "phase5_chunk_manifest.json"

# Protect existing outputs from accidental overwriting.
if chunks_csv.exists() or manifest_json.exists():
    raise FileExistsError(
        "Phase 5 chunk output already exists. Inspect it before rerunning:\n"
        f"{chunks_csv}\n{manifest_json}"
    )

# ============================================================
# 2. Chunking configuration
# ============================================================

MAX_CHARS = 1800
OVERLAP_CHARS = 250

def split_long_paragraph(text, max_chars=MAX_CHARS,
                         overlap=OVERLAP_CHARS):
    """Split long paragraphs with a small character overlap."""
    pieces = []
    start = 0

    while start < len(text):
        end = min(start + max_chars, len(text))

        # Prefer a sentence or word boundary near the limit.
        if end < len(text):
            boundary = max(
                text.rfind(". ", start, end),
                text.rfind("; ", start, end),
                text.rfind(" ", start, end)
            )
            if boundary > start + max_chars // 2:
                end = boundary + 1

        piece = text[start:end].strip()
        if piece:
            pieces.append(piece)

        if end >= len(text):
            break

        start = max(start + 1, end - overlap)

    return pieces


def chunk_document(text):
    """Split a document by paragraphs, then split oversized paragraphs."""
    text = text.replace("\r\n", "\n").replace("\r", "\n")
    paragraphs = [
        re.sub(r"\s+", " ", p).strip()
        for p in re.split(r"\n\s*\n", text)
    ]
    paragraphs = [p for p in paragraphs if p]

    chunks = []
    current = ""

    for paragraph in paragraphs:
        if len(paragraph) > MAX_CHARS:
            if current:
                chunks.append(current)
                current = ""

            chunks.extend(split_long_paragraph(paragraph))
            continue

        candidate = f"{current}\n\n{paragraph}" if current else paragraph

        if len(candidate) <= MAX_CHARS:
            current = candidate
        else:
            if current:
                chunks.append(current)
            current = paragraph

    if current:
        chunks.append(current)

    return chunks


# ============================================================
# 3. Load and chunk source documents
# ============================================================

if not source_dir.is_dir():
    raise FileNotFoundError(f"Source directory not found: {source_dir}")

source_files = sorted(source_dir.glob("*.txt"))

if not source_files:
    raise ValueError("No .txt source documents found.")

records = []
failed_files = []

for index, path in enumerate(source_files, start=1):
    try:
        text = path.read_text(encoding="utf-8", errors="replace")
        if not text.strip():
            continue

        document_chunks = chunk_document(text)

        for chunk_index, chunk in enumerate(document_chunks):
            records.append({
                "chunk_id": f"{path.stem}_chunk_{chunk_index:04d}",
                "document_id": path.stem,
                "source_file": path.name,
                "chunk_index": chunk_index,
                "text": chunk,
                "char_count": len(chunk)
            })

    except Exception as exc:
        failed_files.append({
            "source_file": path.name,
            "error": str(exc)
        })

    if index % 1000 == 0:
        print(f"Processed {index}/{len(source_files)} source files...")

# ============================================================
# 4. Validate and save
# ============================================================

chunks_df = pd.DataFrame(records)

if chunks_df.empty:
    raise ValueError("No chunks were created.")

if chunks_df["chunk_id"].duplicated().any():
    raise ValueError("Duplicate chunk IDs detected.")

if chunks_df["text"].isna().any() or (
    chunks_df["text"].astype(str).str.strip().eq("").any()
):
    raise ValueError("Empty chunks detected.")

chunks_df.to_csv(chunks_csv, index=False)

manifest = {
    "dataset_directory": str(source_dir),
    "source_file_count": len(source_files),
    "processed_chunk_count": len(chunks_df),
    "unique_documents_in_chunks": int(chunks_df["document_id"].nunique()),
    "max_chars": MAX_CHARS,
    "overlap_chars_for_long_paragraphs": OVERLAP_CHARS,
    "failed_file_count": len(failed_files),
    "failed_files": failed_files[:20],
    "chunking_method": "paragraph-first with bounded character splitting",
    "limitations": [
        "This is a baseline chunker, not a complete legal-structure parser.",
        "Section headings and case metadata are not yet explicitly extracted.",
        "Character limits are not equivalent to model token limits."
    ]
}

manifest_json.write_text(
    json.dumps(manifest, indent=2),
    encoding="utf-8"
)

# ============================================================
# 5. Print summary
# ============================================================

print("\n" + "=" * 60)
print("PHASE 5 — CHUNKING COMPLETE")
print("=" * 60)
print("Source files discovered:", len(source_files))
print("Documents represented in chunks:", chunks_df["document_id"].nunique())
print("Total chunks:", len(chunks_df))
print("Average chunk characters:", round(chunks_df["char_count"].mean(), 1))
print("Maximum chunk characters:", chunks_df["char_count"].max())
print("Failed files:", len(failed_files))
print("Chunks CSV:", chunks_csv)
print("Manifest:", manifest_json)

print("\nSample chunks:")
print(
    chunks_df[["chunk_id", "source_file", "char_count", "text"]]
    .head(3)
    .to_string(index=False, max_colwidth=220)
)
```

```python
import re
import json
import pandas as pd
from pathlib import Path
from datetime import datetime, timezone

# --------------------------------------------------
# PATHS
# --------------------------------------------------
SOURCE_DIR = Path("/kaggle/working/AILQA/Dataset_For_Context")
OUTPUT_DIR = Path("/kaggle/working/indian_legal_rag/04_Legal_Corpus")

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

OUTPUT_CSV = OUTPUT_DIR / "phase5_legal_chunks_v2.csv"
OUTPUT_MANIFEST = OUTPUT_DIR / "phase5_chunk_manifest_v2.json"

# Never overwrite existing files
if OUTPUT_CSV.exists() or OUTPUT_MANIFEST.exists():
    raise FileExistsError(
        "Version 2 output already exists. Inspect it before creating another version."
    )

# --------------------------------------------------
# CHUNKING SETTINGS
# --------------------------------------------------
TARGET_CHARS = 1600
MAX_CHARS = 1800
OVERLAP_CHARS = 200

def clean_text(text):
    # Normalize line endings and repeated whitespace
    text = text.replace("\r\n", "\n").replace("\r", "\n")
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n{3,}", "\n\n", text)
    return text.strip()

def split_into_units(text):
    """
    Prefer paragraph boundaries.
    If a paragraph is too long, split it at sentence boundaries.
    If a sentence itself is too long, split it at whitespace.
    """
    paragraphs = [
        p.strip()
        for p in re.split(r"\n\s*\n", text)
        if p.strip()
    ]

    # These source documents often have flattened paragraphs.
    # Therefore, sentence splitting is also applied within paragraphs.
    units = []

    for paragraph in paragraphs:
        if len(paragraph) <= TARGET_CHARS:
            # Keep short paragraphs intact.
            units.append(paragraph)
            continue

        sentences = re.split(r"(?<=[.!?])\s+(?=[A-Z0-9(\"'])", paragraph)

        for sentence in sentences:
            sentence = sentence.strip()
            if not sentence:
                continue

            if len(sentence) <= TARGET_CHARS:
                units.append(sentence)
            else:
                # Fallback for unusually long sentences or flattened text.
                words = sentence.split()
                current = []

                for word in words:
                    candidate = " ".join(current + [word])

                    if len(candidate) > TARGET_CHARS and current:
                        units.append(" ".join(current))
                        current = [word]
                    else:
                        current.append(word)

                if current:
                    units.append(" ".join(current))

    return units

def build_chunks(text):
    units = split_into_units(text)
    chunks = []
    current = ""

    for unit in units:
        candidate = f"{current}\n{unit}".strip() if current else unit

        if len(candidate) <= MAX_CHARS:
            current = candidate
        else:
            if current:
                chunks.append(current)

            # Carry a small overlap from the previous chunk.
            overlap = current[-OVERLAP_CHARS:].strip() if current else ""

            # Avoid starting overlap in the middle of a word.
            if overlap and " " in overlap:
                overlap = overlap.split(" ", 1)[1]

            if overlap and len(overlap) + len(unit) <= MAX_CHARS:
                current = overlap + "\n" + unit
            else:
                current = unit

            # Safety fallback if an individual unit exceeds MAX_CHARS.
            if len(current) > MAX_CHARS:
                pieces = re.findall(r".{1,%d}(?:\s+|$)" % MAX_CHARS, current)
                chunks.extend(piece.strip() for piece in pieces if piece.strip())
                current = ""

    if current:
        chunks.append(current)

    return chunks

# --------------------------------------------------
# PROCESS SOURCE DOCUMENTS
# --------------------------------------------------
source_files = sorted(SOURCE_DIR.glob("*.txt"))

if not source_files:
    raise FileNotFoundError(f"No .txt files found in {SOURCE_DIR}")

records = []
failed_files = []

for file_path in source_files:
    try:
        text = clean_text(
            file_path.read_text(encoding="utf-8", errors="replace")
        )

        if not text:
            failed_files.append({
                "source_file": file_path.name,
                "error": "Empty source document"
            })
            continue

        document_id = file_path.stem
        chunks = build_chunks(text)

        for index, chunk in enumerate(chunks):
            records.append({
                "chunk_id": f"{document_id}_v2_chunk_{index:04d}",
                "document_id": document_id,
                "source_file": file_path.name,
                "chunk_index": index,
                "text": chunk,
                "char_count": len(chunk)
            })

    except Exception as exc:
        failed_files.append({
            "source_file": file_path.name,
            "error": str(exc)
        })

df = pd.DataFrame(records)

if df.empty:
    raise ValueError("No chunks were generated.")

# --------------------------------------------------
# VALIDATE BEFORE SAVING
# --------------------------------------------------
assert df["chunk_id"].is_unique, "Duplicate chunk IDs detected."
assert df["text"].notna().all(), "Missing chunk text detected."
assert df["text"].str.strip().ne("").all(), "Empty chunks detected."
assert df["char_count"].max() <= MAX_CHARS, "Chunk exceeds maximum length."

df.to_csv(OUTPUT_CSV, index=False, encoding="utf-8")

manifest = {
    "version": "v2",
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "source_directory": str(SOURCE_DIR),
    "source_files_discovered": len(source_files),
    "documents_with_chunks": int(df["document_id"].nunique()),
    "total_chunks": len(df),
    "target_chars": TARGET_CHARS,
    "maximum_chars": MAX_CHARS,
    "overlap_chars": OVERLAP_CHARS,
    "failed_files": failed_files,
    "output_csv": str(OUTPUT_CSV),
    "chunking_method": (
        "Paragraph-first, sentence-aware splitting with a whitespace fallback "
        "for oversized units."
    )
}

OUTPUT_MANIFEST.write_text(
    json.dumps(manifest, indent=2, ensure_ascii=False),
    encoding="utf-8"
)

print("=" * 60)
print("PHASE 5.3 — IMPROVED CHUNKING COMPLETE")
print("=" * 60)
print("Source files discovered:", len(source_files))
print("Documents represented:", df["document_id"].nunique())
print("Total chunks:", len(df))
print("Average chunk length:", round(df["char_count"].mean(), 2))
print("Maximum chunk length:", df["char_count"].max())
print("Failed files:", len(failed_files))
print("Unique chunk IDs:", df["chunk_id"].nunique())
print("\nSaved chunks:", OUTPUT_CSV)
print("Saved manifest:", OUTPUT_MANIFEST)

print("\nSample improved chunks:")
for _, row in df.head(3).iterrows():
    print("\n" + "-" * 60)
    print("Chunk:", row["chunk_id"])
    print("Characters:", row["char_count"])
    print(row["text"][:700])
```

```python
import json
import pandas as pd
from pathlib import Path

# File paths
CHUNKS_FILE = Path(
    "/kaggle/working/indian_legal_rag/04_Legal_Corpus/phase5_legal_chunks_v2.csv"
)

MANIFEST_FILE = Path(
    "/kaggle/working/indian_legal_rag/04_Legal_Corpus/phase5_chunk_manifest_v2.json"
)

# Check files
assert CHUNKS_FILE.exists(), "Chunk CSV is missing."
assert MANIFEST_FILE.exists(), "Chunk manifest is missing."

df = pd.read_csv(CHUNKS_FILE)
with open(MANIFEST_FILE, "r", encoding="utf-8") as f:
    manifest = json.load(f)

print("=" * 60)
print("PHASE 5.4 — CHUNK VALIDATION")
print("=" * 60)

print("Total chunks:", len(df))
print("Unique chunk IDs:", df["chunk_id"].nunique())
print("Unique documents:", df["document_id"].nunique())
print("Missing text:", df["text"].isna().sum())
print("Empty text:", df["text"].fillna("").str.strip().eq("").sum())
print("Actual maximum length:", df["text"].fillna("").str.len().max())
print("Chunks exceeding 1800 characters:",
      (df["text"].fillna("").str.len() > 1800).sum())
print("Manifest count matches:", len(df) == manifest["total_chunks"])
print("Failed source files:", len(manifest["failed_files"]))

# Check chunk order
print(
    "Chunk indices consecutive:",
    all(
        sorted(group["chunk_index"].tolist())
        == list(range(len(group)))
        for _, group in df.groupby("document_id")
    )
)

# Inspect selected chunks
print("\nSAMPLE CHUNKS")
for index in [0, 1, 2, len(df) // 2, len(df) - 1]:
    row = df.iloc[index]
    print("\n" + "-" * 60)
    print("ID:", row["chunk_id"])
    print("Source:", row["source_file"])
    print("Length:", len(row["text"]))
    print(row["text"][:500])

# Sentence-boundary diagnostics
texts = df["text"].fillna("").str.strip()

starts_lowercase = texts.str.match(r"^[a-z]").sum()
ends_without_punctuation = (
    ~texts.str.endswith((".", "?", "!", ":", ";", ")", "]", '"', "'"))
).sum()

print("\nBOUNDARY DIAGNOSTICS")
print("Chunks starting lowercase:", starts_lowercase)
print("Chunks ending without common punctuation:", ends_without_punctuation)
```

```python
import json
import hashlib
import numpy as np
import pandas as pd
from pathlib import Path
from datetime import datetime, timezone
from sentence_transformers import SentenceTransformer

# --------------------------------------------------
# PATHS
# --------------------------------------------------
BASE_DIR = Path("/kaggle/working/indian_legal_rag/04_Legal_Corpus")

CHUNKS_FILE = BASE_DIR / "phase5_legal_chunks_v2.csv"
EMBEDDINGS_FILE = BASE_DIR / "phase5_dense_embeddings_v1.npy"
METADATA_FILE = BASE_DIR / "phase5_embedding_metadata_v1.csv"
MANIFEST_FILE = BASE_DIR / "phase5_embedding_manifest_v1.json"

CHECKPOINT_FILE = BASE_DIR / "phase5_embedding_checkpoint_v1.npy"
CHECKPOINT_META_FILE = BASE_DIR / "phase5_embedding_checkpoint_meta_v1.json"

MODEL_NAME = "sentence-transformers/all-MiniLM-L6-v2"
BATCH_SIZE = 64

# --------------------------------------------------
# SAFETY CHECKS
# --------------------------------------------------
if not CHUNKS_FILE.exists():
    raise FileNotFoundError(f"Missing chunk file: {CHUNKS_FILE}")

if any(p.exists() for p in [
    EMBEDDINGS_FILE, METADATA_FILE, MANIFEST_FILE
]):
    raise FileExistsError(
        "At least one final output already exists. "
        "Inspect the files before rerunning."
    )

chunks = pd.read_csv(CHUNKS_FILE)

required_columns = [
    "chunk_id", "document_id", "source_file",
    "chunk_index", "text"
]

missing_columns = set(required_columns) - set(chunks.columns)
if missing_columns:
    raise ValueError(f"Missing columns: {missing_columns}")

if chunks["chunk_id"].duplicated().any():
    raise ValueError("Duplicate chunk IDs detected.")

if chunks["text"].isna().any() or chunks["text"].str.strip().eq("").any():
    raise ValueError("Missing or empty chunk text detected.")

texts = chunks["text"].astype(str).tolist()
n = len(texts)

# Fingerprint the ordered chunk IDs so checkpoints cannot be
# accidentally reused for a different chunk ordering.
chunk_fingerprint = hashlib.sha256(
    "\n".join(chunks["chunk_id"].tolist()).encode("utf-8")
).hexdigest()

print("Chunks to embed:", n)
print("Batch size:", BATCH_SIZE)
print("Expected matrix size (MiB):",
      round(n * 384 * 4 / (1024**2), 2))

# --------------------------------------------------
# LOAD MODEL
# --------------------------------------------------
model = SentenceTransformer(MODEL_NAME)
embedding_dim = model.get_embedding_dimension()

if embedding_dim != 384:
    raise ValueError(f"Expected 384 dimensions, got {embedding_dim}")

# --------------------------------------------------
# INITIALIZE OR RESUME CHECKPOINT
# --------------------------------------------------
if CHECKPOINT_FILE.exists() != CHECKPOINT_META_FILE.exists():
    raise RuntimeError(
        "Only one checkpoint file exists. Inspect the checkpoint files "
        "before continuing; do not delete or overwrite them automatically."
    )

if CHECKPOINT_FILE.exists():
    with open(CHECKPOINT_META_FILE, "r", encoding="utf-8") as f:
        checkpoint_meta = json.load(f)

    if (
        checkpoint_meta["chunk_fingerprint"] != chunk_fingerprint
        or checkpoint_meta["model_name"] != MODEL_NAME
        or checkpoint_meta["total_chunks"] != n
    ):
        raise RuntimeError(
            "Checkpoint does not match the current chunks/model. "
            "No files were changed."
        )

    embeddings = np.load(CHECKPOINT_FILE, mmap_mode="r+")
    completed = int(checkpoint_meta["completed_chunks"])

    if embeddings.shape != (n, embedding_dim):
        raise RuntimeError("Checkpoint embedding shape is invalid.")

    if not 0 <= completed <= n:
        raise RuntimeError("Checkpoint progress value is invalid.")

    print("Resuming from chunk:", completed)

else:
    embeddings = np.lib.format.open_memmap(
        CHECKPOINT_FILE,
        mode="w+",
        dtype=np.float32,
        shape=(n, embedding_dim)
    )

    completed = 0

    checkpoint_meta = {
        "model_name": MODEL_NAME,
        "chunk_fingerprint": chunk_fingerprint,
        "total_chunks": n,
        "embedding_dimension": embedding_dim,
        "completed_chunks": 0,
        "batch_size": BATCH_SIZE
    }

    with open(CHECKPOINT_META_FILE, "w", encoding="utf-8") as f:
        json.dump(checkpoint_meta, f, indent=2)

# --------------------------------------------------
# GENERATE EMBEDDINGS
# --------------------------------------------------
for start in range(completed, n, BATCH_SIZE):
    end = min(start + BATCH_SIZE, n)

    batch_vectors = model.encode(
        texts[start:end],
        batch_size=BATCH_SIZE,
        show_progress_bar=False,
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    embeddings[start:end] = batch_vectors.astype(np.float32)
    embeddings.flush()

    checkpoint_meta["completed_chunks"] = end

    with open(CHECKPOINT_META_FILE, "w", encoding="utf-8") as f:
        json.dump(checkpoint_meta, f, indent=2)

    if end % 1024 == 0 or end == n:
        print(f"Embedded {end:,}/{n:,} chunks")

# --------------------------------------------------
# FINAL VALIDATION
# --------------------------------------------------
if not np.isfinite(embeddings).all():
    raise ValueError("Embeddings contain NaN or infinite values.")

if checkpoint_meta["completed_chunks"] != n:
    raise RuntimeError("Embedding process is incomplete.")

# Copy the completed matrix to the final output.
np.save(EMBEDDINGS_FILE, np.asarray(embeddings))

# Keep metadata in exactly the same row order as the vectors.
metadata = chunks[required_columns].copy()
metadata.to_csv(METADATA_FILE, index=False)

# Save reproducibility manifest.
manifest = {
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "model_name": MODEL_NAME,
    "embedding_dimension": embedding_dim,
    "normalized_embeddings": True,
    "dtype": "float32",
    "total_chunks": n,
    "chunk_fingerprint_sha256": chunk_fingerprint,
    "batch_size": BATCH_SIZE,
    "device": "cuda" if __import__("torch").cuda.is_available() else "cpu",
    "embeddings_file": str(EMBEDDINGS_FILE),
    "metadata_file": str(METADATA_FILE),
    "source_chunks_file": str(CHUNKS_FILE)
}

with open(MANIFEST_FILE, "w", encoding="utf-8") as f:
    json.dump(manifest, f, indent=2)

print("\n" + "=" * 60)
print("PHASE 5.5 — DENSE EMBEDDINGS COMPLETE")
print("=" * 60)
print("Embedding shape:", np.load(EMBEDDINGS_FILE, mmap_mode="r").shape)
print("Metadata rows:", len(metadata))
print("Embedding dimensions:", embedding_dim)
print("Final embeddings:", EMBEDDINGS_FILE)
print("Metadata:", METADATA_FILE)
print("Manifest:", MANIFEST_FILE)
```

```python
import re
import json
import pandas as pd
from pathlib import Path
from datetime import datetime, timezone

# Paths
SOURCE_DIR = Path("/kaggle/working/AILQA/Dataset_For_Context")
OUTPUT_DIR = Path("/kaggle/working/indian_legal_rag/04_Legal_Corpus")

CHUNKS_FILE = OUTPUT_DIR / "phase5_legal_chunks_v2.csv"
MANIFEST_FILE = OUTPUT_DIR / "phase5_chunk_manifest_v2.json"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# Prevent accidental overwriting
if CHUNKS_FILE.exists() or MANIFEST_FILE.exists():
    raise FileExistsError(
        "Chunk output already exists. Inspect it before rerunning."
    )

TARGET_CHARS = 1600
MAX_CHARS = 1800
OVERLAP_CHARS = 200

def clean_text(text):
    text = text.replace("\r\n", "\n").replace("\r", "\n")
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n{3,}", "\n\n", text)
    return text.strip()

def split_into_units(text):
    paragraphs = [
        p.strip()
        for p in re.split(r"\n\s*\n", text)
        if p.strip()
    ]

    units = []

    for paragraph in paragraphs:
        if len(paragraph) <= TARGET_CHARS:
            units.append(paragraph)
            continue

        sentences = re.split(
            r"(?<=[.!?])\s+(?=[A-Z0-9(\"'])",
            paragraph
        )

        for sentence in sentences:
            sentence = sentence.strip()
            if not sentence:
                continue

            if len(sentence) <= TARGET_CHARS:
                units.append(sentence)
            else:
                words = sentence.split()
                current = []

                for word in words:
                    candidate = " ".join(current + [word])

                    if len(candidate) > TARGET_CHARS and current:
                        units.append(" ".join(current))
                        current = [word]
                    else:
                        current.append(word)

                if current:
                    units.append(" ".join(current))

    return units

def build_chunks(text):
    units = split_into_units(text)
    chunks = []
    current = ""

    for unit in units:
        candidate = f"{current}\n{unit}".strip() if current else unit

        if len(candidate) <= MAX_CHARS:
            current = candidate
            continue

        if current:
            chunks.append(current)

        overlap = current[-OVERLAP_CHARS:].strip() if current else ""

        if overlap and " " in overlap:
            overlap = overlap.split(" ", 1)[1]

        if overlap and len(overlap) + len(unit) <= MAX_CHARS:
            current = overlap + "\n" + unit
        else:
            current = unit

        if len(current) > MAX_CHARS:
            words = current.split()
            current = ""

            for word in words:
                candidate = f"{current} {word}".strip()

                if len(candidate) > MAX_CHARS and current:
                    chunks.append(current)
                    current = word
                else:
                    current = candidate

    if current:
        chunks.append(current)

    return chunks

# Process documents in deterministic filename order
source_files = sorted(SOURCE_DIR.glob("*.txt"))

if len(source_files) != 7221:
    raise ValueError(f"Expected 7,221 source files; found {len(source_files)}.")

records = []
failed_files = []

for file_path in source_files:
    try:
        text = clean_text(
            file_path.read_text(encoding="utf-8", errors="replace")
        )

        if not text:
            failed_files.append({
                "source_file": file_path.name,
                "error": "Empty source document"
            })
            continue

        document_id = file_path.stem

        for index, chunk in enumerate(build_chunks(text)):
            records.append({
                "chunk_id": f"{document_id}_v2_chunk_{index:04d}",
                "document_id": document_id,
                "source_file": file_path.name,
                "chunk_index": index,
                "text": chunk,
                "char_count": len(chunk)
            })

    except Exception as exc:
        failed_files.append({
            "source_file": file_path.name,
            "error": str(exc)
        })

df = pd.DataFrame(records)

if df.empty:
    raise ValueError("No chunks were generated.")

assert df["chunk_id"].is_unique
assert df["text"].notna().all()
assert df["text"].str.strip().ne("").all()
assert df["text"].str.len().max() <= MAX_CHARS

df.to_csv(CHUNKS_FILE, index=False, encoding="utf-8")

manifest = {
    "version": "v2",
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "source_directory": str(SOURCE_DIR),
    "source_files_discovered": len(source_files),
    "documents_with_chunks": int(df["document_id"].nunique()),
    "total_chunks": len(df),
    "target_chars": TARGET_CHARS,
    "maximum_chars": MAX_CHARS,
    "overlap_chars": OVERLAP_CHARS,
    "failed_files": failed_files,
    "output_csv": str(CHUNKS_FILE),
    "chunking_method": "Paragraph-first, sentence-aware splitting with whitespace fallback"
}

MANIFEST_FILE.write_text(
    json.dumps(manifest, indent=2, ensure_ascii=False),
    encoding="utf-8"
)

print("=" * 60)
print("PHASE 5.3 — CHUNKS RECREATED")
print("=" * 60)
print("Source files:", len(source_files))
print("Documents represented:", df["document_id"].nunique())
print("Total chunks:", len(df))
print("Average chunk length:", round(df["char_count"].mean(), 2))
print("Maximum chunk length:", df["char_count"].max())
print("Failed files:", len(failed_files))
print("Unique chunk IDs:", df["chunk_id"].nunique())
print("\nChunks saved:", CHUNKS_FILE)
print("Manifest saved:", MANIFEST_FILE)
```

```python
import json
import hashlib
import numpy as np
import pandas as pd
from pathlib import Path
from datetime import datetime, timezone
from sentence_transformers import SentenceTransformer

BASE_DIR = Path("/kaggle/working/indian_legal_rag/04_Legal_Corpus")

CHUNKS_FILE = BASE_DIR / "phase5_legal_chunks_v2.csv"
EMBEDDINGS_FILE = BASE_DIR / "phase5_dense_embeddings_v1.npy"
METADATA_FILE = BASE_DIR / "phase5_embedding_metadata_v1.csv"
MANIFEST_FILE = BASE_DIR / "phase5_embedding_manifest_v1.json"
CHECKPOINT_FILE = BASE_DIR / "phase5_embedding_checkpoint_v1.npy"
CHECKPOINT_META_FILE = BASE_DIR / "phase5_embedding_checkpoint_meta_v1.json"

MODEL_NAME = "sentence-transformers/all-MiniLM-L6-v2"
BATCH_SIZE = 64

# Safety checks
if not CHUNKS_FILE.exists():
    raise FileNotFoundError(f"Missing chunks: {CHUNKS_FILE}")

if any(p.exists() for p in [EMBEDDINGS_FILE, METADATA_FILE, MANIFEST_FILE]):
    raise FileExistsError("A final output already exists. Inspect before rerunning.")

if CHECKPOINT_FILE.exists() != CHECKPOINT_META_FILE.exists():
    raise RuntimeError("Incomplete checkpoint pair. Inspect before proceeding.")

chunks = pd.read_csv(CHUNKS_FILE)

if len(chunks) != 115840:
    raise ValueError(f"Expected 115840 chunks, found {len(chunks)}.")

if chunks["chunk_id"].duplicated().any():
    raise ValueError("Duplicate chunk IDs found.")

texts = chunks["text"].astype(str).tolist()
fingerprint = hashlib.sha256(
    "\n".join(chunks["chunk_id"].tolist()).encode("utf-8")
).hexdigest()

print("Chunks:", len(chunks))
print("Fingerprint:", fingerprint)

# Load model
model = SentenceTransformer(MODEL_NAME)
dimension = model.get_embedding_dimension()

if dimension != 384:
    raise ValueError(f"Expected 384 dimensions, got {dimension}")

# Create or resume checkpoint
if CHECKPOINT_FILE.exists():
    with open(CHECKPOINT_META_FILE, "r", encoding="utf-8") as f:
        checkpoint_meta = json.load(f)

    if (
        checkpoint_meta["chunk_fingerprint"] != fingerprint
        or checkpoint_meta["model_name"] != MODEL_NAME
        or checkpoint_meta["total_chunks"] != len(chunks)
    ):
        raise RuntimeError("Checkpoint does not match current chunks/model.")

    embeddings = np.load(CHECKPOINT_FILE, mmap_mode="r+")
    completed = checkpoint_meta["completed_chunks"]

    if embeddings.shape != (len(chunks), dimension):
        raise RuntimeError("Checkpoint shape is invalid.")

    print("Resuming at:", completed)

else:
    embeddings = np.lib.format.open_memmap(
        CHECKPOINT_FILE,
        mode="w+",
        dtype=np.float32,
        shape=(len(chunks), dimension)
    )
    completed = 0

    checkpoint_meta = {
        "model_name": MODEL_NAME,
        "chunk_fingerprint": fingerprint,
        "total_chunks": len(chunks),
        "embedding_dimension": dimension,
        "completed_chunks": 0,
        "batch_size": BATCH_SIZE
    }

    with open(CHECKPOINT_META_FILE, "w", encoding="utf-8") as f:
        json.dump(checkpoint_meta, f, indent=2)

# Generate embeddings
for start in range(completed, len(chunks), BATCH_SIZE):
    end = min(start + BATCH_SIZE, len(chunks))

    vectors = model.encode(
        texts[start:end],
        batch_size=BATCH_SIZE,
        show_progress_bar=False,
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    embeddings[start:end] = vectors.astype(np.float32)
    embeddings.flush()

    checkpoint_meta["completed_chunks"] = end

    with open(CHECKPOINT_META_FILE, "w", encoding="utf-8") as f:
        json.dump(checkpoint_meta, f, indent=2)

    if end % 1024 == 0 or end == len(chunks):
        print(f"Embedded {end:,}/{len(chunks):,}")

if checkpoint_meta["completed_chunks"] != len(chunks):
    raise RuntimeError("Embedding generation is incomplete.")

if not np.isfinite(embeddings).all():
    raise ValueError("Embeddings contain invalid numeric values.")

# Save final output
np.save(EMBEDDINGS_FILE, np.asarray(embeddings))

metadata = chunks[
    ["chunk_id", "document_id", "source_file", "chunk_index"]
].copy()
metadata.to_csv(METADATA_FILE, index=False)

manifest = {
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "model_name": MODEL_NAME,
    "embedding_dimension": dimension,
    "normalized_embeddings": True,
    "dtype": "float32",
    "total_chunks": len(chunks),
    "chunk_fingerprint_sha256": fingerprint,
    "batch_size": BATCH_SIZE,
    "embeddings_file": str(EMBEDDINGS_FILE),
    "metadata_file": str(METADATA_FILE)
}

with open(MANIFEST_FILE, "w", encoding="utf-8") as f:
    json.dump(manifest, f, indent=2)

print("\nPHASE 5.5 RECREATED SUCCESSFULLY")
print("Embedding shape:", np.load(EMBEDDINGS_FILE, mmap_mode="r").shape)
print("Metadata rows:", len(metadata))
print("Embeddings:", EMBEDDINGS_FILE)
print("Metadata:", METADATA_FILE)
print("Manifest:", MANIFEST_FILE)
```

```python
import json
import hashlib
import importlib.util
import numpy as np
import pandas as pd
from pathlib import Path

BASE_DIR = Path(
    "/kaggle/working/indian_legal_rag/04_Legal_Corpus"
)

CHUNKS_FILE = BASE_DIR / "phase5_legal_chunks_v2.csv"
EMBEDDINGS_FILE = BASE_DIR / "phase5_dense_embeddings_v1.npy"
METADATA_FILE = BASE_DIR / "phase5_embedding_metadata_v1.csv"
MANIFEST_FILE = BASE_DIR / "phase5_embedding_manifest_v1.json"

# 1. Check files
for path in [
    CHUNKS_FILE, EMBEDDINGS_FILE, METADATA_FILE, MANIFEST_FILE
]:
    if not path.exists():
        raise FileNotFoundError(f"Missing file: {path}")

# 2. Load files
chunks = pd.read_csv(CHUNKS_FILE)
metadata = pd.read_csv(METADATA_FILE)
embeddings = np.load(EMBEDDINGS_FILE, mmap_mode="r")

with open(MANIFEST_FILE, "r", encoding="utf-8") as f:
    manifest = json.load(f)

print("=" * 60)
print("PHASE 5.6 — EMBEDDING VALIDATION")
print("=" * 60)

print("Chunk records:", len(chunks))
print("Metadata records:", len(metadata))
print("Embedding shape:", embeddings.shape)
print("Embedding dtype:", embeddings.dtype)

# 3. Validate counts, IDs, and ordering
assert len(chunks) == len(metadata) == embeddings.shape[0]
assert embeddings.shape[1] == 384
assert chunks["chunk_id"].is_unique
assert metadata["chunk_id"].is_unique
assert chunks["chunk_id"].tolist() == metadata["chunk_id"].tolist()

print("\nRecord counts and ordering: PASS")

# 4. Validate numeric values and normalized vector lengths
norm_min = float("inf")
norm_max = 0.0

for start in range(0, len(embeddings), 4096):
    batch = np.asarray(embeddings[start:start + 4096])

    assert np.isfinite(batch).all(), (
        f"Invalid numeric values in batch starting at {start}"
    )

    norms = np.linalg.norm(batch, axis=1)
    norm_min = min(norm_min, float(norms.min()))
    norm_max = max(norm_max, float(norms.max()))

print("All embedding values finite: PASS")
print("Minimum vector norm:", round(norm_min, 5))
print("Maximum vector norm:", round(norm_max, 5))

assert abs(norm_min - 1.0) < 0.01
assert abs(norm_max - 1.0) < 0.01

print("Normalized embeddings: PASS")

# 5. Verify chunk fingerprint
fingerprint = hashlib.sha256(
    "\n".join(chunks["chunk_id"].tolist()).encode("utf-8")
).hexdigest()

assert fingerprint == manifest["chunk_fingerprint_sha256"]

print("Chunk fingerprint matches manifest: PASS")

# 6. Check FAISS availability
faiss_available = importlib.util.find_spec("faiss") is not None
print("\nFAISS installed:", faiss_available)

print("\nRESULT: ALL EMBEDDING CHECKS PASSED")
```

```python
from pathlib import Path
import zipfile
import json

BASE_DIR = Path(
    "/kaggle/working/indian_legal_rag/04_Legal_Corpus"
)

BACKUP_FILE = Path("/kaggle/working/phase5_backup_v1.zip")

files_to_backup = [
    "phase5_legal_chunks_v2.csv",
    "phase5_chunk_manifest_v2.json",
    "phase5_dense_embeddings_v1.npy",
    "phase5_embedding_metadata_v1.csv",
    "phase5_embedding_manifest_v1.json",
]

# Verify every file exists before creating the archive
missing = [name for name in files_to_backup if not (BASE_DIR / name).exists()]

if missing:
    raise FileNotFoundError(f"Missing files: {missing}")

if BACKUP_FILE.exists():
    raise FileExistsError(
        f"Backup already exists: {BACKUP_FILE}. Inspect it before replacing."
    )

# Create backup
with zipfile.ZipFile(
    BACKUP_FILE, "w", compression=zipfile.ZIP_DEFLATED
) as archive:
    for name in files_to_backup:
        archive.write(BASE_DIR / name, arcname=name)
        print("Backed up:", name)

print("\nBackup created:", BACKUP_FILE)
print("Backup size:", round(BACKUP_FILE.stat().st_size / (1024**2), 2), "MB")

# Verify archive contents
with zipfile.ZipFile(BACKUP_FILE, "r") as archive:
    print("\nArchive integrity:", archive.testzip() or "PASS")
    print("Files in archive:", len(archive.namelist()))
```

```python
import faiss
import numpy as np
import os
import json
from datetime import datetime, timezone

BASE_DIR = "/kaggle/working/indian_legal_rag/04_Legal_Corpus"

EMBEDDINGS_PATH = os.path.join(
    BASE_DIR, "phase5_dense_embeddings_v1.npy"
)

INDEX_PATH = os.path.join(
    BASE_DIR, "phase5_faiss_index_v1.index"
)

MANIFEST_PATH = os.path.join(
    BASE_DIR, "phase5_faiss_index_manifest_v1.json"
)

# Prevent accidentally overwriting an existing index
if os.path.exists(INDEX_PATH):
    raise FileExistsError(
        f"Index already exists: {INDEX_PATH}. "
        "Inspect it before rebuilding."
    )

# Load existing embeddings
embeddings = np.load(EMBEDDINGS_PATH)

# Ensure FAISS receives contiguous float32 vectors
embeddings = np.ascontiguousarray(embeddings, dtype=np.float32)

num_vectors, dimension = embeddings.shape

# Inner product on normalized vectors is equivalent to cosine similarity
index = faiss.IndexFlatIP(dimension)

print(f"Building index for {num_vectors:,} vectors...")
index.add(embeddings)

# Validate index size
assert index.ntotal == num_vectors, "Index size mismatch!"

# Save index
faiss.write_index(index, INDEX_PATH)

# Save index manifest
manifest = {
    "index_type": "IndexFlatIP",
    "similarity_metric": "cosine_similarity_for_normalized_vectors",
    "embedding_model": "sentence-transformers/all-MiniLM-L6-v2",
    "embedding_dimension": dimension,
    "total_vectors": int(index.ntotal),
    "embedding_dtype": str(embeddings.dtype),
    "source_embeddings": os.path.basename(EMBEDDINGS_PATH),
    "index_file": os.path.basename(INDEX_PATH),
    "created_at_utc": datetime.now(timezone.utc).isoformat()
}

with open(MANIFEST_PATH, "w", encoding="utf-8") as f:
    json.dump(manifest, f, indent=2)

print("\nFAISS index built successfully!")
print("Total vectors:", f"{index.ntotal:,}")
print("Dimension:", dimension)
print("Index type:", type(index).__name__)
print("Index saved:", INDEX_PATH)
print("Manifest saved:", MANIFEST_PATH)
print("Index size (MB):", round(os.path.getsize(INDEX_PATH) / (1024**2), 2))
```

```python
import faiss
import numpy as np
import pandas as pd
import os

BASE_DIR = "/kaggle/working/indian_legal_rag/04_Legal_Corpus"

INDEX_PATH = os.path.join(
    BASE_DIR, "phase5_faiss_index_v1.index"
)
EMBEDDINGS_PATH = os.path.join(
    BASE_DIR, "phase5_dense_embeddings_v1.npy"
)
METADATA_PATH = os.path.join(
    BASE_DIR, "phase5_embedding_metadata_v1.csv"
)
CHUNKS_PATH = os.path.join(
    BASE_DIR, "phase5_legal_chunks_v2.csv"
)

# Load existing index, embeddings, metadata, and chunk text
index = faiss.read_index(INDEX_PATH)
embeddings = np.load(EMBEDDINGS_PATH, mmap_mode="r")
metadata = pd.read_csv(METADATA_PATH)
chunks = pd.read_csv(CHUNKS_PATH)

# Attach text to metadata using chunk_id
metadata = metadata.merge(
    chunks[["chunk_id", "text"]],
    on="chunk_id",
    how="left",
    validate="one_to_one"
)

# Check mapping
assert len(metadata) == index.ntotal
assert metadata["text"].notna().all(), "Some chunk texts are missing!"
assert metadata["chunk_id"].is_unique, "Duplicate chunk IDs found!"

print("Text mapping validated successfully.")
print("Indexed chunks:", len(metadata))
print("Missing text:", metadata["text"].isna().sum())

# Search using an existing embedding for index verification
query_vector = np.asarray(embeddings[0:1], dtype=np.float32)
scores, indices = index.search(query_vector, 5)

print("\nTop 5 retrieved chunks:")

for rank, (score, idx) in enumerate(
    zip(scores[0], indices[0]), start=1
):
    row = metadata.iloc[int(idx)]

    print(f"\n--- Rank {rank} | Similarity: {score:.4f} ---")
    print("Chunk ID:", row["chunk_id"])
    print("Source file:", row["source_file"])
    print("Text:", str(row["text"])[:500])
```

```python
import faiss
import numpy as np
import pandas as pd
import os

BASE_DIR = "/kaggle/working/indian_legal_rag/04_Legal_Corpus"

# Load the existing FAISS index and chunk data
index = faiss.read_index(
    os.path.join(BASE_DIR, "phase5_faiss_index_v1.index")
)

chunks = pd.read_csv(
    os.path.join(BASE_DIR, "phase5_legal_chunks_v2.csv")
)

metadata = pd.read_csv(
    os.path.join(BASE_DIR, "phase5_embedding_metadata_v1.csv")
)

# Connect metadata to the legal text
metadata = metadata.merge(
    chunks[["chunk_id", "text"]],
    on="chunk_id",
    how="left",
    validate="one_to_one"
)

# Test question
query = "What factors does the court consider when granting bail in a criminal case?"

# Generate the query embedding
query_vector = model.encode(
    [query],
    convert_to_numpy=True,
    normalize_embeddings=True
).astype(np.float32)

# Search the FAISS index
scores, indices = index.search(query_vector, 5)

print("QUESTION:", query)
print("\nTOP 5 RETRIEVED LEGAL PASSAGES")

for rank, (score, idx) in enumerate(
    zip(scores[0], indices[0]), start=1
):
    row = metadata.iloc[int(idx)]

    print(f"\n{'=' * 70}")
    print(f"Rank: {rank} | Similarity: {score:.4f}")
    print("Document:", row["source_file"])
    print("Chunk ID:", row["chunk_id"])
    print("Passage:")
    print(str(row["text"])[:1000])
```

```python
import os
import json
import re
import pandas as pd
from rank_bm25 import BM25Okapi
from datetime import datetime, timezone

BASE_DIR = "/kaggle/working/indian_legal_rag/04_Legal_Corpus"

CHUNKS_PATH = os.path.join(
    BASE_DIR, "phase5_legal_chunks_v2.csv"
)

INDEX_PATH = os.path.join(
    BASE_DIR, "phase6_bm25_corpus_tokens_v1.json"
)

MANIFEST_PATH = os.path.join(
    BASE_DIR, "phase6_bm25_manifest_v1.json"
)

# Avoid overwriting existing artifacts
if os.path.exists(INDEX_PATH) or os.path.exists(MANIFEST_PATH):
    raise FileExistsError(
        "A BM25 artifact already exists. Inspect it before rebuilding."
    )

chunks = pd.read_csv(CHUNKS_PATH)

assert len(chunks) == 115840, (
    f"Unexpected chunk count: {len(chunks)}"
)
assert chunks["chunk_id"].is_unique, "Duplicate chunk IDs found."
assert chunks["text"].notna().all(), "Missing chunk text found."

def tokenize(text):
    # Lowercase and retain legal section numbers and alphanumeric terms
    return re.findall(r"[a-z0-9]+", str(text).lower())

print(f"Tokenizing {len(chunks):,} legal chunks...")

corpus_tokens = [
    tokenize(text) for text in chunks["text"]
]

assert len(corpus_tokens) == len(chunks)
assert all(corpus_tokens), "At least one chunk has no searchable tokens."

# Save tokenized corpus for reproducibility
with open(INDEX_PATH, "w", encoding="utf-8") as f:
    json.dump(corpus_tokens, f, separators=(",", ":"))

manifest = {
    "retrieval_method": "BM25Okapi",
    "tokenization": "lowercase alphanumeric tokens using regex [a-z0-9]+",
    "total_chunks": len(chunks),
    "chunk_file": os.path.basename(CHUNKS_PATH),
    "token_file": os.path.basename(INDEX_PATH),
    "created_at_utc": datetime.now(timezone.utc).isoformat()
}

with open(MANIFEST_PATH, "w", encoding="utf-8") as f:
    json.dump(manifest, f, indent=2)

print("\nBM25 corpus preparation complete.")
print("Chunks:", len(corpus_tokens))
print("Total tokens:", sum(map(len, corpus_tokens)))
print("Token file:", INDEX_PATH)
print("Manifest:", MANIFEST_PATH)
```

```python
import numpy as np
import pandas as pd
import faiss

# Reuse the existing index, model, BM25 index, and tokenizer.
# Load chunks and metadata if they are not already available.

BASE_DIR = "/kaggle/working/indian_legal_rag/04_Legal_Corpus"

chunks = pd.read_csv(
    f"{BASE_DIR}/phase5_legal_chunks_v2.csv"
)

metadata = pd.read_csv(
    f"{BASE_DIR}/phase5_embedding_metadata_v1.csv"
)

metadata = metadata.merge(
    chunks[["chunk_id", "text"]],
    on="chunk_id",
    how="left",
    validate="one_to_one"
)

assert len(metadata) == index.ntotal == len(chunks)

def hybrid_retrieve(query, dense_k=100, sparse_k=100, top_k=5, rrf_k=60):
    # Dense retrieval
    query_vector = model.encode(
        [query],
        convert_to_numpy=True,
        normalize_embeddings=True
    ).astype(np.float32)

    dense_scores, dense_indices = index.search(
        query_vector, dense_k
    )

    # Sparse retrieval
    query_tokens = tokenize(query)
    sparse_scores = bm25_index.get_scores(query_tokens)

    sparse_indices = np.argsort(sparse_scores)[::-1][:sparse_k]

    # Reciprocal Rank Fusion
    fused_scores = {}

    for rank, idx in enumerate(dense_indices[0], start=1):
        idx = int(idx)
        fused_scores[idx] = fused_scores.get(idx, 0.0) + (
            1.0 / (rrf_k + rank)
        )

    for rank, idx in enumerate(sparse_indices, start=1):
        idx = int(idx)
        fused_scores[idx] = fused_scores.get(idx, 0.0) + (
            1.0 / (rrf_k + rank)
        )

    ranked_indices = sorted(
        fused_scores,
        key=fused_scores.get,
        reverse=True
    )[:top_k]

    results = []

    for rank, idx in enumerate(ranked_indices, start=1):
        row = metadata.iloc[idx]

        results.append({
            "rank": rank,
            "rrf_score": fused_scores[idx],
            "chunk_id": row["chunk_id"],
            "source_file": row["source_file"],
            "text": row["text"]
        })

    return pd.DataFrame(results)


query = (
    "What factors does the court consider when granting bail "
    "in a criminal case?"
)

hybrid_results = hybrid_retrieve(query)

print("QUESTION:", query)
print("\nHYBRID RETRIEVAL — TOP 5 RESULTS")

for _, row in hybrid_results.iterrows():
    print("\n" + "=" * 75)
    print(f"Rank: {row['rank']} | RRF score: {row['rrf_score']:.5f}")
    print("Source:", row["source_file"])
    print("Chunk ID:", row["chunk_id"])
    print("Passage:", str(row["text"])[:700].replace("\n", " "))
```

```python
import numpy as np
import pandas as pd
import os
from collections import Counter

BASE_DIR = "/kaggle/working/indian_legal_rag/04_Legal_Corpus"

# Load chunks if they are not currently available
chunks_path = os.path.join(BASE_DIR, "phase5_legal_chunks_v2.csv")
chunks = pd.read_csv(chunks_path)

# Confirm existing index and BM25 data match the corpus
assert len(chunks) == index.ntotal == len(corpus_tokens)
assert chunks["chunk_id"].equals(metadata["chunk_id"]), (
    "Chunk ordering mismatch. Stop and investigate."
)

query = "What factors does the court consider when granting bail in a criminal case?"

# 1. Dense retrieval
query_vector = model.encode(
    [query],
    convert_to_numpy=True,
    normalize_embeddings=True
).astype(np.float32)

dense_scores, dense_indices = index.search(query_vector, 20)

# 2. BM25 retrieval
query_tokens = tokenize(query)
sparse_scores = bm25_index.get_scores(query_tokens)
sparse_indices = np.argsort(sparse_scores)[::-1][:20]

# 3. Reciprocal Rank Fusion
RRF_K = 60
rrf_scores = {}

for rank, idx in enumerate(dense_indices[0], start=1):
    idx = int(idx)
    rrf_scores[idx] = rrf_scores.get(idx, 0) + 1 / (RRF_K + rank)

for rank, idx in enumerate(sparse_indices, start=1):
    idx = int(idx)
    rrf_scores[idx] = rrf_scores.get(idx, 0) + 1 / (RRF_K + rank)

hybrid_indices = sorted(
    rrf_scores,
    key=rrf_scores.get,
    reverse=True
)[:10]

# 4. Inspect top 10 results
results = chunks.iloc[hybrid_indices].copy()

print("TOP 10 HYBRID RESULTS")
print("-" * 60)

for rank, (_, row) in enumerate(results.iterrows(), start=1):
    print(f"{rank}. {row['source_file']} | {row['chunk_id']}")

# 5. Summarize document diversity
source_counts = Counter(results["source_file"])

print("\nDIVERSITY SUMMARY")
print("-" * 60)
print("Results examined:", len(results))
print("Unique source documents:", len(source_counts))
print(
    "Documents represented more than once:",
    sum(count > 1 for count in source_counts.values())
)

print("\nRepeated source documents:")
for source, count in source_counts.items():
    if count > 1:
        print(f"- {source}: {count} results")
```

```python
from sentence_transformers import CrossEncoder

RERANKER_NAME = "cross-encoder/ms-marco-MiniLM-L-6-v2"

reranker = CrossEncoder(
    RERANKER_NAME,
    max_length=512,
    device="cpu"
)

print("Cross-encoder loaded successfully!")
print("Model:", RERANKER_NAME)
```

```python
import numpy as np
import pandas as pd
import os

BASE_DIR = "/kaggle/working/indian_legal_rag/04_Legal_Corpus"

chunks = pd.read_csv(
    os.path.join(BASE_DIR, "phase5_legal_chunks_v2.csv")
)

# Confirm that all indexes refer to the same chunk order
assert len(chunks) == index.ntotal == len(corpus_tokens)
assert chunks["chunk_id"].equals(metadata["chunk_id"]), (
    "Chunk ordering mismatch."
)

query = "What factors does the court consider when granting bail in a criminal case?"

# 1. Dense retrieval
query_vector = model.encode(
    [query],
    convert_to_numpy=True,
    normalize_embeddings=True
).astype(np.float32)

_, dense_indices = index.search(query_vector, 20)

# 2. BM25 retrieval
query_tokens = tokenize(query)
sparse_scores = bm25_index.get_scores(query_tokens)
sparse_indices = np.argsort(sparse_scores)[::-1][:20]

# 3. Reciprocal Rank Fusion
RRF_K = 60
rrf_scores = {}

for rank, idx in enumerate(dense_indices[0], start=1):
    idx = int(idx)
    rrf_scores[idx] = rrf_scores.get(idx, 0.0) + 1 / (RRF_K + rank)

for rank, idx in enumerate(sparse_indices, start=1):
    idx = int(idx)
    rrf_scores[idx] = rrf_scores.get(idx, 0.0) + 1 / (RRF_K + rank)

hybrid_candidates = sorted(
    rrf_scores,
    key=rrf_scores.get,
    reverse=True
)[:20]

# 4. Prepare question-passage pairs for the cross-encoder
candidate_texts = [
    str(chunks.iloc[idx]["text"])
    for idx in hybrid_candidates
]

pairs = [(query, text) for text in candidate_texts]

print("Reranking", len(pairs), "hybrid candidates...")

reranker_scores = reranker.predict(
    pairs,
    batch_size=8,
    show_progress_bar=True
)

# 5. Rank candidates by cross-encoder score
reranked = sorted(
    zip(hybrid_candidates, reranker_scores),
    key=lambda item: float(item[1]),
    reverse=True
)

print("\nQUESTION:", query)
print("\nTOP 5 CROSS-ENCODER RERANKED RESULTS")

for rank, (idx, score) in enumerate(reranked[:5], start=1):
    row = chunks.iloc[int(idx)]

    print("\n" + "=" * 75)
    print(f"Rank: {rank} | Reranker score: {float(score):.4f}")
    print("Source:", row["source_file"])
    print("Chunk ID:", row["chunk_id"])
    print("Passage:", str(row["text"])[:900].replace("\n", " "))
```

```python
import numpy as np
import pandas as pd
import os

BASE_DIR = "/kaggle/working/indian_legal_rag/04_Legal_Corpus"

chunks = pd.read_csv(
    os.path.join(BASE_DIR, "phase5_legal_chunks_v2.csv")
)

test_questions = [
    "What factors does the court consider when granting bail in a criminal case?",
    "What are the rights of an accused person during a criminal trial?",
    "What is the legal procedure for filing an appeal against a criminal conviction?",
    "What conditions are required to establish criminal liability?",
    "What powers does the Supreme Court exercise in criminal appeals?"
]

def retrieve_and_rerank(question, top_k=5, candidate_k=20):
    # Dense retrieval
    query_vector = model.encode(
        [question],
        convert_to_numpy=True,
        normalize_embeddings=True
    ).astype(np.float32)

    _, dense_indices = index.search(query_vector, candidate_k)

    # Sparse retrieval
    query_tokens = tokenize(question)
    sparse_scores = bm25_index.get_scores(query_tokens)
    sparse_indices = np.argsort(sparse_scores)[::-1][:candidate_k]

    # Reciprocal Rank Fusion
    rrf_scores = {}

    for rank, idx in enumerate(dense_indices[0], start=1):
        idx = int(idx)
        rrf_scores[idx] = rrf_scores.get(idx, 0.0) + 1 / (60 + rank)

    for rank, idx in enumerate(sparse_indices, start=1):
        idx = int(idx)
        rrf_scores[idx] = rrf_scores.get(idx, 0.0) + 1 / (60 + rank)

    candidates = sorted(
        rrf_scores,
        key=rrf_scores.get,
        reverse=True
    )[:candidate_k]

    # Cross-encoder reranking
    pairs = [
        (question, str(chunks.iloc[idx]["text"]))
        for idx in candidates
    ]

    scores = reranker.predict(pairs, batch_size=8)

    reranked = sorted(
        zip(candidates, scores),
        key=lambda item: float(item[1]),
        reverse=True
    )

    return [
        {
            "question": question,
            "rank": rank,
            "chunk_id": chunks.iloc[idx]["chunk_id"],
            "source_file": chunks.iloc[idx]["source_file"],
            "reranker_score": float(score),
            "text": str(chunks.iloc[idx]["text"])[:700]
        }
        for rank, (idx, score) in enumerate(reranked[:top_k], start=1)
    ]

all_results = []

for question_number, question in enumerate(test_questions, start=1):
    print(f"\nTesting question {question_number}/{len(test_questions)}")
    results = retrieve_and_rerank(question)
    all_results.extend(results)

    print("Question:", question)

    for result in results[:3]:
        print(
            f"  Rank {result['rank']} | "
            f"Score {result['reranker_score']:.4f} | "
            f"{result['source_file']}"
        )
        print(" ", result["text"][:250].replace("\n", " "))

results_df = pd.DataFrame(all_results)

print("\nTesting complete.")
print("Questions tested:", len(test_questions))
print("Results recorded:", len(results_df))
```

```python
from pathlib import Path
import pandas as pd
import re

project = Path("/kaggle/working/indian_legal_rag")
source_dir = Path("/kaggle/working/AILQA/Dataset_For_Context")
recovery_dir = Path(
    "/kaggle/input/datasets/vaishnavidevi2006/indian-legal-rag-recovery"
)

gold_path = recovery_dir / "phase3_gold_evidence_pilot_RECONSTRUCTED.csv"
benchmark_path = recovery_dir / "phase4_qa_benchmark_v1_RECONSTRUCTED.csv"
review_path = recovery_dir / "phase3_record_review_log_RECONSTRUCTED.csv"

gold = pd.read_csv(gold_path).fillna("")
benchmark = pd.read_csv(benchmark_path).fillna("")
review = pd.read_csv(review_path).fillna("")

expected_ids = {f"CRIM_{i:03d}" for i in range(1, 11)}

print("=== STRUCTURAL VALIDATION ===")
print("Gold evidence rows:", len(gold))
print("Benchmark rows:", len(benchmark))
print("Review log rows:", len(review))
print("Gold IDs complete:", set(gold["question_id"]) == expected_ids)
print("Benchmark IDs complete:", set(benchmark["question_id"]) == expected_ids)
print("Gold IDs unique:", gold["question_id"].is_unique)
print("Benchmark IDs unique:", benchmark["question_id"].is_unique)

print("\n=== SOURCE AND EVIDENCE VALIDATION ===")

def normalize(text):
    return re.sub(r"\s+", " ", str(text)).strip().lower()

results = []

for _, row in gold.iterrows():
    qid = row["question_id"]
    filename = str(row["evidence_document"]).strip()
    source_path = source_dir / filename

    if not source_path.is_file():
        results.append((qid, filename, "SOURCE_MISSING"))
        continue

    source_text = source_path.read_text(
        encoding="utf-8", errors="ignore"
    )

    passage = str(row["evidence_passage"]).strip()

    if not passage:
        status = "PASSAGE_EMPTY"
    elif normalize(passage) in normalize(source_text):
        status = "EXACT_NORMALIZED_MATCH"
    else:
        status = "PASSAGE_NEEDS_REVIEW"

    results.append((qid, filename, status))

for qid, filename, status in results:
    print(f"{qid} | {filename} | {status}")

print("\n=== REVIEW STATUS ===")
for column in [
    "evidence_passage_review",
    "legal_answer_review",
    "metadata_review",
]:
    if column in review.columns:
        print(f"\n{column}:")
        print(review[column].value_counts(dropna=False).to_dict())
```

```python
from pathlib import Path
import pandas as pd
import re

project = Path("/kaggle/working/indian_legal_rag")
source_dir = Path("/kaggle/working/AILQA/Dataset_For_Context")

files = {
    "Phase 3": project / "05_Gold_Evidence" /
        "phase3_gold_evidence_pilot_RECONSTRUCTED.csv",
    "Phase 4": project / "08_Evaluation" /
        "phase4_qa_benchmark_v1_RECONSTRUCTED.csv",
}

def normalize(text):
    return re.sub(r"\s+", " ", str(text)).strip().lower()

expected_ids = {f"CRIM_{i:03d}" for i in range(1, 11)}

for label, path in files.items():
    print(f"\n=== {label} VALIDATION ===")

    df = pd.read_csv(path).fillna("")

    print("Rows:", len(df))
    print("All 10 IDs present:", set(df["question_id"]) == expected_ids)
    print("Question IDs unique:", df["question_id"].is_unique)

    passed = 0

    for _, row in df.iterrows():
        source_path = source_dir / str(row["evidence_document"])

        if not source_path.is_file():
            print(f"FAIL | {row['question_id']} | Source missing")
            continue

        source_text = source_path.read_text(
            encoding="utf-8", errors="ignore"
        )

        passage = str(row["evidence_passage"]).strip()

        if passage and normalize(passage) in normalize(source_text):
            print(f"PASS | {row['question_id']}")
            passed += 1
        else:
            print(f"REVIEW | {row['question_id']}")

    print(f"Evidence matches: {passed}/10")
```

```python
from pathlib import Path
import pandas as pd
import hashlib
import json
from datetime import datetime, timezone

project = Path("/kaggle/working/indian_legal_rag")
gold_dir = project / "05_Gold_Evidence"
eval_dir = project / "08_Evaluation"

gold_dir.mkdir(parents=True, exist_ok=True)
eval_dir.mkdir(parents=True, exist_ok=True)

gold_path = gold_dir / "phase3_gold_evidence_pilot_RECONSTRUCTED.csv"
benchmark_path = eval_dir / "phase4_qa_benchmark_v1_RECONSTRUCTED.csv"
review_path = gold_dir / "phase3_record_review_log_RECONSTRUCTED.csv"
manifest_path = eval_dir / "phase4_benchmark_manifest_RECONSTRUCTED.json"

assert gold_path.exists(), f"Gold evidence file missing: {gold_path}"
assert benchmark_path.exists(), f"Benchmark file missing: {benchmark_path}"

gold = pd.read_csv(gold_path).fillna("")
benchmark = pd.read_csv(benchmark_path).fillna("")

# Recreate the review log if it is missing
if not review_path.exists():
    review = pd.DataFrame({
        "question_id": gold["question_id"],
        "case_name": gold["case_name"],
        "evidence_document": gold["evidence_document"],
        "source_file_exists_in_original_kaggle_workspace": True,
        "case_title_match": True,
        "evidence_passage_review": "PENDING_MANUAL_REVIEW",
        "legal_answer_review": "PENDING_LEGAL_REVIEW",
        "metadata_review": "PENDING_METADATA_REVIEW",
        "review_notes": (
            "Recovery log regenerated. Source and evidence checks passed; "
            "manual legal and metadata review remains pending."
        )
    })
    review.to_csv(review_path, index=False)

expected_ids = {f"CRIM_{i:03d}" for i in range(1, 11)}

for name, df in [("Phase 3", gold), ("Phase 4", benchmark)]:
    assert len(df) == 10, f"{name}: expected 10 records"
    assert set(df["question_id"]) == expected_ids, (
        f"{name}: question IDs incomplete"
    )
    assert df["question_id"].is_unique, f"{name}: duplicate IDs"

def sha256(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()

manifest = {
    "artifact_status": "RECONSTRUCTED_EVIDENCE_MATCHES_VERIFIED",
    "not_claimed_byte_identical_to_original_frozen_benchmark": True,
    "generated_utc": datetime.now(timezone.utc).isoformat(),
    "record_count": len(benchmark),
    "unique_question_ids": int(benchmark["question_id"].nunique()),
    "question_ids": sorted(benchmark["question_id"].tolist()),
    "validation": {
        "phase3_structure": "PASS",
        "phase4_structure": "PASS",
        "source_evidence_match": "PASS",
        "records_with_evidence_match": 10,
        "records_total": 10
    },
    "review_status": {
        "legal_answer_review": "PENDING",
        "metadata_review": "PENDING",
        "manual_evidence_span_review": "PENDING"
    },
    "files": {}
}

for path in [gold_path, review_path, benchmark_path]:
    manifest["files"][path.name] = {
        "sha256": sha256(path),
        "bytes": path.stat().st_size
    }

manifest_path.write_text(
    json.dumps(manifest, indent=2, ensure_ascii=False),
    encoding="utf-8"
)

print("Recovery manifest regenerated successfully.")
print("Review log:", review_path.exists())
print("Gold evidence records:", len(gold))
print("Benchmark records:", len(benchmark))
print("Unique question IDs:", benchmark["question_id"].nunique())
print("Evidence source checks previously passed: 10/10")
print("Legal-answer review: PENDING")
print("Metadata review: PENDING")
print("Manual evidence-span review: PENDING")
print("\nManifest:", manifest_path)
```

```python
import pandas as pd
import json
import hashlib
from pathlib import Path

project = Path("/kaggle/working/indian_legal_rag")

gold_path = project / "05_Gold_Evidence/phase3_gold_evidence_pilot_RECONSTRUCTED.csv"
benchmark_path = project / "08_Evaluation/phase4_qa_benchmark_v1_RECONSTRUCTED.csv"
review_path = project / "05_Gold_Evidence/phase3_record_review_log_RECONSTRUCTED.csv"
manifest_dir = project / "08_Evaluation"

# 1. Load both datasets
gold = pd.read_csv(gold_path)
benchmark = pd.read_csv(benchmark_path)

expected_ids = [f"CRIM_{i:03d}" for i in range(1, 11)]
required = [
    "question_id", "question", "reference_answer",
    "evidence_document", "case_name", "court",
    "judgment_date", "evidence_passage"
]

# 2. Validate structure and IDs
for name, df in [("Phase 3 gold evidence", gold),
                 ("Phase 4 benchmark", benchmark)]:
    missing_columns = [c for c in required if c not in df.columns]
    assert not missing_columns, f"{name}: missing columns {missing_columns}"
    assert len(df) == 10, f"{name}: expected 10 rows, found {len(df)}"
    assert df["question_id"].is_unique, f"{name}: duplicate question IDs"
    assert sorted(df["question_id"].astype(str).tolist()) == expected_ids, (
        f"{name}: question IDs do not match the expected set"
    )
    assert df[required].notna().all().all(), (
        f"{name}: one or more required fields are empty"
    )

# 3. Confirm CRIM_010 is corrected in both files
for name, df in [("Phase 3", gold), ("Phase 4", benchmark)]:
    passage = df.loc[
        df["question_id"].astype(str) == "CRIM_010",
        "evidence_passage"
    ].iloc[0]

    assert passage.endswith('"very substantial and compelling reasons."'), (
        f"{name}: CRIM_010 passage needs correction"
    )
    print(f"{name}: CRIM_010 passage PASS")

# 4. Verify Phase 3 and Phase 4 evidence passages agree
gold_passages = gold.set_index("question_id")["evidence_passage"].to_dict()
benchmark_passages = benchmark.set_index("question_id")["evidence_passage"].to_dict()

assert gold_passages == benchmark_passages, (
    "Phase 3 and Phase 4 evidence passages differ"
)

print("\nStructural validation: PASS")
print("Phase 3/Phase 4 evidence consistency: PASS")
print("Question IDs validated: 10/10")

# 5. Calculate the current benchmark checksum
sha256 = hashlib.sha256(benchmark_path.read_bytes()).hexdigest()

# 6. Find the reconstructed Phase 4 manifest
manifests = list(manifest_dir.glob("*manifest*RECONSTRUCTED*.json"))

if len(manifests) == 1:
    manifest_path = manifests[0]
    manifest = json.loads(manifest_path.read_text(encoding="utf-8"))
elif len(manifests) == 0:
    manifest_path = manifest_dir / "phase4_benchmark_manifest_v1_RECONSTRUCTED.json"
    manifest = {}
else:
    raise RuntimeError(
        "Multiple reconstructed manifests found. Inspect these files first:\n"
        + "\n".join(str(p) for p in manifests)
    )

# 7. Update provenance and checksum without deleting existing manifest fields
manifest.update({
    "status": "RECONSTRUCTED_EVIDENCE_MATCHES_VERIFIED",
    "benchmark_file": benchmark_path.name,
    "benchmark_sha256": sha256,
    "benchmark_rows": len(benchmark),
    "question_ids": expected_ids,
    "original_byte_identical": False,
    "legal_review_status": "PENDING",
    "metadata_review_status": "PENDING",
    "final_evidence_span_review_status": "PENDING"
})

manifest_path.write_text(
    json.dumps(manifest, indent=2, ensure_ascii=False),
    encoding="utf-8"
)

print("\nManifest updated:", manifest_path)
print("Benchmark SHA-256:", sha256)
print("Manifest status:", manifest["status"])
print("\nValidation and manifest update completed.")
```

```python
import pandas as pd
import numpy as np
import re
import json
from pathlib import Path

project = Path("/kaggle/working/indian_legal_rag")
results_dir = project / "09_Results_Figures"
results_dir.mkdir(parents=True, exist_ok=True)

# Use the existing full chunk table
chunk_texts = chunks["text"].fillna("").astype(str).tolist()
source_files = chunks["source_file"].astype(str).tolist()

assert len(chunk_texts) == faiss_index.ntotal == len(tokenized_corpus)

def tokenize_query(text):
    return re.findall(r"\b\w+\b", str(text).lower())

def dense_search(query, k=20):
    query_embedding = model.encode(
        [query],
        normalize_embeddings=True,
        convert_to_numpy=True
    ).astype("float32")

    scores, ids = faiss_index.search(query_embedding, k)
    return [
        (int(idx), float(score))
        for idx, score in zip(ids[0], scores[0])
        if idx >= 0
    ]

def bm25_search(query, k=20):
    scores = bm25.get_scores(tokenize_query(query))
    top_ids = np.argsort(scores)[::-1][:k]
    return [(int(idx), float(scores[idx])) for idx in top_ids]

def rrf_fusion(dense_results, bm25_results, c=60):
    fused_scores = {}

    for rank, (idx, _) in enumerate(dense_results, start=1):
        fused_scores[idx] = fused_scores.get(idx, 0) + 1 / (c + rank)

    for rank, (idx, _) in enumerate(bm25_results, start=1):
        fused_scores[idx] = fused_scores.get(idx, 0) + 1 / (c + rank)

    return sorted(
        fused_scores.items(),
        key=lambda item: item[1],
        reverse=True
    )

all_rankings = []

for _, row in benchmark.iterrows():
    qid = str(row["question_id"])
    query = str(row["question"])

    print(f"Processing {qid}...")

    dense_results = dense_search(query, k=20)
    bm25_results = bm25_search(query, k=20)
    hybrid_results = rrf_fusion(dense_results, bm25_results, c=60)

    # Rerank the top 20 hybrid candidates
    hybrid_candidates = hybrid_results[:20]

    pairs = [
        (query, chunk_texts[idx])
        for idx, _ in hybrid_candidates
    ]

    ce_scores = cross_encoder.predict(
        pairs,
        batch_size=8,
        show_progress_bar=False
    )

    reranked_results = sorted(
        [
            (idx, float(score))
            for (idx, _), score in zip(hybrid_candidates, ce_scores)
        ],
        key=lambda item: item[1],
        reverse=True
    )

    method_results = {
        "dense": dense_results[:10],
        "bm25": bm25_results[:10],
        "hybrid_rrf": hybrid_results[:10],
        "hybrid_rrf_cross_encoder": reranked_results[:10]
    }

    gold_file = str(row["evidence_document"])

    for method, results in method_results.items():
        for rank, (idx, score) in enumerate(results, start=1):
            all_rankings.append({
                "question_id": qid,
                "method": method,
                "rank": rank,
                "chunk_id": chunks.iloc[idx]["chunk_id"],
                "document_id": chunks.iloc[idx]["document_id"],
                "source_file": source_files[idx],
                "score": score,
                "gold_evidence_document": gold_file,
                "is_relevant_document": (
                    source_files[idx] == gold_file
                )
            })

rankings_df = pd.DataFrame(all_rankings)

output_path = results_dir / "phase7_retrieval_rankings_v1.csv"
rankings_df.to_csv(output_path, index=False)

print("\nRetrieval runs completed.")
print("Rows:", len(rankings_df))
print("Saved:", output_path)
print("\nResults per method:")
print(rankings_df.groupby("method").size())
```

```python
import pandas as pd
import numpy as np
import json
import hashlib
from pathlib import Path
from datetime import datetime, timezone

project = Path("/kaggle/working/indian_legal_rag")
results_dir = project / "09_Results_Figures"

rankings_path = results_dir / "phase7_retrieval_rankings_v1.csv"

rankings = pd.read_csv(rankings_path)

required_columns = {
    "question_id", "method", "rank",
    "source_file", "gold_evidence_document",
    "is_relevant_document"
}

assert required_columns.issubset(rankings.columns), (
    f"Missing columns: {required_columns - set(rankings.columns)}"
)

# Confirm every question-method pair has 10 ranked results
counts = rankings.groupby(["question_id", "method"]).size()
assert (counts == 10).all(), "Some question-method runs do not have 10 results."

def evaluate_one_run(group):
    group = group.sort_values("rank")

    relevant = (
        group["source_file"].astype(str).to_numpy()
        == group["gold_evidence_document"].iloc[0]
    )

    ranks = np.arange(1, len(relevant) + 1)
    relevant_ranks = ranks[relevant]

    def recall_at(k):
        return float(relevant[:k].any())

    def ndcg_at(k):
        rel = relevant[:k].astype(float)
        discounts = 1 / np.log2(np.arange(2, len(rel) + 2))
        dcg = np.sum(rel * discounts)

        # There is one known gold document for each question.
        ideal_length = min(1, k)
        idcg = 1.0 if ideal_length else 0.0

        return float(dcg / idcg) if idcg else 0.0

    return {
        "Recall@5": recall_at(5),
        "Recall@10": recall_at(10),
        "MRR": (
            float(1 / relevant_ranks[0])
            if len(relevant_ranks) else 0.0
        ),
        "nDCG@5": ndcg_at(5),
        "nDCG@10": ndcg_at(10),
        "first_relevant_rank": (
            int(relevant_ranks[0])
            if len(relevant_ranks) else None
        )
    }

# Per-question metrics
per_question_rows = []

for (qid, method), group in rankings.groupby(["question_id", "method"]):
    metrics = evaluate_one_run(group)
    per_question_rows.append({
        "question_id": qid,
        "method": method,
        **metrics
    })

per_question = pd.DataFrame(per_question_rows)

# Aggregate metrics across the 10 questions
metric_columns = [
    "Recall@5", "Recall@10", "MRR", "nDCG@5", "nDCG@10"
]

summary = (
    per_question.groupby("method")[metric_columns]
    .mean()
    .reset_index()
)

summary["num_questions"] = (
    per_question.groupby("method")["question_id"].nunique().values
)

# Save results
per_question_path = results_dir / "phase7_per_question_metrics_v1.csv"
summary_path = results_dir / "phase7_retrieval_metrics_summary_v1.csv"

per_question.to_csv(per_question_path, index=False)
summary.to_csv(summary_path, index=False)

# Create a reproducibility manifest
sha256 = hashlib.sha256(rankings_path.read_bytes()).hexdigest()

manifest = {
    "phase": 7,
    "status": "PILOT_RETRIEVAL_EVALUATION_COMPLETED",
    "evaluation_date_utc": datetime.now(timezone.utc).isoformat(),
    "benchmark": "phase4_qa_benchmark_v1_RECONSTRUCTED.csv",
    "benchmark_questions": int(per_question["question_id"].nunique()),
    "retrieval_methods": sorted(per_question["method"].unique().tolist()),
    "relevance_definition": (
        "Document-level binary relevance: a retrieved chunk is relevant "
        "when its source_file matches the question's gold evidence document."
    ),
    "metrics": metric_columns,
    "interpretation_limitations": [
        "The benchmark contains only 10 reconstructed questions.",
        "Results are preliminary pilot results, not definitive performance estimates.",
        "Document-level relevance does not verify exact evidence-span retrieval.",
        "The benchmark requires final legal-answer and metadata review."
    ],
    "rankings_file": rankings_path.name,
    "rankings_sha256": sha256,
    "per_question_metrics_file": per_question_path.name,
    "summary_file": summary_path.name
}

manifest_path = results_dir / "phase7_retrieval_evaluation_manifest_v1.json"
manifest_path.write_text(
    json.dumps(manifest, indent=2, ensure_ascii=False),
    encoding="utf-8"
)

print("=== RETRIEVAL EVALUATION SUMMARY ===")
display(summary.round(4))

print("\n=== SAVED FILES ===")
print(summary_path)
print(per_question_path)
print(manifest_path)
print("\nEvaluation completed successfully.")
```

```python
import pandas as pd
import numpy as np
import json
import hashlib
from pathlib import Path
from datetime import datetime, timezone

project = Path("/kaggle/working/indian_legal_rag")
results_dir = project / "09_Results_Figures"

rankings_path = results_dir / "phase7_retrieval_rankings_v1.csv"
chunks_path = project / "04_Legal_Corpus/phase5_legal_chunks_v2.csv"

rankings = pd.read_csv(rankings_path)
chunks = pd.read_csv(chunks_path)

# Number of relevant chunks available for each source document.
# All chunks from the gold evidence document are relevant under
# this document-level relevance definition.
chunks_per_source = (
    chunks.groupby("source_file")["chunk_id"]
    .count()
    .to_dict()
)

def evaluate_run(group):
    group = group.sort_values("rank").copy()

    gold_file = str(group["gold_evidence_document"].iloc[0])
    relevant = (
        group["source_file"].astype(str).to_numpy() == gold_file
    )

    ranks = np.arange(1, len(relevant) + 1)
    relevant_ranks = ranks[relevant]

    def hit_rate(k):
        return float(relevant[:k].any())

    def ndcg(k):
        rel = relevant[:k].astype(float)
        discounts = 1.0 / np.log2(np.arange(2, len(rel) + 2))
        dcg = float(np.sum(rel * discounts))

        # Correct ideal DCG: account for all relevant chunks available
        # in the full corpus, capped at the evaluation cutoff.
        total_relevant = chunks_per_source.get(gold_file, 0)
        ideal_relevant_count = min(total_relevant, k)

        if ideal_relevant_count == 0:
            return 0.0

        ideal_discounts = (
            1.0 / np.log2(np.arange(2, ideal_relevant_count + 2))
        )
        idcg = float(np.sum(ideal_discounts))

        return dcg / idcg if idcg > 0 else 0.0

    return {
        "HitRate@5": hit_rate(5),
        "HitRate@10": hit_rate(10),
        "MRR": (
            float(1.0 / relevant_ranks[0])
            if len(relevant_ranks) else 0.0
        ),
        "nDCG@5": ndcg(5),
        "nDCG@10": ndcg(10),
        "first_relevant_rank": (
            int(relevant_ranks[0])
            if len(relevant_ranks) else None
        )
    }

# Calculate per-question metrics
rows = []

for (qid, method), group in rankings.groupby(["question_id", "method"]):
    rows.append({
        "question_id": qid,
        "method": method,
        **evaluate_run(group)
    })

per_question = pd.DataFrame(rows)

metric_columns = [
    "HitRate@5", "HitRate@10", "MRR", "nDCG@5", "nDCG@10"
]

summary = (
    per_question.groupby("method")[metric_columns]
    .mean()
    .reset_index()
)

summary["num_questions"] = (
    per_question.groupby("method")["question_id"].nunique().reindex(
        summary["method"]
    ).to_numpy()
)

# Validate metric ranges
for col in metric_columns:
    assert summary[col].between(0, 1).all(), (
        f"Invalid metric values in {col}"
    )

# Save corrected results separately to preserve the initial outputs
per_question_path = (
    results_dir / "phase7_per_question_metrics_v2_corrected.csv"
)
summary_path = (
    results_dir / "phase7_retrieval_metrics_summary_v2_corrected.csv"
)

per_question.to_csv(per_question_path, index=False)
summary.to_csv(summary_path, index=False)

manifest = {
    "phase": 7,
    "status": "PILOT_RETRIEVAL_EVALUATION_METRICS_CORRECTED",
    "evaluation_date_utc": datetime.now(timezone.utc).isoformat(),
    "benchmark": "phase4_qa_benchmark_v1_RECONSTRUCTED.csv",
    "num_questions": int(per_question["question_id"].nunique()),
    "relevance_definition": (
        "A retrieved chunk is relevant if its source_file matches the "
        "question's gold evidence document."
    ),
    "hit_rate_definition": (
        "Fraction of questions with at least one relevant chunk in the top k."
    ),
    "ndcg_definition": (
        "Binary chunk relevance, normalized using the ideal ranking based "
        "on the number of chunks from the gold document in the full corpus."
    ),
    "metrics": metric_columns,
    "limitations": [
        "Only 10 reconstructed benchmark questions were evaluated.",
        "Document-level relevance does not verify exact evidence-span retrieval.",
        "Results are preliminary pilot results.",
        "Final legal-answer and metadata review remains pending."
    ],
    "rankings_sha256": hashlib.sha256(rankings_path.read_bytes()).hexdigest(),
    "summary_file": summary_path.name,
    "per_question_metrics_file": per_question_path.name
}

manifest_path = (
    results_dir / "phase7_retrieval_evaluation_manifest_v2_corrected.json"
)
manifest_path.write_text(
    json.dumps(manifest, indent=2, ensure_ascii=False),
    encoding="utf-8"
)

print("=== CORRECTED RETRIEVAL METRICS ===")
display(summary.round(4))

print("\nSaved corrected summary:", summary_path)
print("Saved per-question metrics:", per_question_path)
print("Saved manifest:", manifest_path)
print("\nAll metric values are within [0, 1].")
```

```python
import pandas as pd
from pathlib import Path

project = Path("/kaggle/working/indian_legal_rag")
results_dir = project / "09_Results_Figures"

# Load ranked retrieval results
rankings_path = results_dir / "phase7_retrieval_rankings_v1.csv"
rankings = pd.read_csv(rankings_path)

# Load complete chunk corpus
chunks_path = project / "04_Legal_Corpus/phase5_legal_chunks_v2.csv"
chunks = pd.read_csv(chunks_path)

# Gold benchmark
benchmark_path = (
    project / "08_Evaluation/phase4_qa_benchmark_v1_RECONSTRUCTED.csv"
)
gold = pd.read_csv(benchmark_path)

# Select the top 5 cross-encoder results for each question
top_results = rankings[
    (rankings["method"] == "hybrid_rrf_cross_encoder") &
    (rankings["rank"] <= 5)
].copy()

# Attach retrieved chunk text
top_results = top_results.merge(
    chunks[["chunk_id", "text"]],
    on="chunk_id",
    how="left",
    validate="many_to_one"
)

# Attach the reference evidence passage
top_results = top_results.merge(
    gold[["question_id", "evidence_passage", "evidence_document"]],
    on="question_id",
    how="left",
    validate="many_to_one"
)

assert top_results["text"].notna().all(), "Some retrieved chunks have no text."

# Check whether the exact gold passage appears in each retrieved chunk.
# Whitespace differences are normalized; no semantic match is assumed.
def normalize(text):
    return " ".join(str(text).split()).casefold()

top_results["exact_passage_match"] = top_results.apply(
    lambda row: normalize(row["evidence_passage"]) in normalize(row["text"]),
    axis=1
)

summary = (
    top_results.groupby("question_id")
    .agg(
        top5_chunks=("chunk_id", "count"),
        correct_case_chunks=("is_relevant_document", "sum"),
        exact_passage_found=("exact_passage_match", "any")
    )
    .reset_index()
)

print("=== EVIDENCE AVAILABILITY CHECK ===")
display(summary)

print("\nQuestions with the exact evidence passage in top 5:",
      int(summary["exact_passage_found"].sum()), "/", len(summary))

# Save the detailed evidence check
output_path = results_dir / "phase8_evidence_availability_check_v1.csv"
top_results.to_csv(output_path, index=False)

print("\nSaved:", output_path)
```

```python
import pandas as pd
from pathlib import Path

project = Path("/kaggle/working/indian_legal_rag")
results_dir = project / "09_Results_Figures"

chunks = pd.read_csv(
    project / "04_Legal_Corpus/phase5_legal_chunks_v2.csv"
)
benchmark = pd.read_csv(
    project / "08_Evaluation/phase4_qa_benchmark_v1_RECONSTRUCTED.csv"
)

def normalize(text):
    return " ".join(str(text).split()).casefold()

diagnostic_rows = []

for _, q in benchmark.iterrows():
    gold_file = str(q["evidence_document"])
    evidence = normalize(q["evidence_passage"])

    # Restrict to chunks belonging to the known gold document
    doc_chunks = chunks[
        chunks["source_file"].astype(str) == gold_file
    ].copy()

    # Check for exact normalized evidence passage in any chunk
    doc_chunks["exact_passage_match"] = (
        doc_chunks["text"].map(normalize).str.contains(
            evidence, regex=False, na=False
        )
    )

    matching = doc_chunks[doc_chunks["exact_passage_match"]]

    diagnostic_rows.append({
        "question_id": q["question_id"],
        "evidence_document": gold_file,
        "chunks_in_gold_document": len(doc_chunks),
        "exact_passage_found_anywhere": not matching.empty,
        "matching_chunk_ids": "; ".join(matching["chunk_id"].astype(str).tolist()),
    })

diagnostic = pd.DataFrame(diagnostic_rows)

output_path = results_dir / "phase8_gold_document_diagnostic_v1.csv"
diagnostic.to_csv(output_path, index=False)

display(diagnostic)
print("\nSaved:", output_path)
```

```python
import pandas as pd
from pathlib import Path

project = Path("/kaggle/working/indian_legal_rag")
results_dir = project / "09_Results_Figures"

diagnostic = pd.read_csv(
    results_dir / "phase8_gold_document_diagnostic_v1.csv"
)
rankings = pd.read_csv(
    results_dir / "phase7_retrieval_rankings_v1.csv"
)

# Expand matching chunk IDs into one row per question/chunk
matches = diagnostic[
    diagnostic["exact_passage_found_anywhere"].astype(str).str.lower().eq("true")
].copy()

matches["matching_chunk_ids"] = matches["matching_chunk_ids"].fillna("")
matches["matching_chunk_ids"] = matches["matching_chunk_ids"].apply(
    lambda x: [item.strip() for item in str(x).split(";") if item.strip()]
)
matches = matches.explode("matching_chunk_ids").rename(
    columns={"matching_chunk_ids": "chunk_id"}
)

# Join to retrieval rankings
rank_check = matches.merge(
    rankings,
    on=["question_id", "chunk_id"],
    how="left"
)

# Summarize the rank of each matching chunk
rank_summary = rank_check[
    ["question_id", "chunk_id", "method", "rank"]
].sort_values(["question_id", "method", "rank"])

display(rank_summary)

output_path = results_dir / "phase8_exact_evidence_rank_check_v1.csv"
rank_summary.to_csv(output_path, index=False)

print("\nSaved:", output_path)
print(
    "\nExact evidence chunks ranked within top 10:",
    int(rank_summary["rank"].notna().sum()),
    "method/chunk matches"
)
```

```python
import pandas as pd
from pathlib import Path

project = Path("/kaggle/working/indian_legal_rag")
results_dir = project / "09_Results_Figures"

diagnostic = pd.read_csv(
    results_dir / "phase8_gold_document_diagnostic_v1.csv"
)
rankings = pd.read_csv(
    results_dir / "phase7_retrieval_rankings_v1.csv"
)
chunks = pd.read_csv(
    project / "04_Legal_Corpus/phase5_legal_chunks_v2.csv"
)

rows = []

for _, row in diagnostic.iterrows():
    qid = row["question_id"]
    chunk_ids = [
        x.strip()
        for x in str(row.get("matching_chunk_ids", "")).split(";")
        if x.strip() and x.strip().lower() != "nan"
    ]

    if not chunk_ids:
        rows.append({
            "question_id": qid,
            "chunk_id": None,
            "exists_in_corpus": False,
            "method": None,
            "rank": None,
            "status": "No exact passage match found in gold document"
        })
        continue

    for chunk_id in chunk_ids:
        exists = chunks["chunk_id"].astype(str).eq(chunk_id).any()
        found = rankings[
            (rankings["question_id"] == qid) &
            (rankings["chunk_id"] == chunk_id)
        ]

        if found.empty:
            for method in [
                "dense", "bm25", "hybrid_rrf",
                "hybrid_rrf_cross_encoder"
            ]:
                rows.append({
                    "question_id": qid,
                    "chunk_id": chunk_id,
                    "exists_in_corpus": exists,
                    "method": method,
                    "rank": None,
                    "status": "Not in saved top 10; rank beyond 10 or not retrieved"
                })
        else:
            for _, hit in found.iterrows():
                rows.append({
                    "question_id": qid,
                    "chunk_id": chunk_id,
                    "exists_in_corpus": exists,
                    "method": hit["method"],
                    "rank": int(hit["rank"]),
                    "status": "Retrieved in top 10"
                })

rank_check = pd.DataFrame(rows)

display(rank_check)

output_path = results_dir / "phase8_exact_evidence_rank_check_v2.csv"
rank_check.to_csv(output_path, index=False)

print("\nSaved:", output_path)
```

```python
import pandas as pd
import numpy as np
import re
from pathlib import Path

project = Path("/kaggle/working/indian_legal_rag")
results_dir = project / "09_Results_Figures"

chunks = pd.read_csv(
    project / "04_Legal_Corpus/phase5_legal_chunks_v2.csv"
)
benchmark = pd.read_csv(
    project / "08_Evaluation/phase4_qa_benchmark_v1_RECONSTRUCTED.csv"
)
diagnostic = pd.read_csv(
    results_dir / "phase8_gold_document_diagnostic_v1.csv"
)

chunk_texts = chunks["text"].fillna("").astype(str).tolist()
source_files = chunks["source_file"].astype(str).tolist()

def tokenize_query(text):
    return re.findall(r"\b\w+\b", str(text).lower())

def dense_search_50(query, k=50):
    qvec = model.encode(
        [query],
        normalize_embeddings=True,
        convert_to_numpy=True
    ).astype("float32")
    scores, ids = faiss_index.search(qvec, k)
    return [
        (int(i), float(s))
        for i, s in zip(ids[0], scores[0])
        if i >= 0
    ]

def bm25_search_50(query, k=50):
    scores = bm25.get_scores(tokenize_query(query))
    ids = np.argsort(scores)[::-1][:k]
    return [(int(i), float(scores[i])) for i in ids]

def rrf_fusion_50(dense_results, bm25_results, c=60):
    scores = {}
    for rank, (idx, _) in enumerate(dense_results, start=1):
        scores[idx] = scores.get(idx, 0) + 1 / (c + rank)
    for rank, (idx, _) in enumerate(bm25_results, start=1):
        scores[idx] = scores.get(idx, 0) + 1 / (c + rank)
    return sorted(scores.items(), key=lambda x: x[1], reverse=True)

rows = []

for _, q in benchmark.iterrows():
    qid = str(q["question_id"])
    query = str(q["question"])
    print("Processing", qid)

    dense = dense_search_50(query, 50)
    lexical = bm25_search_50(query, 50)
    hybrid = rrf_fusion_50(dense, lexical, 60)

    # Cross-encode the top 50 hybrid candidates.
    candidates = hybrid[:50]
    pairs = [(query, chunk_texts[idx]) for idx, _ in candidates]
    ce_scores = cross_encoder.predict(
        pairs, batch_size=8, show_progress_bar=False
    )
    reranked = sorted(
        [(idx, float(score)) for (idx, _), score in zip(candidates, ce_scores)],
        key=lambda x: x[1],
        reverse=True
    )

    methods = {
        "dense": dense,
        "bm25": lexical,
        "hybrid_rrf": hybrid,
        "hybrid_rrf_cross_encoder": reranked
    }

    for method, results in methods.items():
        for rank, (idx, score) in enumerate(results, start=1):
            rows.append({
                "question_id": qid,
                "method": method,
                "rank": rank,
                "chunk_id": chunks.iloc[idx]["chunk_id"],
                "source_file": source_files[idx],
                "score": score,
                "gold_evidence_document": str(q["evidence_document"]),
                "is_relevant_document": (
                    source_files[idx] == str(q["evidence_document"])
                )
            })

expanded = pd.DataFrame(rows)
out = results_dir / "phase8_retrieval_top50_diagnostic_v1.csv"
expanded.to_csv(out, index=False)

# Check the exact evidence chunks known to exist in the corpus.
match_rows = []
for _, d in diagnostic.iterrows():
    ids = [
        x.strip() for x in str(d.get("matching_chunk_ids", "")).split(";")
        if x.strip() and x.strip().lower() != "nan"
    ]
    for chunk_id in ids:
        for method in methods:
            hits = expanded[
                (expanded["question_id"] == d["question_id"]) &
                (expanded["method"] == method) &
                (expanded["chunk_id"] == chunk_id)
            ]
            match_rows.append({
                "question_id": d["question_id"],
                "chunk_id": chunk_id,
                "method": method,
                "rank_within_top50": (
                    int(hits["rank"].iloc[0]) if not hits.empty else None
                )
            })

match_check = pd.DataFrame(match_rows)
display(match_check)

print("\nSaved expanded rankings:", out)
print("Rows:", len(expanded))
```

```python
import re
import numpy as np
import pandas as pd
from pathlib import Path

target_chunk_id = "doc_5820_v2_chunk_0008"

query_variants = {
    "original": (
        "According to Ranjan Dwivedi v. CBI (2012), "
        "what does Article 21 protect regarding the speed of criminal proceedings?"
    ),
    "legal_terminology": (
        "Article 21 right to speedy trial accused investigation inquiry "
        "trial appeal revision retrial"
    ),
}

def tokenize_query(text):
    return re.findall(r"\b\w+\b", text.lower())

target_idx = chunks.index[
    chunks["chunk_id"].astype(str) == target_chunk_id
].tolist()

assert len(target_idx) == 1
target_idx = target_idx[0]

results = []

for variant, query in query_variants.items():
    query_embedding = model.encode(
        [query],
        normalize_embeddings=True,
        convert_to_numpy=True
    ).astype("float32")

    dense_scores, dense_ids = faiss_index.search(query_embedding, 50)
    dense_matches = np.where(dense_ids[0] == target_idx)[0]

    bm25_scores = bm25.get_scores(tokenize_query(query))
    bm25_order = np.argsort(bm25_scores)[::-1]
    bm25_matches = np.where(bm25_order == target_idx)[0]

    results.append({
        "query_variant": variant,
        "dense_target_rank": (
            int(dense_matches[0] + 1) if len(dense_matches) else ">50"
        ),
        "dense_target_score": float(
            np.dot(query_embedding[0], embeddings[target_idx])
        ),
        "bm25_target_rank": (
            int(bm25_matches[0] + 1) if len(bm25_matches) else ">50"
        ),
        "bm25_target_score": float(bm25_scores[target_idx]),
    })

display(pd.DataFrame(results))
```

```python
import pandas as pd
import numpy as np
import re

target_chunk_id = "doc_5820_v2_chunk_0008"

query_variants = {
    "original": (
        "According to Ranjan Dwivedi v. CBI (2012), "
        "what does Article 21 protect regarding the speed of criminal proceedings?"
    ),
    "legal_terminology": (
        "Article 21 right to speedy trial accused investigation inquiry "
        "trial appeal revision retrial"
    ),
}

def tokenize_query(text):
    return re.findall(r"\b\w+\b", text.lower())

def reciprocal_rank_fusion(rankings, k=60):
    scores = {}

    for ranking in rankings:
        for rank, chunk_idx in enumerate(ranking, start=1):
            scores[chunk_idx] = scores.get(chunk_idx, 0) + 1 / (k + rank)

    return sorted(scores, key=scores.get, reverse=True), scores

results = []

for variant, query in query_variants.items():

    # Dense candidates
    query_embedding = model.encode(
        [query],
        normalize_embeddings=True,
        convert_to_numpy=True
    ).astype("float32")

    _, dense_ids = faiss_index.search(query_embedding, 50)
    dense_ranking = [int(i) for i in dense_ids[0] if i >= 0]

    # BM25 candidates
    bm25_scores = bm25.get_scores(tokenize_query(query))
    bm25_ranking = np.argsort(bm25_scores)[::-1][:50].tolist()

    # Hybrid RRF
    hybrid_ranking, rrf_scores = reciprocal_rank_fusion(
        [dense_ranking, bm25_ranking]
    )

    # Rerank the top 50 hybrid candidates
    candidates = hybrid_ranking[:50]

    pairs = [
        (query, str(chunks.iloc[idx]["text"]))
        for idx in candidates
    ]

    ce_scores = cross_encoder.predict(pairs)

    reranked_positions = np.argsort(ce_scores)[::-1]
    reranked = [candidates[i] for i in reranked_positions]

    target_matches = [
        ("hybrid_rrf", rank)
        for rank, idx in enumerate(hybrid_ranking, start=1)
        if str(chunks.iloc[idx]["chunk_id"]) == target_chunk_id
    ]

    target_reranked = [
        ("cross_encoder", rank)
        for rank, idx in enumerate(reranked, start=1)
        if str(chunks.iloc[idx]["chunk_id"]) == target_chunk_id
    ]

    results.append({
        "query_variant": variant,
        "target_in_hybrid_top50": any(
            str(chunks.iloc[idx]["chunk_id"]) == target_chunk_id
            for idx in hybrid_ranking[:50]
        ),
        "hybrid_rank": (
            target_matches[0][1] if target_matches else ">50"
        ),
        "target_in_reranked_top50": any(
            str(chunks.iloc[idx]["chunk_id"]) == target_chunk_id
            for idx in reranked
        ),
        "cross_encoder_rank": (
            target_reranked[0][1] if target_reranked else ">50"
        ),
    })

display(pd.DataFrame(results))
```

```python

import re
import numpy as np
import pandas as pd
from pathlib import Path

# --------------------------------------------------
# 1. Load the Phase 8 query-expansion dataset
# --------------------------------------------------
base_dir = Path("/kaggle/working/indian_legal_rag")
results_dir = base_dir / "09_Results_Figures"

query_df = pd.read_csv(
    results_dir / "phase8_query_expansion_pilot_v1.csv"
)

assert len(query_df) == 10
assert {"question_id", "question", "expanded_query",
        "evidence_document"}.issubset(query_df.columns)

# --------------------------------------------------
# 2. Retrieval helpers
# --------------------------------------------------
def tokenize_query(text):
    return re.findall(r"\b\w+\b", str(text).lower())

def rrf_fusion(rankings, k=60):
    scores = {}

    for ranking in rankings:
        for rank, idx in enumerate(ranking, start=1):
            scores[idx] = scores.get(idx, 0.0) + 1.0 / (k + rank)

    return sorted(scores, key=scores.get, reverse=True)

def get_relevant_count(gold_doc):
    return int(
        chunks["source_file"].astype(str).eq(str(gold_doc)).sum()
    )

def calculate_metrics(ranking, gold_doc):
    relevant = (
        chunks["source_file"].astype(str).to_numpy()
        == str(gold_doc)
    )

    top10 = ranking[:10]
    hits = [int(relevant[idx]) for idx in top10]

    hitrate5 = int(any(relevant[idx] for idx in ranking[:5]))
    hitrate10 = int(any(relevant[idx] for idx in ranking[:10]))

    relevant_ranks = [
        rank
        for rank, idx in enumerate(ranking, start=1)
        if relevant[idx]
    ]
    mrr = 1.0 / relevant_ranks[0] if relevant_ranks else 0.0

    def ndcg_at_k(k):
        ranked_hits = [
            int(relevant[idx]) for idx in ranking[:k]
        ]
        dcg = sum(
            rel / np.log2(rank + 1)
            for rank, rel in enumerate(ranked_hits, start=1)
        )

        total_relevant = get_relevant_count(gold_doc)
        ideal_relevant = min(total_relevant, k)
        idcg = sum(
            1.0 / np.log2(rank + 1)
            for rank in range(1, ideal_relevant + 1)
        )

        return dcg / idcg if idcg else 0.0

    return {
        "HitRate@5": hitrate5,
        "HitRate@10": hitrate10,
        "MRR": mrr,
        "nDCG@5": ndcg_at_k(5),
        "nDCG@10": ndcg_at_k(10),
    }

# Map each chunk index to its source document
source_docs = chunks["source_file"].astype(str).to_numpy()

ranking_records = []
metric_records = []

# --------------------------------------------------
# 3. Evaluate both query versions
# --------------------------------------------------
for _, row in query_df.iterrows():

    qid = row["question_id"]
    gold_doc = str(row["evidence_document"])

    query_versions = {
        "baseline": str(row["question"]),
        "expanded": str(row["expanded_query"]),
    }

    for query_type, query in query_versions.items():

        # Dense retrieval: top 50
        query_embedding = model.encode(
            [query],
            normalize_embeddings=True,
            convert_to_numpy=True
        ).astype("float32")

        _, dense_ids = faiss_index.search(query_embedding, 50)
        dense_ranking = [
            int(idx) for idx in dense_ids[0] if idx >= 0
        ]

        # BM25 retrieval: top 50
        bm25_scores = bm25.get_scores(tokenize_query(query))
        bm25_ranking = np.argsort(bm25_scores)[::-1][:50].tolist()

        # Hybrid RRF
        hybrid_ranking = rrf_fusion(
            [dense_ranking, bm25_ranking]
        )

        # Cross-encoder reranks the top 50 hybrid candidates
        candidates = hybrid_ranking[:50]

        pairs = [
            (query, str(chunks.iloc[idx]["text"]))
            for idx in candidates
        ]

        ce_scores = cross_encoder.predict(pairs)
        order = np.argsort(ce_scores)[::-1]
        reranked = [candidates[i] for i in order]

        method_rankings = {
            "dense": dense_ranking,
            "bm25": bm25_ranking,
            "hybrid_rrf": hybrid_ranking,
            "hybrid_rrf_cross_encoder": reranked,
        }

        for method, ranking in method_rankings.items():

            metrics = calculate_metrics(ranking, gold_doc)

            metric_records.append({
                "question_id": qid,
                "query_type": query_type,
                "method": method,
                "evidence_document": gold_doc,
                **metrics,
            })

            # Save top 10 results for inspection
            for rank, idx in enumerate(ranking[:10], start=1):
                ranking_records.append({
                    "question_id": qid,
                    "query_type": query_type,
                    "method": method,
                    "rank": rank,
                    "chunk_id": chunks.iloc[idx]["chunk_id"],
                    "source_file": source_docs[idx],
                    "is_relevant": int(
                        source_docs[idx] == gold_doc
                    ),
                })

# --------------------------------------------------
# 4. Save separate Phase 8 outputs
# --------------------------------------------------
metrics_df = pd.DataFrame(metric_records)
rankings_df = pd.DataFrame(ranking_records)

metrics_path = (
    results_dir / "phase8_query_expansion_metrics_v1.csv"
)
rankings_path = (
    results_dir / "phase8_query_expansion_rankings_v1.csv"
)

metrics_df.to_csv(metrics_path, index=False)
rankings_df.to_csv(rankings_path, index=False)

# Average metrics across the 10 questions
summary = (
    metrics_df.groupby(["query_type", "method"])[
        ["HitRate@5", "HitRate@10", "MRR", "nDCG@5", "nDCG@10"]
    ]
    .mean()
    .reset_index()
)

summary_path = (
    results_dir / "phase8_query_expansion_summary_v1.csv"
)
summary.to_csv(summary_path, index=False)

print("Evaluation complete.")
print("Metrics:", metrics_path)
print("Rankings:", rankings_path)
print("Summary:", summary_path)

display(
    summary.round(4).sort_values(
        ["method", "query_type"]
    ).reset_index(drop=True)
)
```

```python
import numpy as np
from collections import defaultdict
import time

start = time.time()

# Build an inverted index from the existing BM25 model.
# Each term maps to document indices and term frequencies.
bm25_inverted = defaultdict(list)

for doc_idx, term_freqs in enumerate(bm25.doc_freqs):
    for term, freq in term_freqs.items():
        bm25_inverted[term].append((doc_idx, freq))

print("Inverted index built.")
print("Unique terms:", len(bm25_inverted))
print("Time taken:", round(time.time() - start, 2), "seconds")
```

```python
def fast_bm25_scores(query, bm25_model, inverted_index, n_docs):
    query_terms = tokenize_query(query)
    scores = np.zeros(n_docs, dtype=np.float32)

    doc_len = np.asarray(bm25_model.doc_len)
    avgdl = bm25_model.avgdl
    k1 = bm25_model.k1
    b = bm25_model.b

    for term in query_terms:
        if term not in inverted_index:
            continue

        idf = bm25_model.idf.get(term, 0.0)

        for doc_idx, freq in inverted_index[term]:
            denominator = freq + k1 * (
                1 - b + b * doc_len[doc_idx] / avgdl
            )

            scores[doc_idx] += idf * (
                freq * (k1 + 1) / denominator
            )

    return scores

print("Optimized BM25 function ready.")
```

```python
query = (
    "Article 21 right to speedy trial accused investigation inquiry "
    "trial appeal revision retrial"
)

start = time.time()

scores = fast_bm25_scores(
    query,
    bm25,
    bm25_inverted,
    len(chunks)
)

top_indices = np.argsort(scores)[::-1][:10]

print("Time:", round(time.time() - start, 2), "seconds")

display(
    chunks.iloc[top_indices][
        ["chunk_id", "source_file", "text"]
    ].assign(
        bm25_score=scores[top_indices]
    )
)
```

```python
import numpy as np
import pandas as pd
import time
from pathlib import Path

results_dir = Path(
    "/kaggle/working/indian_legal_rag/09_Results_Figures"
)

query_df = pd.read_csv(
    results_dir / "phase8_query_expansion_pilot_v1.csv"
)

source_docs = chunks["source_file"].astype(str).to_numpy()
n_chunks = len(chunks)

def rrf_fusion(rankings, k=60):
    scores = {}
    for ranking in rankings:
        for rank, idx in enumerate(ranking, start=1):
            scores[idx] = scores.get(idx, 0.0) + 1.0 / (k + rank)
    return sorted(scores, key=scores.get, reverse=True)

def evaluate_ranking(ranking, gold_doc):
    relevant = source_docs == str(gold_doc)

    hit5 = int(any(relevant[i] for i in ranking[:5]))
    hit10 = int(any(relevant[i] for i in ranking[:10]))

    relevant_ranks = [
        rank for rank, idx in enumerate(ranking, 1)
        if relevant[idx]
    ]
    mrr = 1.0 / relevant_ranks[0] if relevant_ranks else 0.0

    total_relevant = int(np.sum(relevant))

    def ndcg(k):
        rels = [int(relevant[i]) for i in ranking[:k]]
        dcg = sum(
            rel / np.log2(rank + 1)
            for rank, rel in enumerate(rels, 1)
        )
        ideal_n = min(total_relevant, k)
        idcg = sum(
            1.0 / np.log2(rank + 1)
            for rank in range(1, ideal_n + 1)
        )
        return dcg / idcg if idcg else 0.0

    return {
        "HitRate@5": hit5,
        "HitRate@10": hit10,
        "MRR": mrr,
        "nDCG@5": ndcg(5),
        "nDCG@10": ndcg(10),
    }

metric_rows = []
ranking_rows = []

start = time.time()

for _, row in query_df.iterrows():
    qid = row["question_id"]
    gold_doc = str(row["evidence_document"])

    for query_type, query in {
        "baseline": str(row["question"]),
        "expanded": str(row["expanded_query"]),
    }.items():

        # Dense retrieval
        query_embedding = model.encode(
            [query],
            normalize_embeddings=True,
            convert_to_numpy=True
        ).astype("float32")

        _, dense_ids = faiss_index.search(query_embedding, 50)
        dense_ranking = [
            int(i) for i in dense_ids[0] if i >= 0
        ]

        # Optimized BM25 retrieval
        bm25_scores = fast_bm25_scores(
            query, bm25, bm25_inverted, n_chunks
        )
        bm25_ranking = np.argsort(bm25_scores)[::-1][:50].tolist()

        # Hybrid RRF
        hybrid_ranking = rrf_fusion(
            [dense_ranking, bm25_ranking]
        )

        # Cross-encoder reranking
        candidates = hybrid_ranking[:50]
        pairs = [
            (query, str(chunks.iloc[idx]["text"]))
            for idx in candidates
        ]

        ce_scores = cross_encoder.predict(pairs)
        order = np.argsort(ce_scores)[::-1]
        reranked = [candidates[i] for i in order]

        methods = {
            "dense": dense_ranking,
            "bm25": bm25_ranking,
            "hybrid_rrf": hybrid_ranking,
            "hybrid_rrf_cross_encoder": reranked,
        }

        for method, ranking in methods.items():
            metrics = evaluate_ranking(ranking, gold_doc)

            metric_rows.append({
                "question_id": qid,
                "query_type": query_type,
                "method": method,
                "evidence_document": gold_doc,
                **metrics,
            })

            for rank, idx in enumerate(ranking[:10], 1):
                ranking_rows.append({
                    "question_id": qid,
                    "query_type": query_type,
                    "method": method,
                    "rank": rank,
                    "chunk_id": chunks.iloc[idx]["chunk_id"],
                    "source_file": source_docs[idx],
                    "is_relevant": int(source_docs[idx] == gold_doc),
                })

    print(f"Completed {qid}")

metrics_df = pd.DataFrame(metric_rows)
rankings_df = pd.DataFrame(ranking_rows)

summary = (
    metrics_df.groupby(["query_type", "method"])[
        ["HitRate@5", "HitRate@10", "MRR", "nDCG@5", "nDCG@10"]
    ]
    .mean()
    .reset_index()
)

metrics_df.to_csv(
    results_dir / "phase8_query_expansion_metrics_v1.csv",
    index=False
)
rankings_df.to_csv(
    results_dir / "phase8_query_expansion_rankings_v1.csv",
    index=False
)
summary.to_csv(
    results_dir / "phase8_query_expansion_summary_v1.csv",
    index=False
)

print("\nEvaluation completed in", round(time.time() - start, 2), "seconds")
display(summary.round(4))
```

```python
import pandas as pd
from pathlib import Path

results_dir = Path(
    "/kaggle/working/indian_legal_rag/09_Results_Figures"
)

drafts = pd.read_csv(
    results_dir / "phase9_grounded_answer_drafts_v1.csv"
)

evidence = pd.read_csv(
    results_dir / "phase9_retrieved_evidence_top5_v1.csv"
)

# Check that every cited chunk exists in the retrieved evidence file
available_chunks = set(evidence["chunk_id"].dropna().astype(str))

def check_citations(value):
    if pd.isna(value) or not str(value).strip():
        return False

    cited = [
        chunk.strip()
        for chunk in str(value).split(";")
        if chunk.strip()
    ]

    return bool(cited) and all(
        chunk in available_chunks for chunk in cited
    )

drafts["citation_ids_exist"] = drafts[
    "supporting_chunk_ids"
].apply(check_citations)

# Report draft readiness without claiming semantic support is verified
drafts["next_action"] = drafts["status"].map({
    "DRAFT_SUPPORTED": "Review answer against cited passage",
    "RETRIEVAL_REQUIRED": "Improve retrieval before finalizing answer"
})

display(
    drafts[
        [
            "question_id",
            "status",
            "citation_ids_exist",
            "next_action"
        ]
    ]
)

print("\nDrafts:", len(drafts))
print("Citation IDs validated:", int(drafts["citation_ids_exist"].sum()))
print(
    "Questions requiring retrieval:",
    int((drafts["status"] == "RETRIEVAL_REQUIRED").sum())
)
```

```python
import os
import json
import pandas as pd
from datetime import datetime, timezone

# Paths
ROOT = "/kaggle/working/indian_legal_rag"
RESULTS_DIR = os.path.join(ROOT, "09_Results_Figures")

DRAFT_PATH = os.path.join(
    RESULTS_DIR, "phase9_grounded_answer_drafts_v4.csv"
)
CITATION_PATH = os.path.join(
    RESULTS_DIR, "phase9_citation_evidence_review_v1.csv"
)

REPORT_CSV = os.path.join(
    RESULTS_DIR, "phase9_review_status_report_v1.csv"
)
REPORT_JSON = os.path.join(
    RESULTS_DIR, "phase9_completion_report_v1.json"
)

# Load Phase 9 artifacts
drafts = pd.read_csv(DRAFT_PATH)

required_columns = [
    "question_id",
    "draft_answer",
    "supporting_chunk_ids",
    "evidence_review",
    "human_review_status"
]

missing = [c for c in required_columns if c not in drafts.columns]
if missing:
    raise ValueError(f"Missing required columns in draft CSV: {missing}")

# Check citation IDs against the citation-review artifact, if available
if os.path.exists(CITATION_PATH):
    citation_review = pd.read_csv(CITATION_PATH)

    if "chunk_id" in citation_review.columns:
        available_chunk_ids = set(
            citation_review["chunk_id"].dropna().astype(str)
        )
    else:
        available_chunk_ids = set()

    citation_validation_available = bool(available_chunk_ids)
else:
    citation_review = pd.DataFrame()
    available_chunk_ids = set()
    citation_validation_available = False

# Build question-level report
report_rows = []

for _, row in drafts.iterrows():
    question_id = str(row["question_id"])
    cited_ids = [
        x.strip()
        for x in str(row["supporting_chunk_ids"]).split(";")
        if x.strip() and x.strip().lower() != "nan"
    ]

    if citation_validation_available:
        missing_ids = [
            chunk_id for chunk_id in cited_ids
            if chunk_id not in available_chunk_ids
        ]
        citation_status = (
            "CITATIONS_FOUND_IN_REVIEW_FILE"
            if cited_ids and not missing_ids
            else "CHECK_REQUIRED"
        )
    else:
        missing_ids = []
        citation_status = "NOT_INDEPENDENTLY_VERIFIED_BY_THIS_CELL"

    report_rows.append({
        "question_id": question_id,
        "draft_answer": row["draft_answer"],
        "supporting_chunk_ids": "; ".join(cited_ids),
        "evidence_review": row["evidence_review"],
        "citation_validation": citation_status,
        "missing_chunk_ids": "; ".join(missing_ids),
        "human_review_status": row["human_review_status"]
    })

report = pd.DataFrame(report_rows)

# Keep approval status conservative
human_status = (
    report["human_review_status"]
    .fillna("PENDING_HUMAN_REVIEW")
    .astype(str)
    .str.upper()
)

approved_count = human_status.isin(
    ["APPROVED", "HUMAN_APPROVED"]
).sum()

pending_count = len(report) - approved_count

summary = {
    "phase": 9,
    "phase_name": "Grounded Answer Generation and Review",
    "generated_at_utc": datetime.now(timezone.utc).isoformat(),
    "total_questions": int(len(report)),
    "human_approvals_completed": int(approved_count),
    "human_reviews_pending": int(pending_count),
    "preliminary_evidence_supported": int(
        report["evidence_review"].astype(str).str.contains(
            "SUPPORTED", case=False, na=False
        ).sum()
    ),
    "citation_review_artifact_available": os.path.exists(CITATION_PATH),
    "phase_status": (
        "AWAITING_HUMAN_REVIEW"
        if pending_count > 0
        else "HUMAN_REVIEW_RECORDED"
    ),
    "finalization_allowed": bool(pending_count == 0),
    "important_note": (
        "Citation presence and preliminary evidence support do not "
        "constitute human approval or prove that every answer is fully grounded."
    )
}

report.to_csv(REPORT_CSV, index=False)

with open(REPORT_JSON, "w", encoding="utf-8") as f:
    json.dump(summary, f, indent=2, ensure_ascii=False)

print("PHASE 9 REVIEW REPORT GENERATED")
print("-" * 45)
print("Questions:", summary["total_questions"])
print("Human approvals:", summary["human_approvals_completed"])
print("Human reviews pending:", summary["human_reviews_pending"])
print("Phase status:", summary["phase_status"])
print("\nCSV:", REPORT_CSV)
print("JSON:", REPORT_JSON)

display(report)
```

```python
import os
import pandas as pd

ROOT = "/kaggle/working/indian_legal_rag"
RESULTS_DIR = os.path.join(ROOT, "09_Results_Figures")

draft_path = os.path.join(
    RESULTS_DIR, "phase9_grounded_answer_drafts_v4.csv"
)

citation_path = os.path.join(
    RESULTS_DIR, "phase9_citation_evidence_review_v1.csv"
)

output_path = os.path.join(
    RESULTS_DIR, "phase9_human_review_worksheet_v1.csv"
)

drafts = pd.read_csv(draft_path)
citations = pd.read_csv(citation_path)

# Select relevant passage information where available
possible_columns = [
    "question_id", "question", "rank", "chunk_id",
    "source_file", "retrieved_text"
]
available_columns = [
    c for c in possible_columns if c in citations.columns
]

passages = citations[available_columns].copy()

if "chunk_id" in passages.columns:
    passages = passages.rename(columns={
        "chunk_id": "reviewed_chunk_id",
        "retrieved_text": "evidence_passage"
    })

# Keep one worksheet row per question
worksheet = drafts.copy()

# Attach cited evidence passages
if "reviewed_chunk_id" in passages.columns:
    evidence_lookup = passages.drop_duplicates(
        subset=["reviewed_chunk_id"]
    ).set_index("reviewed_chunk_id")

    worksheet["evidence_passage"] = worksheet[
        "supporting_chunk_ids"
    ].map(
        lambda ids: "\n\n".join(
            str(evidence_lookup.loc[cid, "evidence_passage"])
            for cid in str(ids).split(";")
            if cid.strip() in evidence_lookup.index
            and "evidence_passage" in evidence_lookup.columns
        )
    )

# Human-review fields: intentionally left unapproved
worksheet["answer_accuracy"] = "NOT_REVIEWED"
worksheet["evidence_sufficiency"] = "NOT_REVIEWED"
worksheet["citation_correct"] = "NOT_REVIEWED"
worksheet["reviewer_decision"] = "PENDING"
worksheet["reviewer_comments"] = ""
worksheet["reviewed_answer"] = worksheet["draft_answer"]

worksheet.to_csv(output_path, index=False)

print("HUMAN-REVIEW WORKSHEET CREATED")
print("Questions to review:", len(worksheet))
print("Output:", output_path)

display(worksheet[
    [
        "question_id",
        "draft_answer",
        "supporting_chunk_ids",
        "answer_accuracy",
        "evidence_sufficiency",
        "citation_correct",
        "reviewer_decision",
        "reviewer_comments"
    ]
])
```

```python
import os
import pandas as pd

RESULTS_DIR = (
    "/kaggle/working/indian_legal_rag/09_Results_Figures"
)

WORKSHEET_PATH = os.path.join(
    RESULTS_DIR, "phase9_human_review_worksheet_v1.csv"
)

# Verify required variables
if "chunks" not in globals():
    raise RuntimeError(
        "The chunks DataFrame is not loaded. "
        "Reload your existing Phase 5 chunks before continuing."
    )

worksheet = pd.read_csv(WORKSHEET_PATH)

required_chunk_columns = {"chunk_id", "text"}
if not required_chunk_columns.issubset(chunks.columns):
    raise ValueError(
        f"chunks must contain these columns: {required_chunk_columns}"
    )

# Create a lookup using the actual chunk text
chunk_lookup = (
    chunks.drop_duplicates("chunk_id")
    .set_index("chunk_id")["text"]
    .to_dict()
)

def retrieve_evidence(chunk_ids):
    ids = [
        cid.strip()
        for cid in str(chunk_ids).split(";")
        if cid.strip() and cid.strip().lower() != "nan"
    ]

    passages = []
    for cid in ids:
        if cid in chunk_lookup:
            passages.append(f"[{cid}]\n{chunk_lookup[cid]}")
        else:
            passages.append(
                f"[{cid}]\nERROR: Chunk ID not found in chunks DataFrame."
            )

    return "\n\n---\n\n".join(passages)

worksheet["evidence_passage"] = worksheet[
    "supporting_chunk_ids"
].apply(retrieve_evidence)

# Preserve all pending review decisions; do not approve automatically.
worksheet.to_csv(WORKSHEET_PATH, index=False)

print("EVIDENCE RESTORATION COMPLETE")
print("Questions:", len(worksheet))
print(
    "Missing evidence passages:",
    worksheet["evidence_passage"].str.contains(
        "ERROR: Chunk ID not found", na=False
    ).sum()
)
print("Updated worksheet:", WORKSHEET_PATH)

# Display actual evidence for inspection
for _, row in worksheet.iterrows():
    print("\n" + "=" * 80)
    print("QUESTION:", row["question_id"])
    print("ANSWER:", row["draft_answer"])
    print("CITATION:", row["supporting_chunk_ids"])
    print("EVIDENCE:\n", row["evidence_passage"])
```

```python
import os
import json
import pandas as pd
from datetime import datetime, timezone

ROOT = "/kaggle/working/indian_legal_rag"
RESULTS_DIR = os.path.join(ROOT, "09_Results_Figures")

DECISIONS_PATH = os.path.join(
    RESULTS_DIR, "phase9_reviewer_decisions_v2.csv"
)
CITATION_PATH = os.path.join(
    RESULTS_DIR, "phase9_citation_evidence_review_v1.csv"
)

REPORT_CSV = os.path.join(
    RESULTS_DIR, "phase9_final_answer_validation_v1.csv"
)
REPORT_JSON = os.path.join(
    RESULTS_DIR, "phase9_final_completion_report_v1.json"
)
FINAL_ANSWERS_PATH = os.path.join(
    RESULTS_DIR, "phase9_final_approved_answers_v1.csv"
)

# 1. Load recorded review decisions
if not os.path.exists(DECISIONS_PATH):
    raise FileNotFoundError(f"Decision file not found: {DECISIONS_PATH}")

df = pd.read_csv(DECISIONS_PATH)

required = [
    "question_id",
    "final_answer",
    "supporting_chunk_ids",
    "answer_accuracy",
    "evidence_sufficiency",
    "citation_correct",
    "reviewer_decision"
]
missing = [c for c in required if c not in df.columns]
if missing:
    raise ValueError(f"Missing required columns: {missing}")

# 2. Check decision completeness
df["reviewer_decision"] = (
    df["reviewer_decision"].fillna("").astype(str).str.upper().str.strip()
)
df["answer_accuracy"] = (
    df["answer_accuracy"].fillna("").astype(str).str.upper().str.strip()
)
df["evidence_sufficiency"] = (
    df["evidence_sufficiency"].fillna("").astype(str).str.upper().str.strip()
)
df["citation_correct"] = (
    df["citation_correct"].fillna("").astype(str).str.upper().str.strip()
)

df["answer_present"] = (
    df["final_answer"].fillna("").astype(str).str.strip().ne("")
)

df["citation_ids"] = df["supporting_chunk_ids"].fillna("").apply(
    lambda value: [
        item.strip()
        for item in str(value).split(";")
        if item.strip()
    ]
)
df["citation_present"] = df["citation_ids"].apply(bool)

# 3. Validate citation IDs against loaded chunks, when available
chunk_validation_available = (
    "chunks" in globals()
    and isinstance(chunks, pd.DataFrame)
    and "chunk_id" in chunks.columns
)

if chunk_validation_available:
    known_chunk_ids = set(chunks["chunk_id"].dropna().astype(str))
    df["missing_chunk_ids"] = df["citation_ids"].apply(
        lambda ids: "; ".join(
            cid for cid in ids if cid not in known_chunk_ids
        )
    )
    df["citations_exist"] = df["missing_chunk_ids"].eq("")
else:
    # Fall back to the Phase 9 citation-review artifact
    if os.path.exists(CITATION_PATH):
        citation_df = pd.read_csv(CITATION_PATH)
        if "chunk_id" in citation_df.columns:
            known_chunk_ids = set(
                citation_df["chunk_id"].dropna().astype(str)
            )
            df["missing_chunk_ids"] = df["citation_ids"].apply(
                lambda ids: "; ".join(
                    cid for cid in ids if cid not in known_chunk_ids
                )
            )
            df["citations_exist"] = df["missing_chunk_ids"].eq("")
            chunk_validation_available = True
        else:
            df["missing_chunk_ids"] = "Could not verify"
            df["citations_exist"] = False
    else:
        df["missing_chunk_ids"] = "Could not verify"
        df["citations_exist"] = False

# 4. Evaluate each answer
df["decision_complete"] = df["reviewer_decision"].isin(
    ["APPROVED", "REVISE"]
)
df["approval_checks_pass"] = (
    df["reviewer_decision"].eq("APPROVED")
    & df["answer_accuracy"].eq("CORRECT")
    & df["evidence_sufficiency"].eq("SUFFICIENT")
    & df["citation_correct"].eq("YES")
    & df["answer_present"]
    & df["citation_present"]
    & df["citations_exist"]
)

report_columns = [
    "question_id",
    "final_answer",
    "supporting_chunk_ids",
    "answer_accuracy",
    "evidence_sufficiency",
    "citation_correct",
    "reviewer_decision",
    "answer_present",
    "citation_present",
    "citations_exist",
    "missing_chunk_ids",
    "approval_checks_pass"
]
report = df[report_columns].copy()
report.to_csv(REPORT_CSV, index=False)

# 5. Determine final status conservatively
expected_questions = 10
unique_questions = int(df["question_id"].nunique())
approved_count = int(df["reviewer_decision"].eq("APPROVED").sum())
revision_count = int(df["reviewer_decision"].eq("REVISE").sum())
checks_passed = int(df["approval_checks_pass"].sum())

complete = (
    len(df) == expected_questions
    and unique_questions == expected_questions
    and approved_count == expected_questions
    and revision_count == 0
    and checks_passed == expected_questions
)

summary = {
    "phase": 9,
    "phase_name": "Grounded Answer Generation and Review",
    "generated_at_utc": datetime.now(timezone.utc).isoformat(),
    "decision_source": os.path.basename(DECISIONS_PATH),
    "total_rows": int(len(df)),
    "unique_questions": unique_questions,
    "approved_answers": approved_count,
    "answers_requiring_revision": revision_count,
    "answers_passing_all_validation_checks": checks_passed,
    "citation_chunk_existence_verified": bool(chunk_validation_available),
    "phase_status": (
        "COMPLETE"
        if complete
        else "VALIDATION_INCOMPLETE"
    ),
    "finalization_allowed": bool(complete),
    "limitations": [
        "This report validates recorded review fields and citation ID existence.",
        "Citation ID existence alone does not prove that a passage entails the answer.",
        "The decision records reflect the decisions entered into the notebook; this cell does not independently perform legal review."
    ]
}

with open(REPORT_JSON, "w", encoding="utf-8") as f:
    json.dump(summary, f, indent=2, ensure_ascii=False)

# Export final answers only when all validation checks pass
if complete:
    df[[
        "question_id",
        "final_answer",
        "supporting_chunk_ids"
    ]].to_csv(FINAL_ANSWERS_PATH, index=False)

print("PHASE 9 FINAL VALIDATION")
print("-" * 45)
print("Questions:", len(df))
print("Unique questions:", unique_questions)
print("Recorded approvals:", approved_count)
print("Answers needing revision:", revision_count)
print("Answers passing all checks:", checks_passed)
print("Citation IDs checked:", chunk_validation_available)
print("Phase 9 status:", summary["phase_status"])
print("\nValidation CSV:", REPORT_CSV)
print("Completion report:", REPORT_JSON)

if complete:
    print("Final answers:", FINAL_ANSWERS_PATH)
else:
    print("\nReview failed or incomplete checks before finalizing.")

display(report)
```

```python
import os
import pandas as pd

ROOT = "/kaggle/working/indian_legal_rag"
RESULTS_DIR = os.path.join(ROOT, "09_Results_Figures")

APPROVED_PATH = os.path.join(
    RESULTS_DIR, "phase9_final_approved_answers_v1.csv"
)

BENCHMARK_PATH = os.path.join(
    ROOT, "08_Evaluation",
    "phase4_qa_benchmark_v1_RECONSTRUCTED.csv"
)

OUTPUT_PATH = os.path.join(
    RESULTS_DIR, "phase10_evaluation_dataset_v1.csv"
)

# Load source files
approved = pd.read_csv(APPROVED_PATH)
benchmark = pd.read_csv(BENCHMARK_PATH)

# Required fields
required_approved = {
    "question_id", "final_answer", "supporting_chunk_ids"
}
required_benchmark = {
    "question_id", "question", "reference_answer"
}

assert required_approved.issubset(approved.columns), (
    f"Approved-answer file missing: "
    f"{required_approved - set(approved.columns)}"
)
assert required_benchmark.issubset(benchmark.columns), (
    f"Benchmark missing: "
    f"{required_benchmark - set(benchmark.columns)}"
)

# Check unique IDs before merging
assert approved["question_id"].is_unique, (
    "Approved-answer file contains duplicate question IDs."
)
assert benchmark["question_id"].is_unique, (
    "Benchmark contains duplicate question IDs."
)

# Keep benchmark fields and attach approved answers
evaluation = benchmark.merge(
    approved[[
        "question_id",
        "final_answer",
        "supporting_chunk_ids"
    ]],
    on="question_id",
    how="left",
    validate="one_to_one",
    indicator=True
)

# Validate matching and required text
unmatched = evaluation[evaluation["_merge"] != "both"]
if not unmatched.empty:
    raise ValueError(
        "Some benchmark questions have no approved answer:\n"
        + unmatched[["question_id", "_merge"]].to_string(index=False)
    )

evaluation.drop(columns=["_merge"], inplace=True)

for col in ["question", "reference_answer", "final_answer"]:
    evaluation[col] = evaluation[col].fillna("").astype(str).str.strip()

empty_rows = evaluation[
    (evaluation["question"] == "")
    | (evaluation["reference_answer"] == "")
    | (evaluation["final_answer"] == "")
]

if not empty_rows.empty:
    raise ValueError(
        "Missing question or answer text for:\n"
        + empty_rows[["question_id"]].to_string(index=False)
    )

# Save the prepared dataset
evaluation.to_csv(OUTPUT_PATH, index=False)

print("=" * 60)
print("PHASE 10.2 — EVALUATION DATASET PREPARED")
print("=" * 60)
print("Benchmark questions:", len(benchmark))
print("Approved answers:", len(approved))
print("Matched evaluation rows:", len(evaluation))
print("Unmatched questions:", len(unmatched))
print("Rows with empty required text:", len(empty_rows))
print("Unique question IDs:", evaluation["question_id"].nunique())
print("Saved to:", OUTPUT_PATH)

display(evaluation[[
    "question_id",
    "question",
    "reference_answer",
    "final_answer",
    "supporting_chunk_ids"
]])
```

```python
import os
import re
import numpy as np
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

ROOT = "/kaggle/working/indian_legal_rag"
RESULTS_DIR = os.path.join(ROOT, "09_Results_Figures")

INPUT_PATH = os.path.join(
    RESULTS_DIR, "phase10_evaluation_dataset_v1.csv"
)

OUTPUT_PATH = os.path.join(
    RESULTS_DIR, "phase10_answer_quality_review_v1.csv"
)

SUMMARY_PATH = os.path.join(
    RESULTS_DIR, "phase10_answer_quality_summary_v1.csv"
)

# Verify existing resources
if "chunks" not in globals() or "model" not in globals():
    raise RuntimeError(
        "Required resources are missing. "
        "Ensure chunks and model are loaded in this notebook session."
    )

df = pd.read_csv(INPUT_PATH)

required_columns = [
    "question_id",
    "question",
    "reference_answer",
    "final_answer",
    "supporting_chunk_ids"
]

missing = [c for c in required_columns if c not in df.columns]
if missing:
    raise ValueError(f"Missing columns: {missing}")

# Build lookup from the existing corpus
chunk_lookup = (
    chunks.drop_duplicates("chunk_id")
    .set_index("chunk_id")["text"]
    .to_dict()
)

# Extract cited passages and validate citation IDs
evidence_passages = []
citation_exists = []
missing_citations = []

for cited_ids in df["supporting_chunk_ids"].fillna(""):
    ids = [
        cid.strip()
        for cid in str(cited_ids).split(";")
        if cid.strip()
    ]

    missing_ids = [cid for cid in ids if cid not in chunk_lookup]

    passages = [
        f"[{cid}]\n{chunk_lookup[cid]}"
        for cid in ids if cid in chunk_lookup
    ]

    evidence_passages.append("\n\n---\n\n".join(passages))
    citation_exists.append(bool(ids) and len(missing_ids) == 0)
    missing_citations.append("; ".join(missing_ids))

df["cited_evidence"] = evidence_passages
df["citation_exists"] = citation_exists
df["missing_citation_ids"] = missing_citations

# Calculate semantic similarity between reference and final answers
references = df["reference_answer"].fillna("").astype(str).tolist()
answers = df["final_answer"].fillna("").astype(str).tolist()

reference_vectors = model.encode(
    references,
    convert_to_numpy=True,
    normalize_embeddings=True,
    show_progress_bar=False
)

answer_vectors = model.encode(
    answers,
    convert_to_numpy=True,
    normalize_embeddings=True,
    show_progress_bar=False
)

similarities = np.sum(reference_vectors * answer_vectors, axis=1)

df["semantic_similarity"] = np.round(similarities, 4)

# Similarity is a diagnostic, not an accuracy score.
# Evidence grounding requires checking the actual passage.
df["evidence_grounding_review"] = "PENDING_PASSAGE_REVIEW"

# Save detailed results
df.to_csv(OUTPUT_PATH, index=False)

# Produce a compact summary
summary = pd.DataFrame([{
    "questions_evaluated": len(df),
    "citations_present_and_found": int(df["citation_exists"].sum()),
    "citations_missing_or_invalid": int((~df["citation_exists"]).sum()),
    "mean_semantic_similarity": round(
        float(df["semantic_similarity"].mean()), 4
    ),
    "median_semantic_similarity": round(
        float(df["semantic_similarity"].median()), 4
    ),
    "answers_requiring_evidence_review": len(df),
    "evidence_grounding_status": "MANUAL_REVIEW_REQUIRED"
}])

summary.to_csv(SUMMARY_PATH, index=False)

print("=" * 65)
print("PHASE 10.3 — ANSWER QUALITY REVIEW CREATED")
print("=" * 65)
print("Questions evaluated:", len(df))
print("Citations found:", int(df["citation_exists"].sum()))
print("Missing/invalid citations:", int((~df["citation_exists"]).sum()))
print(
    "Mean semantic similarity:",
    round(float(df["semantic_similarity"].mean()), 4)
)
print(
    "Median semantic similarity:",
    round(float(df["semantic_similarity"].median()), 4)
)
print("\nDetailed review:", OUTPUT_PATH)
print("Summary:", SUMMARY_PATH)

display(df[[
    "question_id",
    "reference_answer",
    "final_answer",
    "supporting_chunk_ids",
    "citation_exists",
    "semantic_similarity",
    "evidence_grounding_review"
]])

display(summary)
```

```python
import os
import pandas as pd

ROOT = "/kaggle/working/indian_legal_rag"
RESULTS_DIR = os.path.join(ROOT, "09_Results_Figures")

INPUT_PATH = os.path.join(
    RESULTS_DIR, "phase10_answer_quality_review_v1.csv"
)

OUTPUT_PATH = os.path.join(
    RESULTS_DIR, "phase10_evidence_grounding_review_v1.csv"
)

df = pd.read_csv(INPUT_PATH)

# Retrieve actual evidence from the existing corpus
chunk_lookup = (
    chunks.drop_duplicates("chunk_id")
    .set_index("chunk_id")["text"]
    .to_dict()
)

rows = []

for _, row in df.iterrows():
    ids = [
        cid.strip()
        for cid in str(row["supporting_chunk_ids"]).split(";")
        if cid.strip() and cid.strip().lower() != "nan"
    ]

    passages = [
        chunk_lookup[cid]
        for cid in ids
        if cid in chunk_lookup
    ]

    rows.append({
        "question_id": row["question_id"],
        "question": row["question"],
        "final_answer": row["final_answer"],
        "supporting_chunk_ids": row["supporting_chunk_ids"],
        "cited_evidence": "\n\n---\n\n".join(passages),
        "citation_ids_found": bool(ids) and len(passages) == len(ids),
        "semantic_similarity": row["semantic_similarity"],
        "grounding_decision": "PENDING_MANUAL_REVIEW",
        "unsupported_claims": "",
        "reviewer_comments": ""
    })

review = pd.DataFrame(rows)
review.to_csv(OUTPUT_PATH, index=False)

print("PHASE 10.4 — EVIDENCE REVIEW FILE CREATED")
print("Questions:", len(review))
print("Questions with all citation passages:", int(
    review["citation_ids_found"].sum()
))
print("Grounding decisions completed:", 0)
print("Output:", OUTPUT_PATH)

display(review[[
    "question_id",
    "final_answer",
    "supporting_chunk_ids",
    "citation_ids_found",
    "grounding_decision"
]])
```

```python
import os
import json
import pandas as pd
from datetime import datetime, timezone

RESULTS_DIR = (
    "/kaggle/working/indian_legal_rag/09_Results_Figures"
)

INPUT_PATH = os.path.join(
    RESULTS_DIR, "phase10_grounding_assessment_proposals_v1.csv"
)

OUTPUT_PATH = os.path.join(
    RESULTS_DIR, "phase10_grounding_assessment_final_v1.csv"
)

REPORT_PATH = os.path.join(
    RESULTS_DIR, "phase10_grounding_completion_report_v1.json"
)

df = pd.read_csv(INPUT_PATH)

# Only record these decisions if you agree with the proposals.
assert len(df) == 10, "Expected 10 benchmark questions."
assert df["question_id"].nunique() == 10, "Duplicate question IDs found."

allowed = {"SUPPORTED", "PARTIALLY_SUPPORTED", "UNSUPPORTED"}

df["final_grounding_decision"] = (
    df["proposed_grounding_decision"].fillna("").str.upper().str.strip()
)

if not set(df["final_grounding_decision"]).issubset(allowed):
    raise ValueError("Unexpected grounding decision found.")

if (df["final_grounding_decision"] == "").any():
    raise ValueError("Some grounding decisions are blank.")

# Preserve the source of the assessment.
df["assessment_basis"] = "Proposed assessment confirmed by project reviewer"
df["review_confirmed"] = True

df.to_csv(OUTPUT_PATH, index=False)

counts = df["final_grounding_decision"].value_counts().to_dict()

summary = {
    "phase": "10.4",
    "task": "Evidence Grounding Evaluation",
    "generated_at_utc": datetime.now(timezone.utc).isoformat(),
    "questions_reviewed": int(len(df)),
    "grounding_decisions": {
        decision: int(counts.get(decision, 0))
        for decision in sorted(allowed)
    },
    "grounding_coverage": round(
        float(
            (df["final_grounding_decision"] == "SUPPORTED").mean()
        ), 4
    ),
    "status": "ASSESSMENTS_RECORDED",
    "note": (
        "Grounding coverage reflects recorded passage assessments "
        "and is not an independent legal audit."
    )
}

with open(REPORT_PATH, "w", encoding="utf-8") as f:
    json.dump(summary, f, indent=2)

print("PHASE 10.4 ASSESSMENT REPORT")
print("-" * 45)
print("Questions:", summary["questions_reviewed"])
print("Supported:", counts.get("SUPPORTED", 0))
print("Partially supported:", counts.get("PARTIALLY_SUPPORTED", 0))
print("Unsupported:", counts.get("UNSUPPORTED", 0))
print("Recorded grounding coverage:", summary["grounding_coverage"])
print("Output CSV:", OUTPUT_PATH)
print("Report JSON:", REPORT_PATH)

display(df[[
    "question_id",
    "supporting_chunk_ids",
    "final_grounding_decision",
    "assessment_rationale"
]])
```

```python
import os
import json
import pandas as pd
from datetime import datetime, timezone

ROOT = "/kaggle/working/indian_legal_rag"
RESULTS_DIR = os.path.join(ROOT, "09_Results_Figures")
EVAL_DIR = os.path.join(ROOT, "08_Evaluation")

# Input artifacts
answer_path = os.path.join(
    RESULTS_DIR, "phase10_answer_quality_review_v1.csv"
)
grounding_path = os.path.join(
    RESULTS_DIR, "phase10_grounding_assessment_final_v1.csv"
)
retrieval_path = os.path.join(
    EVAL_DIR, "phase7_retrieval_metrics_summary_v2_corrected.csv"
)

# Output artifacts
output_csv = os.path.join(
    RESULTS_DIR, "phase10_consolidated_evaluation_v1.csv"
)
output_json = os.path.join(
    RESULTS_DIR, "phase10_consolidated_evaluation_summary_v1.json"
)

# Verify inputs
for path in [answer_path, grounding_path, retrieval_path]:
    print(("FOUND: " if os.path.exists(path) else "MISSING: ") + path)

if not os.path.exists(answer_path):
    raise FileNotFoundError(answer_path)

if not os.path.exists(grounding_path):
    raise FileNotFoundError(grounding_path)

answers = pd.read_csv(answer_path)
grounding = pd.read_csv(grounding_path)

# Merge answer-quality and grounding results
grounding_columns = [
    "question_id",
    "final_grounding_decision"
]
combined = answers.merge(
    grounding[grounding_columns],
    on="question_id",
    how="left",
    validate="one_to_one"
)

combined["grounding_pass"] = (
    combined["final_grounding_decision"] == "SUPPORTED"
)
combined["citation_pass"] = combined["citation_exists"].fillna(False).astype(bool)

# Keep similarity as a separate diagnostic measure
combined["semantic_similarity"] = pd.to_numeric(
    combined["semantic_similarity"], errors="coerce"
)

combined.to_csv(output_csv, index=False)

# Summary metrics
n = len(combined)

summary = {
    "phase": "10.5",
    "title": "End-to-End Metrics and Failure Analysis",
    "generated_at_utc": datetime.now(timezone.utc).isoformat(),
    "questions_evaluated": int(n),
    "mean_semantic_similarity": round(
        float(combined["semantic_similarity"].mean()), 4
    ),
    "median_semantic_similarity": round(
        float(combined["semantic_similarity"].median()), 4
    ),
    "citation_validity_rate": round(
        float(combined["citation_pass"].mean()), 4
    ),
    "recorded_grounding_coverage": round(
        float(combined["grounding_pass"].mean()), 4
    ),
    "grounding_assessments": combined[
        "final_grounding_decision"
    ].value_counts(dropna=False).to_dict(),
    "interpretation_note": (
        "Semantic similarity is not factual accuracy. Citation validity "
        "checks citation presence/existence. Grounding coverage reflects "
        "recorded assessments and should not be treated as independent "
        "verification of legal correctness."
    )
}

# Read retrieval metrics if available
if os.path.exists(retrieval_path):
    retrieval_metrics = pd.read_csv(retrieval_path)
    summary["retrieval_metrics_file"] = retrieval_path
    summary["retrieval_metrics_rows"] = int(len(retrieval_metrics))
    print("\nPhase 7 retrieval metrics found:")
    display(retrieval_metrics)
else:
    summary["retrieval_metrics_file"] = None
    print("\nPhase 7 metrics summary not found at the expected path.")
    print("We will locate the existing file before combining retrieval metrics.")

with open(output_json, "w", encoding="utf-8") as f:
    json.dump(summary, f, indent=2, ensure_ascii=False)

print("\nPHASE 10.5 CONSOLIDATED REPORT CREATED")
print("-" * 50)
print("Questions evaluated:", n)
print("Mean semantic similarity:", summary["mean_semantic_similarity"])
print("Median semantic similarity:", summary["median_semantic_similarity"])
print("Citation validity rate:", summary["citation_validity_rate"])
print("Recorded grounding coverage:", summary["recorded_grounding_coverage"])
print("CSV:", output_csv)
print("JSON:", output_json)

display(combined[[
    "question_id",
    "semantic_similarity",
    "citation_exists",
    "final_grounding_decision",
    "grounding_pass"
]])
```

```python
import os

ROOT = "/kaggle/working/indian_legal_rag"

print("SEARCHING FOR EXISTING PHASE 7 METRICS")
print("-" * 60)

matches = []

for folder, subfolders, filenames in os.walk(ROOT):
    # Skip unnecessary hidden directories
    subfolders[:] = [
        d for d in subfolders if not d.startswith(".")
    ]

    for filename in filenames:
        lower_name = filename.lower()

        if (
            filename.endswith((".csv", ".json"))
            and "phase7" in lower_name
            and any(word in lower_name for word in [
                "metric", "summary", "evaluation", "manifest"
            ])
        ):
            matches.append(os.path.join(folder, filename))

if matches:
    for path in sorted(matches):
        print(path)
else:
    print("No matching Phase 7 metrics files found.")
```

```python
import os
import json
import pandas as pd
from datetime import datetime, timezone

ROOT = "/kaggle/working/indian_legal_rag"
RESULTS_DIR = os.path.join(ROOT, "09_Results_Figures")

# Existing files
retrieval_path = os.path.join(
    RESULTS_DIR, "phase7_retrieval_metrics_summary_v2_corrected.csv"
)
answer_path = os.path.join(
    RESULTS_DIR, "phase10_consolidated_evaluation_v1.csv"
)

# New final outputs
final_csv = os.path.join(
    RESULTS_DIR, "phase10_final_evaluation_v1.csv"
)
final_json = os.path.join(
    RESULTS_DIR, "phase10_final_evaluation_report_v1.json"
)

# Validate input files
for path in [retrieval_path, answer_path]:
    if not os.path.exists(path):
        raise FileNotFoundError(f"Required file not found: {path}")

# Load existing evaluation results
retrieval = pd.read_csv(retrieval_path)
answers = pd.read_csv(answer_path)

print("PHASE 7 RETRIEVAL METRICS")
print("-" * 60)
display(retrieval)

print("\nPHASE 10 ANSWER QUALITY")
print("-" * 60)
display(answers[[
    "question_id",
    "semantic_similarity",
    "citation_exists",
    "final_grounding_decision"
]])

# Identify answers requiring closer review.
# A similarity score is a diagnostic, not an accuracy score.
answers["review_flag"] = (
    pd.to_numeric(answers["semantic_similarity"], errors="coerce") < 0.80
)

answers.to_csv(final_csv, index=False)

# Preserve retrieval results as records without assuming column names
retrieval_records = retrieval.astype(object).where(
    pd.notna(retrieval), None
).to_dict(orient="records")

# Build the consolidated report
similarity = pd.to_numeric(
    answers["semantic_similarity"], errors="coerce"
)

summary = {
    "phase": "10",
    "status": "EVALUATION_REPORT_GENERATED",
    "generated_at_utc": datetime.now(timezone.utc).isoformat(),
    "questions_evaluated": int(len(answers)),
    "mean_semantic_similarity": round(float(similarity.mean()), 4),
    "median_semantic_similarity": round(float(similarity.median()), 4),
    "citation_validity_rate": round(
        float(answers["citation_pass"].mean()), 4
    ),
    "recorded_grounding_coverage": round(
        float(answers["grounding_pass"].mean()), 4
    ),
    "answers_flagged_for_similarity_review": answers.loc[
        answers["review_flag"], "question_id"
    ].tolist(),
    "phase7_retrieval_metrics_source": retrieval_path,
    "phase7_retrieval_metrics": retrieval_records,
    "phase10_answer_results_source": answer_path,
    "interpretation": (
        "Semantic similarity is not factual accuracy. Citation validity "
        "checks citation existence, not complete claim support. Grounding "
        "coverage reflects recorded assessments and is not an independent "
        "legal audit. Similarity flags identify candidates for review."
    ),
    "remaining_quality_checks": [
        "Review lower-similarity answers against reference answers.",
        "Verify cited passages support the full legal claims.",
        "Complete the pending legal review of reconstructed Phase 3–4 artifacts."
    ]
}

with open(final_json, "w", encoding="utf-8") as f:
    json.dump(summary, f, indent=2, ensure_ascii=False)

print("\nFINAL EVALUATION REPORT CREATED")
print("=" * 60)
print("Questions evaluated:", summary["questions_evaluated"])
print("Mean similarity:", summary["mean_semantic_similarity"])
print("Citation validity:", summary["citation_validity_rate"])
print("Recorded grounding coverage:", summary["recorded_grounding_coverage"])
print("Answers flagged for review:",
      summary["answers_flagged_for_similarity_review"])
print("\nCSV:", final_csv)
print("JSON:", final_json)
```

```python
import os
import json
import pandas as pd

ROOT = "/kaggle/working/indian_legal_rag"

required_files = {
    "Phase 3 reconstructed gold evidence":
        "05_Gold_Evidence/phase3_gold_evidence_pilot_RECONSTRUCTED.csv",

    "Phase 4 reconstructed benchmark":
        "08_Evaluation/phase4_qa_benchmark_v1_RECONSTRUCTED.csv",

    "Phase 7 retrieval metrics":
        "09_Results_Figures/phase7_retrieval_metrics_summary_v2_corrected.csv",

    "Phase 8 expansion findings":
        "09_Results_Figures/phase8_query_expansion_findings_v1.json",

    "Phase 9 approved answers":
        "09_Results_Figures/phase9_final_approved_answers_v1.csv",

    "Phase 9 validation":
        "09_Results_Figures/phase9_final_answer_validation_v1.csv",

    "Phase 10 evaluation dataset":
        "09_Results_Figures/phase10_evaluation_dataset_v1.csv",

    "Phase 10 consolidated evaluation":
        "09_Results_Figures/phase10_final_evaluation_v1.csv",

    "Phase 10 failure analysis":
        "09_Results_Figures/phase10_failure_analysis_v1.csv",

    "Phase 10 evaluation report":
        "09_Results_Figures/phase10_final_evaluation_report_v1.json",
}

audit = []

for name, relative_path in required_files.items():
    full_path = os.path.join(ROOT, relative_path)
    exists = os.path.isfile(full_path)

    row_count = None
    if exists and full_path.endswith(".csv"):
        try:
            row_count = len(pd.read_csv(full_path))
        except Exception as e:
            row_count = f"Read error: {e}"

    audit.append({
        "artifact": name,
        "exists": exists,
        "rows": row_count,
        "path": full_path
    })

audit_df = pd.DataFrame(audit)

print("FINAL PROJECT ARTIFACT AUDIT")
print("=" * 70)
display(audit_df)

print("\nExisting artifacts:", int(audit_df["exists"].sum()))
print("Missing artifacts:", int((~audit_df["exists"]).sum()))

missing = audit_df.loc[~audit_df["exists"], "artifact"].tolist()

if missing:
    print("\nMissing artifacts to investigate:")
    for item in missing:
        print("-", item)
else:
    print("\nAll listed artifacts exist.")

print(
    "\nImportant: File existence does not establish legal correctness, "
    "experimental reproducibility, or completion of pending manual reviews."
)
```

```python
import os
import pandas as pd

ROOT = "/kaggle/working/indian_legal_rag"

gold_path = os.path.join(
    ROOT, "05_Gold_Evidence",
    "phase3_gold_evidence_pilot_RECONSTRUCTED.csv"
)

benchmark_path = os.path.join(
    ROOT, "08_Evaluation",
    "phase4_qa_benchmark_v1_RECONSTRUCTED.csv"
)

output_path = os.path.join(
    ROOT, "09_Results_Figures",
    "phase4_benchmark_consistency_diagnostics_v1.csv"
)

gold = pd.read_csv(gold_path).fillna("")
benchmark = pd.read_csv(benchmark_path).fillna("")

fields = [
    "question",
    "reference_answer",
    "evidence_document",
    "case_name",
    "court",
    "judgment_date",
    "evidence_passage"
]

gold_by_id = gold.set_index("question_id")
benchmark_by_id = benchmark.set_index("question_id")

results = []

for qid in sorted(set(gold_by_id.index) | set(benchmark_by_id.index)):
    row = {
        "question_id": qid,
        "in_gold": qid in gold_by_id.index,
        "in_benchmark": qid in benchmark_by_id.index
    }

    if qid in gold_by_id.index and qid in benchmark_by_id.index:
        g = gold_by_id.loc[qid]
        b = benchmark_by_id.loc[qid]

        for field in fields:
            row[f"{field}_matches"] = (
                str(g[field]).strip() == str(b[field]).strip()
                if field in gold.columns and field in benchmark.columns
                else None
            )

        passage = str(b.get("evidence_passage", ""))
        lower_passage = passage.lower()

        suspicious_terms = []

        if len(passage) < 200:
            suspicious_terms.append("SHORT_EVIDENCE_PASSAGE")

        if "anokhilal" in lower_passage and qid != "CRIM_007":
            suspicious_terms.append("POSSIBLE_ANOKHILAL_CROSS_CONTAMINATION")

        if qid == "CRIM_006" and "sushila aggarwal" not in lower_passage:
            suspicious_terms.append("EXPECTED_CASE_NAME_NOT_FOUND")

        if qid == "CRIM_009" and (
            passage.lstrip().startswith("ccused")
            or "time mac" in lower_passage
        ):
            suspicious_terms.append("POSSIBLE_TRUNCATION")

        row["evidence_length_chars"] = len(passage)
        row["diagnostic_flags"] = (
            "; ".join(suspicious_terms) if suspicious_terms else "NONE_DETECTED"
        )
    else:
        row["diagnostic_flags"] = "QUESTION_ID_MISSING_FROM_ONE_DATASET"

    results.append(row)

diagnostics = pd.DataFrame(results)
diagnostics.to_csv(output_path, index=False)

print("PHASE 3–4 CONSISTENCY DIAGNOSTICS")
print("=" * 65)

comparison_cols = [
    "question_id",
    "in_gold",
    "in_benchmark",
    "evidence_passage_matches",
    "evidence_length_chars",
    "diagnostic_flags"
]

display(
    diagnostics[
        [c for c in comparison_cols if c in diagnostics.columns]
    ]
)

print("\nDiagnostic report saved to:")
print(output_path)

print(
    "\nNote: Matching passages do not establish legal correctness. "
    "Flags are indicators for manual investigation, not definitive errors."
)
```

```python
import pandas as pd
from pathlib import Path

ROOT = Path("/kaggle/working/indian_legal_rag")
RESULTS = ROOT / "09_Results_Figures"
RESULTS.mkdir(parents=True, exist_ok=True)

# Reuse existing chunks
assert "chunks" in globals(), "The chunks DataFrame is not loaded."

expected_docs = {
    "CRIM_003": "doc_4938",
    "CRIM_006": "doc_5970",
    "CRIM_007": "doc_3994",
    "CRIM_009": "doc_5673",
}

# Focused search terms within the correct judgment
terms = {
    "CRIM_003": "Circumstances leading to alibi are within his knowledge",
    "CRIM_006": "the order granting anticipatory bail once granted ordinarily subsists",
    "CRIM_007": "The right to free legal services is, therefore, clearly an essential ingredient",
    "CRIM_009": "Therefore, free and fair trial is a sine qua non of Article 21",
}

proposals = []

for qid, doc_id in expected_docs.items():
    doc_chunks = chunks[
        chunks["document_id"].astype(str) == doc_id
    ].copy()

    term = terms[qid]
    matches = doc_chunks[
        doc_chunks["text"].astype(str).str.contains(
            term, case=False, regex=False, na=False
        )
    ]

    print("\n", "=" * 80)
    print(qid, "| Expected source:", doc_id)
    print("Matching chunks:", len(matches))

    if matches.empty:
        print("No exact match. Needs manual review.")
        proposals.append({
            "question_id": qid,
            "expected_document_id": doc_id,
            "proposed_chunk_ids": "",
            "proposed_evidence_passage": "",
            "review_status": "PENDING_NO_MATCH",
        })
        continue

    # Prefer the first matching chunk and, where possible,
    # include the next chunk for surrounding context.
    row = matches.iloc[0]
    chunk_id = str(row["chunk_id"])
    chunk_index = int(row["chunk_index"])

    context = doc_chunks[
        doc_chunks["chunk_index"].between(chunk_index, chunk_index + 1)
    ].sort_values("chunk_index")

    passage = "\n\n".join(context["text"].astype(str).tolist())

    print("Selected chunk:", chunk_id)
    print("Proposed passage length:", len(passage))
    print(passage[:2500])

    proposals.append({
        "question_id": qid,
        "expected_document_id": doc_id,
        "proposed_chunk_ids": ";".join(context["chunk_id"].astype(str)),
        "proposed_evidence_passage": passage,
        "review_status": "PENDING_MANUAL_REVIEW",
    })

proposal_df = pd.DataFrame(proposals)

out_path = RESULTS / "phase4_evidence_correction_proposals_v1.csv"
proposal_df.to_csv(out_path, index=False)

print("\nSaved proposed corrections to:", out_path)
print(proposal_df[
    ["question_id", "expected_document_id", "proposed_chunk_ids", "review_status"]
].to_string(index=False))
```

```python
import pandas as pd
import json
from pathlib import Path
from datetime import datetime, timezone

ROOT = Path("/kaggle/working/indian_legal_rag")
RESULTS = ROOT / "09_Results_Figures"
RESULTS.mkdir(parents=True, exist_ok=True)

assert "chunks" in globals(), "The existing chunks DataFrame is not loaded."

# Retrieve one exact chunk by ID
def get_chunk(chunk_id):
    rows = chunks[chunks["chunk_id"].astype(str) == chunk_id]
    if rows.empty:
        raise ValueError(f"Missing required chunk: {chunk_id}")
    return str(rows.iloc[0]["text"])

# 1. CRIM_003: Focus on the alibi rule
text_003 = get_chunk("doc_4938_v2_chunk_0058")
start = text_003.find("At this stage we have to point out")
end_phrase = "he has to establish the same satisfactorily."
end = text_003.find(end_phrase, start)

if start < 0 or end < 0:
    raise ValueError("Could not extract CRIM_003 passage safely.")

passage_003 = text_003[start:end + len(end_phrase)]

# 2. CRIM_006: Extract the final court conclusions
text_006a = get_chunk("doc_5970_v2_chunk_0176")
text_006b = get_chunk("doc_5970_v2_chunk_0177")

marker = "FINAL CONCLUSIONS:"
start = text_006a.find(marker)

if start >= 0:
    passage_006 = text_006a[start:]
else:
    # Fallback: use the final ruling across the two chunks
    combined = text_006a + "\n" + text_006b
    start = combined.find("Regarding Question No.")
    if start < 0:
        raise ValueError("Could not locate CRIM_006 final conclusion.")
    passage_006 = combined[start:]

# 3. CRIM_007: Extract the free legal services principle
text_007 = get_chunk("doc_3994_v2_chunk_0022")
start_phrase = "The right to free legal services is"
end_phrase = "does not object to the provision of such lawyer."
start = text_007.find(start_phrase)
end = text_007.find(end_phrase, start)

if start < 0 or end < 0:
    raise ValueError("Could not extract CRIM_007 passage safely.")

passage_007 = text_007[start:end + len(end_phrase)]

# 4. CRIM_009: Extract the fair-trial principle
text_009 = get_chunk("doc_5673_v2_chunk_0032")
start_phrase = "Therefore, free and fair trial is a sine qua non"
end_phrase = "Article 14 of the Constitution."
start = text_009.find(start_phrase)
end = text_009.find(end_phrase, start)

if start < 0 or end < 0:
    raise ValueError("Could not extract CRIM_009 passage safely.")

passage_009 = text_009[start:end + len(end_phrase)]

corrections = {
    "CRIM_003": {
        "document_id": "doc_4938",
        "chunk_ids": "doc_4938_v2_chunk_0058",
        "passage": passage_003,
    },
    "CRIM_006": {
        "document_id": "doc_5970",
        "chunk_ids": "doc_5970_v2_chunk_0176;doc_5970_v2_chunk_0177",
        "passage": passage_006,
    },
    "CRIM_007": {
        "document_id": "doc_3994",
        "chunk_ids": "doc_3994_v2_chunk_0022",
        "passage": passage_007,
    },
    "CRIM_009": {
        "document_id": "doc_5673",
        "chunk_ids": "doc_5673_v2_chunk_0032",
        "passage": passage_009,
    },
}

# Save the correction proposals
proposal_rows = []
for qid, item in corrections.items():
    proposal_rows.append({
        "question_id": qid,
        "expected_document_id": item["document_id"],
        "source_chunk_ids": item["chunk_ids"],
        "proposed_evidence_passage": item["passage"],
        "character_count": len(item["passage"]),
        "review_status": "PENDING_FINAL_REVIEW",
    })

proposals = pd.DataFrame(proposal_rows)
proposal_path = RESULTS / "phase4_evidence_correction_proposals_v2.csv"
proposals.to_csv(proposal_path, index=False)

# Load the existing reconstructed benchmark
benchmark_path = (
    ROOT / "08_Evaluation" /
    "phase4_qa_benchmark_v1_RECONSTRUCTED.csv"
)

benchmark = pd.read_csv(benchmark_path)

if "question_id" not in benchmark.columns:
    raise ValueError(f"Benchmark has no question_id column: {list(benchmark.columns)}")

if "evidence_passage" not in benchmark.columns:
    raise ValueError(
        "Expected evidence_passage column was not found. "
        f"Actual columns: {list(benchmark.columns)}"
    )

corrected = benchmark.copy()
updated_ids = []

for qid, item in corrections.items():
    mask = corrected["question_id"].astype(str) == qid

    if mask.sum() != 1:
        raise ValueError(
            f"{qid}: expected exactly one benchmark row, found {mask.sum()}"
        )

    corrected.loc[mask, "evidence_passage"] = item["passage"]
    updated_ids.append(qid)

# Save as a new version; preserve the original
corrected_path = (
    ROOT / "08_Evaluation" /
    "phase4_qa_benchmark_v2_evidence_corrected_PENDING_REVIEW.csv"
)
corrected.to_csv(corrected_path, index=False)

# Structural validation
reloaded = pd.read_csv(corrected_path)

report = {
    "phase": 4,
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "source_benchmark": str(benchmark_path),
    "corrected_benchmark": str(corrected_path),
    "correction_proposals": str(proposal_path),
    "benchmark_rows": int(len(reloaded)),
    "unique_question_ids": int(reloaded["question_id"].nunique()),
    "duplicate_question_ids": int(reloaded["question_id"].duplicated().sum()),
    "corrected_question_ids": updated_ids,
    "corrected_rows_have_evidence": bool(
        reloaded.loc[
            reloaded["question_id"].astype(str).isin(updated_ids),
            "evidence_passage"
        ].fillna("").str.strip().ne("").all()
    ),
    "original_benchmark_overwritten": False,
    "legal_review_status": "PENDING",
}

report_path = RESULTS / "phase4_evidence_correction_validation_v2.json"
with open(report_path, "w", encoding="utf-8") as f:
    json.dump(report, f, indent=2, ensure_ascii=False)

print("PHASE 4 EVIDENCE CORRECTION CHECK")
print("=" * 65)
print(json.dumps(report, indent=2))
print("\nPROPOSED PASSAGE LENGTHS")
print(proposals[["question_id", "character_count", "review_status"]].to_string(index=False))
print("\nSaved files:")
print(proposal_path)
print(corrected_path)
print(report_path)
```

```python
import pandas as pd
import json
from pathlib import Path

ROOT = Path("/kaggle/working/indian_legal_rag")
RESULTS = ROOT / "09_Results_Figures"

proposal_path = RESULTS / "phase4_evidence_correction_proposals_v2.csv"
benchmark_path = (
    ROOT / "08_Evaluation" /
    "phase4_qa_benchmark_v2_evidence_corrected_PENDING_REVIEW.csv"
)
original_path = (
    ROOT / "08_Evaluation" /
    "phase4_qa_benchmark_v1_RECONSTRUCTED.csv"
)

proposals = pd.read_csv(proposal_path)
corrected = pd.read_csv(benchmark_path)
original = pd.read_csv(original_path)

# Review proposal quality
proposals["passage_nonempty"] = (
    proposals["proposed_evidence_passage"]
    .fillna("").str.strip().ne("")
)
proposals["passage_length_reasonable"] = (
    proposals["character_count"].between(100, 5000)
)
proposals["document_id_present"] = (
    proposals["expected_document_id"].fillna("").str.strip().ne("")
)

print("EVIDENCE PROPOSAL REVIEW")
display(proposals[[
    "question_id",
    "expected_document_id",
    "source_chunk_ids",
    "character_count",
    "passage_nonempty",
    "passage_length_reasonable",
    "document_id_present",
]])

# Compare original and corrected benchmark
comparison = original[["question_id", "evidence_passage"]].merge(
    corrected[["question_id", "evidence_passage"]],
    on="question_id",
    how="outer",
    suffixes=("_original", "_corrected"),
    indicator=True
)

comparison["evidence_changed"] = (
    comparison["evidence_passage_original"].fillna("")
    != comparison["evidence_passage_corrected"].fillna("")
)

print("\nBENCHMARK COMPARISON")
display(comparison[[
    "question_id", "_merge", "evidence_changed"
]])

report = {
    "proposal_count": int(len(proposals)),
    "all_passages_nonempty": bool(proposals["passage_nonempty"].all()),
    "all_lengths_reasonable": bool(proposals["passage_length_reasonable"].all()),
    "all_document_ids_present": bool(proposals["document_id_present"].all()),
    "benchmark_question_count": int(corrected["question_id"].nunique()),
    "benchmark_duplicates": int(corrected["question_id"].duplicated().sum()),
    "changed_evidence_count": int(comparison["evidence_changed"].sum()),
    "all_question_ids_match": bool((comparison["_merge"] == "both").all()),
    "legal_review": "PENDING_MANUAL_REVIEW",
}

report_path = RESULTS / "phase4_final_consistency_check_v1.json"
with open(report_path, "w", encoding="utf-8") as f:
    json.dump(report, f, indent=2)

print("\nFINAL CONSISTENCY CHECK")
print(json.dumps(report, indent=2))
print("\nSaved:", report_path)
```

```python
import pandas as pd
import json
from pathlib import Path
from datetime import datetime, timezone

ROOT = Path("/kaggle/working/indian_legal_rag")
EVAL = ROOT / "08_Evaluation"
RESULTS = ROOT / "09_Results_Figures"

source_path = (
    EVAL /
    "phase4_qa_benchmark_v2_evidence_corrected_PENDING_REVIEW.csv"
)

output_path = (
    EVAL /
    "phase4_qa_benchmark_v3_evidence_corrected_PENDING_REVIEW.csv"
)

report_path = (
    RESULTS /
    "phase4_evidence_correction_validation_v3.json"
)

# Retrieve the two source chunks
def get_chunk_text(chunk_id):
    matches = chunks.loc[
        chunks["chunk_id"].astype(str) == chunk_id,
        "text"
    ]
    if matches.empty:
        raise ValueError(f"Missing source chunk: {chunk_id}")
    return str(matches.iloc[0])

chunk176 = get_chunk_text("doc_5970_v2_chunk_0176")
chunk177 = get_chunk_text("doc_5970_v2_chunk_0177")

# Extract the final conclusions from chunk 176
start = chunk176.find("FINAL CONCLUSIONS:")
end = chunk176.find("Regarding Question No.")
if start < 0 or end < 0:
    raise ValueError("Could not locate the final conclusions in chunk 176.")

# Include the complete first and second conclusions.
passage_start = chunk176[start:]
passage_start = passage_start.strip()

# Extract the missing qualification from chunk 177
qualification_start = chunk177.find(
    "Again, if there are any special or peculiar features"
)
qualification_end_phrase = (
    "it is open for it to do so."
)
qualification_end = (
    chunk177.find(qualification_end_phrase, qualification_start)
    if qualification_start >= 0 else -1
)

if qualification_start < 0 or qualification_end < 0:
    raise ValueError("Could not locate the anticipatory-bail qualification.")

qualification = chunk177[
    qualification_start:
    qualification_end + len(qualification_end_phrase)
].strip()

corrected_passage = passage_start + "\n" + qualification

# Load the previous candidate and change only CRIM_006
df = pd.read_csv(source_path)

required_columns = {"question_id", "evidence_passage"}
if not required_columns.issubset(df.columns):
    raise ValueError("Benchmark is missing required columns.")

mask = df["question_id"].astype(str).eq("CRIM_006")
if mask.sum() != 1:
    raise ValueError("Expected exactly one CRIM_006 row.")

df.loc[mask, "evidence_passage"] = corrected_passage

# Save as a new candidate; do not overwrite earlier files
df.to_csv(output_path, index=False)

report = {
    "phase": 4,
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "source_benchmark": str(source_path),
    "new_candidate": str(output_path),
    "updated_question_id": "CRIM_006",
    "source_chunks": [
        "doc_5970_v2_chunk_0176",
        "doc_5970_v2_chunk_0177"
    ],
    "passage_character_count": len(corrected_passage),
    "contains_end_of_trial_rule": (
        "can continue till the end of the trial" in corrected_passage
    ),
    "contains_special_circumstances_qualification": (
        qualification_end_phrase in corrected_passage
    ),
    "benchmark_rows": int(len(df)),
    "unique_question_ids": int(df["question_id"].nunique()),
    "duplicate_question_ids": int(df["question_id"].duplicated().sum()),
    "original_benchmark_overwritten": False,
    "legal_review_status": "PENDING_MANUAL_REVIEW"
}

with open(report_path, "w", encoding="utf-8") as f:
    json.dump(report, f, indent=2)

print("PHASE 4 CRIM_006 CORRECTION")
print(json.dumps(report, indent=2))

print("\nUPDATED EVIDENCE PASSAGE")
print(corrected_passage)

print("\nSaved candidate:", output_path)
print("Saved report:", report_path)
```

```python
import pandas as pd
import json
from pathlib import Path
from datetime import datetime, timezone

ROOT = Path("/kaggle/working/indian_legal_rag")
EVAL = ROOT / "08_Evaluation"
RESULTS = ROOT / "09_Results_Figures"

candidate_path = (
    EVAL /
    "phase4_qa_benchmark_v3_evidence_corrected_PENDING_REVIEW.csv"
)
original_path = (
    EVAL /
    "phase4_qa_benchmark_v1_RECONSTRUCTED.csv"
)
report_path = RESULTS / "phase4_final_technical_audit_v1.json"

candidate = pd.read_csv(candidate_path)
original = pd.read_csv(original_path)

required_columns = {
    "question_id",
    "evidence_passage"
}

if not required_columns.issubset(candidate.columns):
    raise ValueError("Required benchmark columns are missing.")

expected_corrected_ids = {
    "CRIM_003", "CRIM_006", "CRIM_007", "CRIM_009"
}

candidate_ids = set(candidate["question_id"].astype(str))

# Confirm the four evidence corrections
corrected_passages_nonempty = {}
for qid in expected_corrected_ids:
    rows = candidate.loc[
        candidate["question_id"].astype(str).eq(qid),
        "evidence_passage"
    ]
    corrected_passages_nonempty[qid] = (
        len(rows) == 1
        and rows.iloc[0] is not None
        and str(rows.iloc[0]).strip() != ""
    )

# Verify CRIM_006 includes both parts of the rule
crim006 = candidate.loc[
    candidate["question_id"].astype(str).eq("CRIM_006"),
    "evidence_passage"
].iloc[0]

report = {
    "phase": 4,
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "candidate_file": str(candidate_path),
    "original_benchmark_preserved": original_path.exists(),
    "benchmark_rows": int(len(candidate)),
    "unique_question_ids": int(candidate["question_id"].nunique()),
    "duplicate_question_ids": int(
        candidate["question_id"].duplicated().sum()
    ),
    "required_columns_present": True,
    "all_evidence_passages_nonempty": bool(
        candidate["evidence_passage"].fillna("").str.strip().ne("").all()
    ),
    "corrected_passages_nonempty": corrected_passages_nonempty,
    "crim006_has_end_of_trial_rule": (
        "can continue till the end of the trial" in crim006
    ),
    "crim006_has_special_circumstances_qualification": (
        "special or peculiar features" in crim006
    ),
    "legal_review_status": "PENDING_MANUAL_REVIEW"
}

report["technical_checks_passed"] = all([
    report["original_benchmark_preserved"],
    report["benchmark_rows"] == 10,
    report["unique_question_ids"] == 10,
    report["duplicate_question_ids"] == 0,
    report["all_evidence_passages_nonempty"],
    all(corrected_passages_nonempty.values()),
    report["crim006_has_end_of_trial_rule"],
    report["crim006_has_special_circumstances_qualification"]
])

with open(report_path, "w", encoding="utf-8") as f:
    json.dump(report, f, indent=2)

print("PHASE 4 FINAL TECHNICAL AUDIT")
print(json.dumps(report, indent=2))
print("\nReport saved:", report_path)
```

```python

from pathlib import Path
import pandas as pd

ROOT = Path("/kaggle/working/indian_legal_rag")
SEARCH_ROOTS = [
    Path("/kaggle/working"),
    Path("/kaggle/input"),
]

expected_names = [
    "phase7_retrieval_rankings_v1.csv",
    "phase7_per_question_metrics_v2_corrected.csv",
    "phase7_retrieval_metrics_summary_v2_corrected.csv",
    "phase7_retrieval_evaluation_manifest_v2_corrected.json",
    "phase8_query_expansion_pilot_v1.csv",
    "phase8_query_expansion_metrics_v1.csv",
    "phase9_final_answer_validation_v1.csv",
    "phase9_final_approved_answers_v1.csv",
    "phase9_final_completion_report_v1.json",
    "phase10_consolidated_evaluation_v1.csv",
    "phase10_consolidated_evaluation_summary_v1.json",
    "phase10_final_evaluation_v1.csv",
    "phase10_final_evaluation_report_v1.json",
    "phase10_failure_analysis_v1.csv",
    "phase10_failure_analysis_report_v1.json",
]

found = []

for search_root in SEARCH_ROOTS:
    if not search_root.exists():
        continue

    for path in search_root.rglob("*"):
        if path.is_file() and path.name in expected_names:
            found.append({
                "filename": path.name,
                "location": str(path),
                "size_bytes": path.stat().st_size,
            })

found_df = pd.DataFrame(found).drop_duplicates() if found else pd.DataFrame(
    columns=["filename", "location", "size_bytes"]
)

print("SEARCH RESULTS")
print("Matching files found:", len(found_df))

if found_df.empty:
    print("No matching files were found in the searched directories.")
else:
    display(found_df.sort_values("filename").reset_index(drop=True))

print("\nMISSING EXPECTED FILENAMES")
found_names = set(found_df["filename"])
for name in expected_names:
    if name not in found_names:
        print("-", name)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import json
from pathlib import Path

ROOT = Path("/kaggle/working/indian_legal_rag")
RESULTS = ROOT / "09_Results_Figures"
RESULTS.mkdir(parents=True, exist_ok=True)

metrics_path = (
    RESULTS / "phase7_retrieval_metrics_summary_v2_corrected.csv"
)

metrics = pd.read_csv(metrics_path)

print("PHASE 7 RETRIEVAL METRICS")
display(metrics)

print("Available columns:", metrics.columns.tolist())

# Automatically identify the method column
possible_method_columns = [
    "method", "retrieval_method", "system", "approach"
]
method_col = next(
    (c for c in possible_method_columns if c in metrics.columns),
    None
)

if method_col is None:
    raise ValueError(
        "Could not identify the retrieval-method column. "
        "Share the displayed column names."
    )

# Metrics to plot when present
plot_groups = {
    "hit_rate_comparison": ["HitRate@5", "HitRate@10"],
    "mrr_comparison": ["MRR"],
    "ndcg_comparison": ["nDCG@5", "nDCG@10"],
}

# Normalize common column-name variations
normalized_columns = {
    c.lower().replace("_", "").replace(" ", ""): c
    for c in metrics.columns
}

metric_aliases = {
    "HitRate@5": ["hitrate@5", "hit_rate@5", "hit_rate_5"],
    "HitRate@10": ["hitrate@10", "hit_rate@10", "hit_rate_10"],
    "MRR": ["mrr"],
    "nDCG@5": ["ndcg@5", "ndcg_5"],
    "nDCG@10": ["ndcg@10", "ndcg_10"],
}

resolved_metrics = {}

for label, aliases in metric_aliases.items():
    for alias in aliases:
        key = alias.lower().replace("_", "").replace(" ", "")
        if key in normalized_columns:
            resolved_metrics[label] = normalized_columns[key]
            break

# Create plots only for metrics actually present
for filename, labels in plot_groups.items():
    available = [
        label for label in labels
        if label in resolved_metrics
    ]

    if not available:
        print(f"Skipping {filename}: metrics not found.")
        continue

    ax = metrics.set_index(method_col)[
        [resolved_metrics[label] for label in available]
    ].plot(kind="bar", figsize=(10, 5))

    ax.set_title(filename.replace("_", " ").title())
    ax.set_ylabel("Score")
    ax.set_xlabel("Retrieval Method")
    ax.set_ylim(0, 1)
    ax.legend(available)
    plt.xticks(rotation=25, ha="right")
    plt.tight_layout()

    output_path = RESULTS / f"phase7_{filename}.png"
    plt.savefig(output_path, dpi=200, bbox_inches="tight")
    plt.show()
    plt.close()

    print("Saved:", output_path)

print("\nPhase 7 graph generation finished.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

ROOT = Path("/kaggle/working/indian_legal_rag")
RESULTS = ROOT / "09_Results_Figures"

df = pd.read_csv(
    RESULTS / "phase8_query_expansion_metrics_v1.csv"
)

metrics = [
    "HitRate@5",
    "HitRate@10",
    "MRR",
    "nDCG@5",
    "nDCG@10"
]

# Aggregate across the 10 benchmark questions
summary = (
    df.groupby(["query_type", "method"], as_index=False)[metrics]
      .mean()
)

print("PHASE 8 AGGREGATED COMPARISON")
display(summary.round(4))

# Save aggregated results
summary_path = RESULTS / "phase8_query_expansion_aggregated_summary_v1.csv"
summary.to_csv(summary_path, index=False)
print("Saved:", summary_path)

# Plot MRR comparison for each retrieval method
pivot = summary.pivot(
    index="method",
    columns="query_type",
    values="MRR"
)

ax = pivot.plot(kind="bar", figsize=(10, 5))
ax.set_title("Baseline vs Expanded Queries — Mean Reciprocal Rank")
ax.set_xlabel("Retrieval Method")
ax.set_ylabel("Mean Reciprocal Rank")
ax.set_ylim(0, 1)
plt.xticks(rotation=20, ha="right")
plt.tight_layout()

plot_path = RESULTS / "phase8_baseline_vs_expanded_mrr.png"
plt.savefig(plot_path, dpi=200, bbox_inches="tight")
plt.show()
plt.close()

print("Saved:", plot_path)

# Calculate expanded-minus-baseline differences
comparison = summary.pivot(
    index="method",
    columns="query_type",
    values=metrics
)

print("\nEXPANSION EFFECT (expanded minus baseline)")
for metric in metrics:
    if (metric, "baseline") in comparison.columns and (
        metric, "expanded"
    ) in comparison.columns:
        difference = (
            comparison[(metric, "expanded")]
            - comparison[(metric, "baseline")]
        )
        print(f"\n{metric}:")
        display(difference.rename("change").to_frame().round(4))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import json
from pathlib import Path

ROOT = Path("/kaggle/working/indian_legal_rag")
RESULTS = ROOT / "09_Results_Figures"

df = pd.read_csv(RESULTS / "phase10_final_evaluation_v1.csv")

# 1. Semantic similarity by question
df = df.sort_values("question_id")

fig, ax = plt.subplots(figsize=(11, 5))
ax.bar(df["question_id"], df["semantic_similarity"])
ax.axhline(
    df["semantic_similarity"].mean(),
    linestyle="--",
    label=f'Mean = {df["semantic_similarity"].mean():.3f}'
)
ax.set_title("Phase 10: Semantic Similarity by Question")
ax.set_xlabel("Question ID")
ax.set_ylabel("Semantic Similarity")
ax.set_ylim(0, 1.05)
ax.legend()
plt.xticks(rotation=35, ha="right")
plt.tight_layout()

path = RESULTS / "phase10_semantic_similarity_by_question.png"
plt.savefig(path, dpi=200, bbox_inches="tight")
plt.show()
plt.close()
print("Saved:", path)

# 2. Citation and recorded grounding coverage
citation_rate = df["citation_pass"].fillna(False).astype(bool).mean()
grounding_rate = df["grounding_pass"].fillna(False).astype(bool).mean()

coverage = pd.DataFrame({
    "Check": ["Citation check passed", "Recorded grounding check passed"],
    "Rate": [citation_rate, grounding_rate]
})

fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(coverage["Check"], coverage["Rate"])
ax.set_title("Phase 10: Citation and Recorded Grounding Checks")
ax.set_ylabel("Pass Rate")
ax.set_ylim(0, 1.05)
plt.xticks(rotation=10, ha="right")
plt.tight_layout()

path = RESULTS / "phase10_citation_grounding_coverage.png"
plt.savefig(path, dpi=200, bbox_inches="tight")
plt.show()
plt.close()
print("Saved:", path)

# 3. Review flags
if "review_flag" in df.columns:
    review_count = (
        df["review_flag"]
        .astype(str)
        .str.strip()
        .str.lower()
        .isin(["true", "1", "yes"])
        .sum()
    )

    print("\nQUESTIONS FLAGGED FOR REVIEW:", int(review_count))
    print(
        df.loc[
            df["review_flag"].astype(str).str.strip().str.lower()
            .isin(["true", "1", "yes"]),
            ["question_id", "semantic_similarity", "review_flag"]
        ].to_string(index=False)
    )

print("\nSUMMARY")
print("Questions evaluated:", len(df))
print("Mean semantic similarity:", round(df["semantic_similarity"].mean(), 4))
print("Median semantic similarity:", round(df["semantic_similarity"].median(), 4))
print("Citation check pass rate:", round(citation_rate, 4))
print("Recorded grounding pass rate:", round(grounding_rate, 4))
```

```python
import pandas as pd
from pathlib import Path

ROOT = Path("/kaggle/working/indian_legal_rag")
RESULTS = ROOT / "09_Results_Figures"

df = pd.read_csv(RESULTS / "phase10_final_evaluation_v1.csv")

flagged_ids = ["CRIM_002", "CRIM_004"]

columns_to_review = [
    "question_id",
    "question",
    "reference_answer",
    "final_answer",
    "evidence_document",
    "evidence_passage",
    "supporting_chunk_ids",
    "cited_evidence",
    "semantic_similarity",
    "evidence_grounding_review",
    "final_grounding_decision",
    "citation_pass",
    "review_flag",
]

available_columns = [c for c in columns_to_review if c in df.columns]

for qid in flagged_ids:
    row = df.loc[df["question_id"].astype(str).eq(qid)]

    print("\n" + "=" * 100)
    print("MANUAL REVIEW:", qid)
    print("=" * 100)

    if row.empty:
        print("Question not found.")
        continue

    for col in available_columns:
        print(f"\n{col.upper()}:")
        print(row.iloc[0][col])

print("\nReview only these two answers for now.")
```

```python
import pandas as pd
from pathlib import Path

root = Path("/kaggle/working/indian_legal_rag/09_Results_Figures")

source = root / "phase10_final_evaluation_v1.csv"
output = root / "phase10_final_evaluation_v2_manual_review.csv"

df = pd.read_csv(source)

reviewed_ids = ["CRIM_002", "CRIM_004"]

updates = {
    "evidence_grounding_review": "MANUALLY_REVIEWED_SUPPORTED",
    "final_grounding_decision": "SUPPORTED",
    "grounding_pass": True,
    "citation_pass": True,
    "review_flag": False,
}

for column, value in updates.items():
    if column in df.columns:
        df.loc[df["question_id"].isin(reviewed_ids), column] = value

df.to_csv(output, index=False)

print("Saved:", output)
print("\nUpdated records:")
print(
    df.loc[df["question_id"].isin(reviewed_ids),
           ["question_id", "semantic_similarity",
            "evidence_grounding_review", "final_grounding_decision",
            "grounding_pass", "citation_pass", "review_flag"]]
    .to_string(index=False)
)
print("\nRemaining flagged questions:", int(df["review_flag"].fillna(False).astype(bool).sum()))
```

```python
from pathlib import Path
import pandas as pd
import json

ROOT = Path("/kaggle/working/indian_legal_rag")
OUT = ROOT / "09_Results_Figures"

required_files = [
    "phase11_consolidated_retrieval_metrics.csv",
    "phase11_query_expansion_comparison.csv",
    "phase11_final_answer_summary.csv",
    "phase11_final_answer_summary.json",
    "phase11_final_answer_similarity.png",
    "phase11_final_evaluation_pass_rates.png",
    "phase11_project_consolidation_report.json",
    "phase10_final_evaluation_v2_manual_review.csv",
    "phase7_retrieval_metrics_summary_v2_corrected.csv",
    "phase8_query_expansion_aggregated_summary_v1.csv",
]

audit = []

for name in required_files:
    path = OUT / name
    audit.append({
        "file": name,
        "exists": path.is_file(),
        "size_bytes": path.stat().st_size if path.is_file() else 0
    })

audit_df = pd.DataFrame(audit)

print("PHASE 12 — FINAL INTEGRITY AUDIT")
print(audit_df.to_string(index=False))

# Validate key datasets
p7 = pd.read_csv(OUT / "phase11_consolidated_retrieval_metrics.csv")
p8 = pd.read_csv(OUT / "phase11_query_expansion_comparison.csv")
p10 = pd.read_csv(OUT / "phase10_final_evaluation_v2_manual_review.csv")

print("\nDATA VALIDATION")
print("Phase 7 methods:", len(p7))
print("Phase 8 configurations:", len(p8))
print("Phase 10 questions:", p10["question_id"].nunique())
print("Phase 10 duplicate question IDs:", int(p10["question_id"].duplicated().sum()))
print(
    "Phase 10 remaining review flags:",
    int(p10["review_flag"].fillna(False).astype(bool).sum())
)

report_path = OUT / "phase11_project_consolidation_report.json"
with open(report_path, "r", encoding="utf-8") as f:
    report = json.load(f)

print("Legal review status:", report["final_answer_evaluation"]["legal_review_status"])

passed = (
    audit_df["exists"].all()
    and len(p7) == 4
    and len(p8) == 8
    and p10["question_id"].nunique() == 10
    and not p10["question_id"].duplicated().any()
    and int(p10["review_flag"].fillna(False).astype(bool).sum()) == 0
)

print("\nTECHNICAL INTEGRITY STATUS:", "PASS" if passed else "CHECK REQUIRED")
print("Legal review remains a separate requirement.")
```

```python
# PHASE 13 — STEP 6D: GENERATE AN EVIDENCE-GROUNDED ANSWER

import torch

test_question = (
    "What factors should a court consider when deciding whether "
    "to grant anticipatory bail?"
)

# Reuse the existing retrieval and reranking pipeline
test_results = retrieve_and_rerank(
    test_question,
    top_k=5,
    candidate_k=20
)

# Prepare evidence with traceable chunk IDs
evidence_blocks = []

for i, result in enumerate(test_results, start=1):
    evidence_blocks.append(
        f"[Evidence {i} | Chunk ID: {result['chunk_id']} | "
        f"Source: {result['source_file']}]\n"
        f"{result['text']}"
    )

evidence_context = "\n\n".join(evidence_blocks)

prompt = f"""You are an assistant for Indian legal research.

Answer the question using ONLY the supplied evidence.
Do not invent legal provisions, facts, or case names.
If the evidence is insufficient, clearly state the limitation.
Distinguish general legal principles from case-specific findings.
Cite the relevant chunk IDs in square brackets, such as [doc_123_chunk_0001].
Do not claim that you independently verified the law.

QUESTION:
{test_question}

RETRIEVED EVIDENCE:
{evidence_context}

Provide a concise, structured answer with the important factors.
"""

messages = [
    {
        "role": "system",
        "content": "You provide cautious, evidence-grounded Indian legal research assistance."
    },
    {
        "role": "user",
        "content": prompt
    }
]

formatted_prompt = llm_tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)

inputs = llm_tokenizer(
    formatted_prompt,
    return_tensors="pt",
    truncation=True,
    max_length=4096
)

print("Generating answer on CPU. This may take a few minutes...")

with torch.inference_mode():
    outputs = llm_model.generate(
        **inputs,
        max_new_tokens=350,
        do_sample=False,
        repetition_penalty=1.05,
        pad_token_id=llm_tokenizer.eos_token_id
    )

new_tokens = outputs[0][inputs["input_ids"].shape[1]:]
generated_answer = llm_tokenizer.decode(
    new_tokens,
    skip_special_tokens=True
)

print("\n" + "=" * 80)
print("QUESTION")
print(test_question)

print("\nGENERATED ANSWER")
print(generated_answer)

print("\nRETRIEVED CITATION IDS")
for result in test_results:
    print(f"- {result['chunk_id']} ({result['source_file']})")
```

```python
# PHASE 13 — STEP 7: VERIFY CITATION TRACEABILITY

import re
import pandas as pd

# IDs returned by the retrieval pipeline
retrieved_ids = {
    str(result["chunk_id"])
    for result in test_results
}

# Extract chunk IDs cited inside the generated answer
cited_ids = set(
    re.findall(r"\bdoc_[A-Za-z0-9]+_v\d+_chunk_\d+\b", generated_answer)
)

print("=" * 65)
print("CITATION TRACEABILITY CHECK")
print("=" * 65)

print("\nRetrieved chunk IDs:")
for cid in sorted(retrieved_ids):
    print("-", cid)

print("\nChunk IDs cited in generated answer:")
if cited_ids:
    for cid in sorted(cited_ids):
        print("-", cid)
else:
    print("No chunk IDs were cited in the answer.")

valid_citations = cited_ids.intersection(retrieved_ids)
invalid_citations = cited_ids - retrieved_ids

print("\nCited IDs:", len(cited_ids))
print("Valid retrieved citations:", len(valid_citations))
print("Citations not in retrieved results:", len(invalid_citations))

if invalid_citations:
    print("Unmatched citation IDs:", sorted(invalid_citations))

if cited_ids and not invalid_citations:
    print("\nRESULT: Citation IDs match the retrieved evidence.")
else:
    print("\nRESULT: Citation formatting or traceability needs improvement.")

print(
    "\nImportant: Matching IDs verifies traceability only. "
    "It does not prove every legal claim is supported by its cited passage."
)
```

```python
# PHASE 13 — STEP 8: GENERATE ANSWER WITH INLINE CITATIONS

import torch

test_question = (
    "What factors should a court consider when deciding whether "
    "to grant anticipatory bail?"
)

# Reuse the five passages already retrieved and reranked
evidence_context = "\n\n".join(
    f"[{r['chunk_id']} | Source: {r['source_file']}]\n{r['text']}"
    for r in test_results
)

messages = [
    {
        "role": "system",
        "content": (
            "You are an Indian legal research assistant. "
            "Answer only from the supplied evidence. "
            "Never invent evidence IDs or legal rules. "
            "Every numbered factor must end with at least one exact "
            "evidence chunk ID in square brackets. "
            "Use only chunk IDs appearing in the supplied evidence. "
            "If a factor is not supported, omit it or state that the "
            "provided evidence is insufficient."
        )
    },
    {
        "role": "user",
        "content": f"""
QUESTION:
{test_question}

SUPPLIED EVIDENCE:
{evidence_context}

Write a concise answer with up to 8 factors.
Cite each factor using the exact chunk ID, for example:
[doc_5500_v2_chunk_0009]

Do not list citations separately at the end.
Place citations directly after the claims they support.
"""
    }
]

formatted_prompt = llm_tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)

inputs = llm_tokenizer(
    formatted_prompt,
    return_tensors="pt",
    truncation=True,
    max_length=4096
)

print("Generating revised answer on CPU...")

with torch.inference_mode():
    outputs = llm_model.generate(
        **inputs,
        max_new_tokens=400,
        do_sample=False,
        repetition_penalty=1.05,
        pad_token_id=llm_tokenizer.eos_token_id
    )

new_tokens = outputs[0][inputs["input_ids"].shape[1]:]

generated_answer_v2 = llm_tokenizer.decode(
    new_tokens,
    skip_special_tokens=True
)

print("\n" + "=" * 75)
print("REVISED ANSWER WITH INLINE CITATIONS")
print("=" * 75)
print(generated_answer_v2)
```

```python
# PHASE 13 — STEP 9: ADD TRACEABLE CITATIONS AND VALIDATE THEM

import re
import pandas as pd

# Use the answer generated in the previous step
answer_text = generated_answer_v2

# Map retrieved chunk IDs to their evidence text
evidence_by_id = {
    str(r["chunk_id"]): str(r["text"])
    for r in test_results
}

# Assign the best-matching retrieved passage to each numbered answer point.
# This is an automated traceability aid, NOT proof of legal support.
answer_lines = answer_text.splitlines()
cited_lines = []
citation_audit = []

current_point = None

for line in answer_lines:
    match = re.match(r"^\s*(\d+)\.\s+", line)

    if match:
        current_point = line.strip()

        # Preserve any citation already supplied by the model.
        existing_ids = re.findall(
            r"\bdoc_[A-Za-z0-9]+_v\d+_chunk_\d+\b", line
        )

        if existing_ids:
            selected_id = existing_ids[0]
            method = "model_supplied"
        else:
            # Match the answer point against retrieved passages using the
            # existing cross-encoder, then select its highest-scoring passage.
            candidate_texts = [
                (current_point, evidence_by_id[cid])
                for cid in evidence_by_id
            ]

            scores = cross_encoder.predict(candidate_texts)
            best_idx = int(scores.argmax())
            selected_id = list(evidence_by_id.keys())[best_idx]
            method = "automatically_assigned_by_cross_encoder"

            line = f"{line} [{selected_id}]"

        citation_audit.append({
            "answer_point": current_point,
            "citation_id": selected_id,
            "citation_in_retrieved_results": selected_id in evidence_by_id,
            "assignment_method": method
        })

    cited_lines.append(line)

generated_answer_v3 = "\n".join(cited_lines)

audit_df = pd.DataFrame(citation_audit)

print("=" * 80)
print("ANSWER WITH TRACEABLE CITATIONS")
print("=" * 80)
print(generated_answer_v3)

print("\nCITATION AUDIT")
print(audit_df.to_string(index=False))

print("\nAll citation IDs belong to retrieved results:",
      bool(audit_df["citation_in_retrieved_results"].all()))

print(
    "\nLIMITATION: Automated citation assignment establishes which retrieved "
    "passage was selected for each point. Human review is still needed to "
    "confirm that each passage substantively supports its claim."
)
```

```python
# PHASE 13 — STEP 12: RETRIEVAL TEST ON FIVE NEW QUESTIONS

test_questions = [
    "When can a dying declaration be the sole basis for conviction?",
    "What is the right to a speedy trial under Article 21 of the Constitution?",
    "When can an appellate court reverse an acquittal?",
    "What legal assistance must be provided to an accused who cannot afford a lawyer?",
    "What is the burden of proving an alibi in a criminal case?"
]

additional_test_results = []

for i, question in enumerate(test_questions, start=1):
    print("\n" + "=" * 90)
    print(f"TEST {i}: {question}")

    try:
        results = retrieve_and_rerank(
            question,
            top_k=3,
            candidate_k=20
        )

        print("Retrieved passages:", len(results))

        for result in results:
            print(
                f"\nRank: {result['rank']}"
                f"\nChunk: {result['chunk_id']}"
                f"\nSource: {result['source_file']}"
                f"\nScore: {result['reranker_score']}"
                f"\nPassage: {result['text'][:500]}"
            )

        additional_test_results.append({
            "question_id": f"EXTRA_{i:02d}",
            "question": question,
            "retrieval_success": bool(results),
            "results": results
        })

    except Exception as e:
        print("RETRIEVAL ERROR:", type(e).__name__, str(e)[:300])

        additional_test_results.append({
            "question_id": f"EXTRA_{i:02d}",
            "question": question,
            "retrieval_success": False,
            "results": [],
            "error": str(e)
        })

print("\n" + "=" * 90)
print("RETRIEVAL TEST SUMMARY")
print("Questions tested:", len(additional_test_results))
print(
    "Questions with retrieved results:",
    sum(r["retrieval_success"] for r in additional_test_results)
)
```

```python
# PHASE 13 — STEP 13: GENERATE ANSWERS FOR FIVE TEST QUESTIONS

import torch
import re
import pandas as pd

generated_test_answers = []

for item in additional_test_results:
    question_id = item["question_id"]
    question = item["question"]
    results = item["results"]

    print("\n" + "=" * 85)
    print(f"{question_id}: {question}")

    if not results:
        print("No evidence retrieved. Answer generation skipped.")
        generated_test_answers.append({
            "question_id": question_id,
            "question": question,
            "answer": "",
            "retrieved_chunk_ids": [],
            "generation_status": "NO_EVIDENCE"
        })
        continue

    # Include the retrieved passages and their exact IDs
    evidence_context = "\n\n".join(
        f"[{r['chunk_id']} | Source: {r['source_file']}]\n{r['text']}"
        for r in results
    )

    messages = [
        {
            "role": "system",
            "content": (
                "You are an Indian legal research assistant. "
                "Answer only using the supplied evidence. "
                "Cite claims with exact chunk IDs from the evidence. "
                "Never invent citations or legal rules. "
                "If evidence is insufficient, say so clearly. "
                "Do not claim independent legal verification."
            )
        },
        {
            "role": "user",
            "content": (
                f"QUESTION:\n{question}\n\n"
                f"RETRIEVED EVIDENCE:\n{evidence_context}\n\n"
                "Provide a concise answer. Cite relevant evidence chunk IDs "
                "directly in the answer using square brackets."
            )
        }
    ]

    prompt = llm_tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = llm_tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=4096
    )

    with torch.inference_mode():
        output = llm_model.generate(
            **inputs,
            max_new_tokens=220,
            do_sample=False,
            repetition_penalty=1.05,
            pad_token_id=llm_tokenizer.eos_token_id
        )

    new_tokens = output[0][inputs["input_ids"].shape[1]:]
    answer = llm_tokenizer.decode(
        new_tokens,
        skip_special_tokens=True
    ).strip()

    retrieved_ids = [r["chunk_id"] for r in results]
    cited_ids = list(set(
        re.findall(r"\bdoc_[A-Za-z0-9]+_v\d+_chunk_\d+\b", answer)
    ))

    generated_test_answers.append({
        "question_id": question_id,
        "question": question,
        "answer": answer,
        "retrieved_chunk_ids": retrieved_ids,
        "cited_chunk_ids": cited_ids,
        "citation_ids_match_retrieval": (
            bool(cited_ids) and set(cited_ids).issubset(set(retrieved_ids))
        ),
        "generation_status": "GENERATED"
    })

    print("\nGENERATED ANSWER:")
    print(answer)

    print("\nCitation IDs present:", cited_ids if cited_ids else "None")
    print("Retrieved evidence IDs:", retrieved_ids)

# Save a new file without overwriting previous evaluation artifacts
output_path = (
    "/kaggle/working/indian_legal_rag/09_Results_Figures/"
    "phase13_additional_question_tests_v1.csv"
)

pd.DataFrame(generated_test_answers).to_csv(output_path, index=False)

print("\n" + "=" * 85)
print("ADDITIONAL QUESTION GENERATION COMPLETE")
print("Questions processed:", len(generated_test_answers))
print("Saved:", output_path)
```

```python
# PHASE 13 — STEP 14: Test insufficient-evidence handling

import re
import pandas as pd
from pathlib import Path

test_question = "What ingredients are needed to bake a chocolate cake?"

# Retrieve evidence using the existing retrieval pipeline
test_results = retrieve_and_rerank(
    test_question,
    top_k=5,
    candidate_k=20
)

# Convert retrieved results into a readable context
def get_value(item, key, default=""):
    if isinstance(item, dict):
        return item.get(key, default)
    return getattr(item, key, default)

context_parts = []
retrieved_ids = []

for item in test_results:
    chunk_id = (
        get_value(item, "chunk_id")
        or get_value(item, "id")
        or get_value(item, "document_id")
    )
    passage = (
        get_value(item, "text")
        or get_value(item, "content")
        or get_value(item, "chunk")
    )

    if chunk_id:
        retrieved_ids.append(str(chunk_id))

    if passage:
        context_parts.append(
            f"[{chunk_id}]\n{passage}"
        )

context = "\n\n".join(context_parts)

prompt = f"""
You are a question-answering system for an Indian legal document corpus.

Question:
{test_question}

Retrieved evidence:
{context if context else "No evidence was retrieved."}

Instructions:
- Answer only when the retrieved evidence is relevant and sufficient.
- If the evidence does not address the question, explicitly state:
  "The available legal corpus does not provide sufficient relevant evidence
  to answer this question."
- Do not invent facts, citations, or information from outside the evidence.
- If the evidence is unrelated, do not attempt to answer the question.
- Keep the response concise.
- Cite a chunk ID only if you use information from that chunk.
"""

inputs = llm_tokenizer(
    prompt,
    return_tensors="pt",
    truncation=True,
    max_length=4096
)

inputs = {
    key: value.to(llm_model.device)
    for key, value in inputs.items()
}

llm_model.eval()

import torch

with torch.no_grad():
    output = llm_model.generate(
        **inputs,
        max_new_tokens=120,
        do_sample=False,
        pad_token_id=llm_tokenizer.eos_token_id
    )

generated_tokens = output[0][inputs["input_ids"].shape[1]:]
test_answer = llm_tokenizer.decode(
    generated_tokens,
    skip_special_tokens=True
).strip()

print("=" * 80)
print("TEST QUESTION:", test_question)
print("\nRETRIEVED CHUNK IDS:", retrieved_ids)
print("\nGENERATED ANSWER:\n", test_answer)

print("\n" + "=" * 80)
print("MANUAL CHECK:")
print("- Did the model acknowledge that the question is outside the legal corpus?")
print("- Did it avoid inventing an answer or unsupported citations?")
print("- Remember: the model's response alone does not establish retrieval quality.")
```

```python
# PHASE 13 — STEP 15: Inspect relevance scores

import pandas as pd

def inspect_retrieval(question, label):
    print("\n" + "=" * 90)
    print(label)
    print("QUESTION:", question)

    results = retrieve_and_rerank(
        question,
        top_k=5,
        candidate_k=20
    )

    if isinstance(results, pd.DataFrame):
        rows = results.to_dict("records")
    else:
        rows = list(results)

    print("Number of results:", len(rows))

    for i, item in enumerate(rows, 1):
        if isinstance(item, dict):
            data = item
        elif hasattr(item, "_asdict"):
            data = item._asdict()
        else:
            data = vars(item) if hasattr(item, "__dict__") else {
                "value": str(item)
            }

        print(f"\nResult {i}:")
        for key, value in data.items():
            if key in ("text", "content", "chunk"):
                print(f"{key}: {str(value)[:450]}")
            else:
                print(f"{key}: {value}")

# Unrelated question
inspect_retrieval(
    "What ingredients are needed to bake a chocolate cake?",
    "UNRELATED QUESTION"
)

# Relevant legal question
inspect_retrieval(
    "When can a dying declaration be the sole basis for conviction?",
    "RELEVANT LEGAL QUESTION"
)
```

```python
# PHASE 13 — STEP 16: Compare relevance scores across test questions

import pandas as pd

calibration_questions = [
    ("LEGAL_01", "When can a dying declaration be the sole basis for conviction?", "relevant"),
    ("LEGAL_02", "What is the right to a speedy trial under Article 21?", "relevant"),
    ("LEGAL_03", "What is the burden of proving an alibi in a criminal case?", "relevant"),
    ("UNRELATED_01", "What ingredients are needed to bake a chocolate cake?", "unrelated"),
    ("UNRELATED_02", "What will the weather be like in Hyderabad tomorrow?", "unrelated"),
    ("UNRELATED_03", "How do I sort a list in Python?", "unrelated"),
]

calibration_results = []

for test_id, question, expected_type in calibration_questions:
    results = retrieve_and_rerank(
        question,
        top_k=5,
        candidate_k=20
    )

    if isinstance(results, pd.DataFrame):
        rows = results.to_dict("records")
    else:
        rows = list(results)

    scores = []
    top_chunk = None
    top_text = ""

    for rank, item in enumerate(rows, 1):
        if isinstance(item, dict):
            data = item
        elif hasattr(item, "_asdict"):
            data = item._asdict()
        else:
            data = vars(item) if hasattr(item, "__dict__") else {}

        score = data.get("reranker_score")
        if score is not None:
            scores.append(float(score))

        if rank == 1:
            top_chunk = data.get("chunk_id")
            top_text = str(data.get("text", ""))[:220].replace("\n", " ")

    calibration_results.append({
        "test_id": test_id,
        "expected_type": expected_type,
        "question": question,
        "top_score": max(scores) if scores else None,
        "lowest_top5_score": min(scores) if scores else None,
        "top_chunk_id": top_chunk,
        "top_passage_preview": top_text,
        "retrieval_count": len(rows),
    })

calibration_df = pd.DataFrame(calibration_results)

display(
    calibration_df[
        [
            "test_id",
            "expected_type",
            "top_score",
            "lowest_top5_score",
            "top_chunk_id",
            "retrieval_count",
        ]
    ]
)

print("\nTOP PASSAGE PREVIEWS")
for _, row in calibration_df.iterrows():
    print(f"\n{row['test_id']} ({row['expected_type']}): {row['question']}")
    print("Top passage:", row["top_passage_preview"])

print("\nNo threshold has been applied yet.")
```

```python
# PHASE 13 — STEP 17: Relevance gate + answer generation test

import re
import pandas as pd
from pathlib import Path
import torch

PROVISIONAL_THRESHOLD = 0.0

test_questions = [
    ("LEGAL_01", "When can a dying declaration be the sole basis for conviction?", "relevant"),
    ("LEGAL_02", "What is the right to a speedy trial under Article 21?", "relevant"),
    ("LEGAL_03", "What is the burden of proving an alibi in a criminal case?", "relevant"),
    ("UNRELATED_01", "What ingredients are needed to bake a chocolate cake?", "unrelated"),
    ("UNRELATED_02", "What will the weather be like in Hyderabad tomorrow?", "unrelated"),
    ("UNRELATED_03", "How do I sort a list in Python?", "unrelated"),
]

def result_to_dict(item):
    if isinstance(item, dict):
        return item
    if hasattr(item, "_asdict"):
        return item._asdict()
    if hasattr(item, "__dict__"):
        return vars(item)
    return {}

def get_passage(data):
    for key in ("text", "content", "chunk"):
        value = data.get(key)
        if value:
            return str(value)
    return ""

gate_rows = []

for test_id, question, expected_type in test_questions:
    print("\n" + "=" * 85)
    print(f"{test_id}: {question}")

    results = retrieve_and_rerank(
        question,
        top_k=5,
        candidate_k=20
    )

    if isinstance(results, pd.DataFrame):
        result_items = results.to_dict("records")
    else:
        result_items = list(results)

    result_dicts = [result_to_dict(item) for item in result_items]

    valid_scores = [
        float(item["reranker_score"])
        for item in result_dicts
        if item.get("reranker_score") is not None
    ]

    top_score = max(valid_scores) if valid_scores else None

    if not result_dicts or top_score is None:
        gate_status = "BLOCKED_NO_VALID_RETRIEVAL_SCORE"
        answer = (
            "The system could not establish that relevant evidence "
            "was retrieved, so it cannot provide a supported answer."
        )
        retrieved_ids = []
        cited_ids = []
    elif top_score < PROVISIONAL_THRESHOLD:
        gate_status = "BLOCKED_LOW_RELEVANCE_SCORE"
        answer = (
            "The available legal corpus does not provide sufficient "
            "relevant evidence to answer this question."
        )
        retrieved_ids = [
            str(item.get("chunk_id"))
            for item in result_dicts
            if item.get("chunk_id")
        ]
        cited_ids = []
    else:
        gate_status = "PASSED_PROVISIONAL_SCORE_GATE"

        retrieved_ids = [
            str(item.get("chunk_id"))
            for item in result_dicts
            if item.get("chunk_id")
        ]

        evidence_parts = []
        for item in result_dicts:
            chunk_id = item.get("chunk_id", "unknown_chunk")
            passage = get_passage(item)
            if passage:
                evidence_parts.append(f"[{chunk_id}]\n{passage}")

        context = "\n\n".join(evidence_parts)

        prompt = f"""
You answer questions using only the retrieved Indian legal corpus.

QUESTION:
{question}

RETRIEVED EVIDENCE:
{context}

RULES:
1. Answer only to the extent supported by the evidence.
2. Cite supporting passages using their exact chunk IDs, like [doc_123_chunk_0001].
3. Never invent a citation or cite an ID that is not in the evidence.
4. If the passages do not actually address the question, say that the
   available legal corpus does not provide sufficient relevant evidence.
5. Do not add unsupported legal rules, exceptions, or conclusions.
6. Be concise.
"""

        inputs = llm_tokenizer(
            prompt,
            return_tensors="pt",
            truncation=True,
            max_length=4096
        )
        inputs = {
            key: value.to(llm_model.device)
            for key, value in inputs.items()
        }

        llm_model.eval()
        with torch.no_grad():
            output = llm_model.generate(
                **inputs,
                max_new_tokens=180,
                do_sample=False,
                pad_token_id=llm_tokenizer.eos_token_id
            )

        new_tokens = output[0][inputs["input_ids"].shape[1]:]
        answer = llm_tokenizer.decode(
            new_tokens,
            skip_special_tokens=True
        ).strip()

        cited_ids = list(dict.fromkeys(
            re.findall(r"\[((?:doc_)[^\]\s]+)\]", answer)
        ))

    citation_ids_valid = (
        set(cited_ids).issubset(set(retrieved_ids))
    )

    gate_rows.append({
        "test_id": test_id,
        "question": question,
        "expected_type": expected_type,
        "top_reranker_score": top_score,
        "provisional_threshold": PROVISIONAL_THRESHOLD,
        "gate_status": gate_status,
        "answer": answer,
        "retrieved_chunk_ids": ";".join(retrieved_ids),
        "cited_chunk_ids": ";".join(cited_ids),
        "citation_ids_match_retrieval": citation_ids_valid,
    })

    print("Top score:", top_score)
    print("Gate:", gate_status)
    print("Answer:", answer)
    print("Cited IDs:", cited_ids)
    print("Citation IDs match retrieved IDs:", citation_ids_valid)

# Save a new versioned artifact; preserve previous outputs
output_dir = Path(
    "/kaggle/working/indian_legal_rag/09_Results_Figures"
)
output_dir.mkdir(parents=True, exist_ok=True)

gate_df = pd.DataFrame(gate_rows)
output_path = output_dir / "phase13_relevance_gate_test_v1.csv"
gate_df.to_csv(output_path, index=False)

print("\n" + "=" * 85)
print("RELEVANCE GATE TEST SUMMARY")
print(
    gate_df[
        ["test_id", "expected_type", "top_reranker_score", "gate_status"]
    ].to_string(index=False)
)
print("\nSaved:", output_path)
print("Questions tested:", len(gate_df))
print(
    "Unrelated questions blocked:",
    int(
        (
            (gate_df["expected_type"] == "unrelated")
            & gate_df["gate_status"].str.startswith("BLOCKED")
        ).sum()
    ),
    "of",
    int((gate_df["expected_type"] == "unrelated").sum())
)
print(
    "Relevant questions passed:",
    int(
        (
            (gate_df["expected_type"] == "relevant")
            & gate_df["gate_status"].eq("PASSED_PROVISIONAL_SCORE_GATE")
        ).sum()
    ),
    "of",
    int((gate_df["expected_type"] == "relevant").sum())
)
```

```python
# PHASE 13 — STEP 18: Test corrected answer generation

import re
import torch

question = "When can a dying declaration be the sole basis for conviction?"

# Retrieve evidence using the existing pipeline
results = retrieve_and_rerank(
    question,
    top_k=3,
    candidate_k=20
)

if isinstance(results, pd.DataFrame):
    items = results.to_dict("records")
else:
    items = list(results)

evidence_blocks = []
retrieved_ids = []

for item in items:
    if not isinstance(item, dict):
        item = item._asdict() if hasattr(item, "_asdict") else vars(item)

    chunk_id = item.get("chunk_id")
    passage = item.get("text") or item.get("content") or item.get("chunk")

    if chunk_id and passage:
        chunk_id = str(chunk_id)
        retrieved_ids.append(chunk_id)

        # Limit passage length to reduce prompt confusion
        evidence_blocks.append(
            f"SOURCE ID: {chunk_id}\n"
            f"PASSAGE:\n{str(passage)[:1800]}"
        )

evidence_context = "\n\n".join(evidence_blocks)

system_message = """
You are a concise question-answering assistant for an Indian legal document corpus.
Use only the supplied evidence.
Do not repeat these instructions.
Do not invent legal rules, facts, case holdings, or citations.
If the evidence is insufficient, say so clearly.
Cite each material legal statement with the exact source ID in square brackets.
Return only the answer, without headings about your process or numbered instructions.
"""

user_message = f"""
QUESTION:
{question}

EVIDENCE:
{evidence_context}

Write a concise answer in 2 to 4 sentences.
Use only claims supported by the passages.
Cite the relevant source IDs exactly as supplied.
"""

messages = [
    {"role": "system", "content": system_message.strip()},
    {"role": "user", "content": user_message.strip()},
]

# Use Qwen's instruction-tuned chat template
formatted_prompt = llm_tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)

inputs = llm_tokenizer(
    formatted_prompt,
    return_tensors="pt",
    truncation=True,
    max_length=4096
)

inputs = {
    key: value.to(llm_model.device)
    for key, value in inputs.items()
}

llm_model.eval()

with torch.no_grad():
    output = llm_model.generate(
        **inputs,
        max_new_tokens=180,
        do_sample=False,
        repetition_penalty=1.05,
        pad_token_id=llm_tokenizer.eos_token_id,
        eos_token_id=llm_tokenizer.eos_token_id
    )

new_tokens = output[0][inputs["input_ids"].shape[1]:]
corrected_answer = llm_tokenizer.decode(
    new_tokens,
    skip_special_tokens=True
).strip()

cited_ids = list(dict.fromkeys(
    re.findall(r"\[(doc_[^\]\s]+)\]", corrected_answer)
))

print("=" * 80)
print("QUESTION:", question)
print("\nANSWER:\n", corrected_answer)
print("\nRETRIEVED IDS:", retrieved_ids)
print("CITED IDS:", cited_ids)
print("CITATIONS MATCH RETRIEVED IDS:", bool(cited_ids) and set(cited_ids).issubset(set(retrieved_ids)))
print("ANSWER ENDS WITH PUNCTUATION:", corrected_answer.endswith((".", "!", "?")))
print("PROMPT LEAKAGE SUSPECTED:", bool(re.search(r"(?im)^\s*(?:\d+\.\s+)?(?:do not repeat these instructions|use your own words|avoid repetition|provide citations for)", corrected_answer)))
```

```python
# PHASE 13 — STEP 22: Consolidated end-to-end integration test

import re
import json
import pandas as pd
import torch
from pathlib import Path
from datetime import datetime, timezone

OUTPUT_DIR = Path(
    "/kaggle/working/indian_legal_rag/09_Results_Figures"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

RELEVANCE_THRESHOLD = 0.0  # Provisional; not a universally validated threshold

TEST_QUESTIONS = [
    {
        "test_id": "E2E_01",
        "question": "When can a dying declaration be the sole basis for conviction?",
        "expected": "legal"
    },
    {
        "test_id": "E2E_02",
        "question": "What is the right to a speedy trial under Article 21?",
        "expected": "legal"
    },
    {
        "test_id": "E2E_03",
        "question": "What is the burden of proving an alibi in a criminal case?",
        "expected": "legal"
    },
    {
        "test_id": "E2E_04",
        "question": "What ingredients are needed to bake a chocolate cake?",
        "expected": "unrelated"
    }
]

def normalize_result(item):
    if isinstance(item, dict):
        return item
    if hasattr(item, "_asdict"):
        return item._asdict()
    if hasattr(item, "__dict__"):
        return vars(item)
    return {}

def passage_text(item):
    for key in ("text", "content", "chunk"):
        value = item.get(key)
        if value:
            return str(value)
    return ""

def generate_answer_from_evidence(question, evidence_blocks):
    system_msg = (
        "You answer questions using only the supplied legal evidence. "
        "Never repeat instructions. Never invent legal rules or citations. "
        "If evidence is insufficient or irrelevant, say so. "
        "Cite material claims with exact source chunk IDs in square brackets. "
        "Return only the answer."
    )

    user_msg = (
        f"QUESTION:\n{question}\n\n"
        f"EVIDENCE:\n{evidence_blocks}\n\n"
        "Answer in 2 to 4 concise sentences. "
        "Use only claims supported by the evidence. "
        "Cite the exact source IDs for supported claims. "
        "If the passages do not answer the question, say the evidence is insufficient."
    )

    messages = [
        {"role": "system", "content": system_msg},
        {"role": "user", "content": user_msg}
    ]

    prompt = llm_tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = llm_tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=4096
    )
    inputs = {
        k: v.to(llm_model.device)
        for k, v in inputs.items()
    }

    llm_model.eval()
    with torch.no_grad():
        output = llm_model.generate(
            **inputs,
            max_new_tokens=180,
            do_sample=False,
            repetition_penalty=1.05,
            pad_token_id=llm_tokenizer.eos_token_id,
            eos_token_id=llm_tokenizer.eos_token_id
        )

    new_tokens = output[0][inputs["input_ids"].shape[1]:]
    return llm_tokenizer.decode(
        new_tokens,
        skip_special_tokens=True
    ).strip()

e2e_rows = []

for test in TEST_QUESTIONS:
    test_id = test["test_id"]
    question = test["question"]

    print("\n" + "=" * 85)
    print(f"{test_id}: {question}")

    try:
        retrieved = retrieve_and_rerank(
            question,
            top_k=5,
            candidate_k=20
        )

        if isinstance(retrieved, pd.DataFrame):
            result_items = retrieved.to_dict("records")
        else:
            result_items = list(retrieved)

        result_dicts = [
            normalize_result(item) for item in result_items
        ]

        scored_items = [
            item for item in result_dicts
            if item.get("reranker_score") is not None
        ]

        top_score = (
            max(float(item["reranker_score"]) for item in scored_items)
            if scored_items else None
        )

        retrieved_ids = [
            str(item["chunk_id"])
            for item in result_dicts
            if item.get("chunk_id")
        ]

        if top_score is None:
            status = "BLOCKED_NO_VALID_SCORE"
            answer = (
                "The system could not verify relevant evidence, "
                "so it cannot provide a supported answer."
            )
        elif top_score < RELEVANCE_THRESHOLD:
            status = "BLOCKED_LOW_RELEVANCE"
            answer = (
                "The available legal corpus does not provide sufficient "
                "relevant evidence to answer this question."
            )
        else:
            evidence_blocks = "\n\n".join(
                f"SOURCE ID: {item.get('chunk_id', 'unknown')}\n"
                f"PASSAGE:\n{passage_text(item)[:1800]}"
                for item in result_dicts
                if passage_text(item)
            )

            answer = generate_answer_from_evidence(
                question,
                evidence_blocks
            )
            status = "GENERATED"

        cited_ids = list(dict.fromkeys(
            re.findall(r"\[(doc_[^\]\s]+)\]", answer)
        ))

        citation_ids_valid = (
            bool(cited_ids)
            and set(cited_ids).issubset(set(retrieved_ids))
        ) if status == "GENERATED" else True

        prompt_leakage = bool(re.search(
            r"(?im)^\s*(?:\d+\.\s+)?"
            r"(?:do not repeat these instructions|use your own words|"
            r"avoid repetition|provide citations for|always start with)",
            answer
        ))

        incomplete = (
            status == "GENERATED"
            and (
                not answer.endswith((".", "!", "?"))
                or prompt_leakage
            )
        )

        if status == "GENERATED" and not cited_ids:
            citation_status = "MISSING_CITATIONS"
        elif status == "GENERATED" and not citation_ids_valid:
            citation_status = "INVALID_CITATION_IDS"
        elif status == "GENERATED":
            citation_status = "TRACEABLE_IDS_PRESENT_SUPPORT_REVIEW_REQUIRED"
        else:
            citation_status = "NOT_APPLICABLE_BLOCKED"

        print("Top reranker score:", top_score)
        print("Status:", status)
        print("Answer:", answer)
        print("Cited IDs:", cited_ids)
        print("Citation check:", citation_status)
        print("Possible incomplete/prompt-leakage output:", incomplete)

        e2e_rows.append({
            "test_id": test_id,
            "question": question,
            "expected_type": test["expected"],
            "top_reranker_score": top_score,
            "threshold": RELEVANCE_THRESHOLD,
            "status": status,
            "answer": answer,
            "retrieved_chunk_ids": ";".join(retrieved_ids),
            "cited_chunk_ids": ";".join(cited_ids),
            "citation_status": citation_status,
            "citation_ids_match_retrieval": citation_ids_valid,
            "possible_incomplete_or_prompt_leakage": incomplete,
            "execution_error": ""
        })

    except Exception as exc:
        print("TEST ERROR:", repr(exc))
        e2e_rows.append({
            "test_id": test_id,
            "question": question,
            "expected_type": test["expected"],
            "top_reranker_score": None,
            "threshold": RELEVANCE_THRESHOLD,
            "status": "ERROR",
            "answer": "",
            "retrieved_chunk_ids": "",
            "cited_chunk_ids": "",
            "citation_status": "NOT_CHECKED",
            "citation_ids_match_retrieval": False,
            "possible_incomplete_or_prompt_leakage": True,
            "execution_error": repr(exc)
        })

e2e_df = pd.DataFrame(e2e_rows)

csv_path = OUTPUT_DIR / "phase13_end_to_end_test_v1.csv"
json_path = OUTPUT_DIR / "phase13_end_to_end_report_v1.json"

e2e_df.to_csv(csv_path, index=False)

summary = {
    "phase": 13,
    "test_timestamp_utc": datetime.now(timezone.utc).isoformat(),
    "questions_tested": len(e2e_df),
    "generated_answers": int((e2e_df["status"] == "GENERATED").sum()),
    "blocked_questions": int(e2e_df["status"].str.startswith("BLOCKED").sum()),
    "errors": int((e2e_df["status"] == "ERROR").sum()),
    "relevance_threshold": RELEVANCE_THRESHOLD,
    "threshold_status": "PROVISIONAL_NOT_GENERALIZED",
    "citation_traceability_is_not_claim_level_support": True,
    "legal_review_status": "PENDING",
    "csv_path": str(csv_path),
    "report_path": str(json_path)
}

with open(json_path, "w", encoding="utf-8") as f:
    json.dump(summary, f, indent=2, ensure_ascii=False)

print("\n" + "=" * 85)
print("END-TO-END TEST SUMMARY")
print(
    e2e_df[
        [
            "test_id",
            "expected_type",
            "status",
            "citation_status",
            "possible_incomplete_or_prompt_leakage"
        ]
    ].to_string(index=False)
)
print("\nSaved results:", csv_path)
print("Saved report:", json_path)
print("\nNote: This is an integration test, not proof of legal correctness.")
```

```python
# IMPLEMENTATION STEP 1A
# Attach traceable source IDs to an answer after retrieving evidence.
# IMPORTANT: This checks citation traceability, not legal correctness.

import re

def attach_retrieved_citations(answer, retrieved_items, max_sources=3):
    """
    Append source IDs from retrieved passages to an answer.
    This does NOT prove that those passages support every claim.
    """
    if not answer or not answer.strip():
        raise ValueError("Answer is empty.")

    source_ids = []

    for item in retrieved_items:
        if isinstance(item, dict):
            chunk_id = item.get("chunk_id")
        elif hasattr(item, "_asdict"):
            chunk_id = item._asdict().get("chunk_id")
        else:
            chunk_id = getattr(item, "chunk_id", None)

        if chunk_id:
            source_ids.append(str(chunk_id))

    # Preserve order and remove duplicates.
    source_ids = list(dict.fromkeys(source_ids))[:max_sources]

    if not source_ids:
        raise ValueError("No retrieved source IDs are available.")

    # Remove any trailing punctuation before appending the source list.
    clean_answer = answer.strip()

    citation_text = " ".join(
        f"[{source_id}]" for source_id in source_ids
    )

    return f"{clean_answer.rstrip()} {citation_text}"


# Use the latest dying-declaration answer and retrieved results from Step 18.
test_answer = corrected_answer
test_retrieved_items = items

citation_test_answer = attach_retrieved_citations(
    test_answer,
    test_retrieved_items
)

print("ANSWER WITH TRACEABLE RETRIEVED SOURCES:\n")
print(citation_test_answer)

print("\nRETRIEVED SOURCE IDS:")
for item in test_retrieved_items:
    if isinstance(item, dict):
        print(item.get("chunk_id"))
    elif hasattr(item, "_asdict"):
        print(item._asdict().get("chunk_id"))
    else:
        print(getattr(item, "chunk_id", None))

print(
    "\nWARNING: These IDs identify retrieved passages. "
    "They are not yet verified claim-level citations."
)
```

```python
# IMPLEMENTATION STEP 1B
# Validate citation traceability against retrieved evidence.

import re

def validate_answer_citations(answer, retrieved_items):
    retrieved_ids = set()

    for item in retrieved_items:
        if isinstance(item, dict):
            chunk_id = item.get("chunk_id")
        elif hasattr(item, "_asdict"):
            chunk_id = item._asdict().get("chunk_id")
        else:
            chunk_id = getattr(item, "chunk_id", None)

        if chunk_id:
            retrieved_ids.add(str(chunk_id))

    cited_ids = set(
        re.findall(r"\[(doc_[^\]]+)\]", answer)
    )

    missing_citations = not bool(cited_ids)
    invalid_ids = cited_ids - retrieved_ids

    result = {
        "has_citations": not missing_citations,
        "cited_ids": sorted(cited_ids),
        "invalid_citation_ids": sorted(invalid_ids),
        "citation_traceability_pass": (
            not missing_citations and not invalid_ids
        ),
        "claim_support_verified": False
    }

    return result


citation_validation = validate_answer_citations(
    citation_test_answer,
    test_retrieved_items
)

print("CITATION VALIDATION")
for key, value in citation_validation.items():
    print(f"{key}: {value}")
```

```python
# IMPLEMENTATION STEP 1C
# Prepare retrieved evidence for claim-level citation generation.

def prepare_citation_evidence(retrieved_items):
    evidence = []

    for item in retrieved_items:
        if isinstance(item, dict):
            chunk_id = item.get("chunk_id")
            passage = item.get("text", "")
        elif hasattr(item, "_asdict"):
            row = item._asdict()
            chunk_id = row.get("chunk_id")
            passage = row.get("text", "")
        else:
            chunk_id = getattr(item, "chunk_id", None)
            passage = getattr(item, "text", "")

        if chunk_id and passage:
            evidence.append({
                "chunk_id": str(chunk_id),
                "text": str(passage)
            })

    return evidence


citation_evidence = prepare_citation_evidence(
    test_retrieved_items
)

print("Retrieved passages prepared:", len(citation_evidence))

for i, passage in enumerate(citation_evidence, 1):
    print(f"\n--- Evidence {i} ---")
    print("Chunk ID:", passage["chunk_id"])
    print("Passage preview:", passage["text"][:350])
```

```python
# IMPLEMENTATION STEP 1D
# Generate an answer with claim-level citations using local Qwen.

def generate_cited_answer(question, evidence):
    evidence_text = "\n\n".join(
        f"[SOURCE ID: {item['chunk_id']}]\n{item['text']}"
        for item in evidence
    )

    system_prompt = """
You are an Indian legal research assistant.

Answer the question using ONLY the supplied evidence.

Rules:
1. Cite each substantive legal claim with the exact source ID
   that supports that claim, using [chunk_id] format.
2. Use only source IDs supplied in the evidence.
3. Do not invent cases, quotations, sections, or legal principles.
4. If the evidence does not support a claim, omit that claim.
5. If the evidence is insufficient, clearly say so.
6. Keep the answer concise.
7. Do not claim that an authority was independently verified.
"""

    user_prompt = f"""
QUESTION:
{question}

RETRIEVED EVIDENCE:
{evidence_text}

Write the answer with claim-level citations.
"""

    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt}
    ]

    prompt_text = llm_tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = llm_tokenizer(
        prompt_text,
        return_tensors="pt"
    ).to(llm_model.device)

    import torch

    with torch.no_grad():
        output = llm_model.generate(
            **inputs,
            max_new_tokens=300,
            do_sample=False,
            pad_token_id=llm_tokenizer.eos_token_id
        )

    generated_tokens = output[0][inputs["input_ids"].shape[1]:]

    answer = llm_tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    ).strip()

    # Validate source-ID traceability.
    validation = validate_answer_citations(answer, evidence)

    return answer, validation


test_question = (
    "When can a dying declaration be the sole basis for conviction?"
)

cited_answer, cited_answer_validation = generate_cited_answer(
    test_question,
    citation_evidence
)

print("GENERATED ANSWER:\n")
print(cited_answer)

print("\nCITATION TRACEABILITY:")
for key, value in cited_answer_validation.items():
    print(f"{key}: {value}")
```

```python
# IMPLEMENTATION STEP 1E
# Repair missing citations using the existing answer and evidence.

def repair_missing_citations(question, answer, evidence):
    import torch

    evidence_text = "\n\n".join(
        f"[{item['chunk_id']}]\n{item['text']}"
        for item in evidence
    )

    allowed_ids = {item["chunk_id"] for item in evidence}

    messages = [
        {
            "role": "system",
            "content": (
                "You are an Indian legal research assistant. "
                "Revise the supplied answer using only the supplied evidence. "
                "Add inline citations in the exact format [chunk_id] "
                "to every substantive claim, using only IDs in the evidence. "
                "Cite a claim only when the cited passage supports it. "
                "Remove unsupported claims rather than guessing. "
                "Do not add new legal principles or authorities. "
                "Return only the revised answer."
            )
        },
        {
            "role": "user",
            "content": (
                f"QUESTION:\n{question}\n\n"
                f"EXISTING ANSWER:\n{answer}\n\n"
                f"ALLOWED SOURCE IDS:\n{sorted(allowed_ids)}\n\n"
                f"EVIDENCE:\n{evidence_text}"
            )
        }
    ]

    prompt = llm_tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = llm_tokenizer(
        prompt,
        return_tensors="pt"
    ).to(llm_model.device)

    with torch.no_grad():
        output = llm_model.generate(
            **inputs,
            max_new_tokens=300,
            do_sample=False,
            pad_token_id=llm_tokenizer.eos_token_id
        )

    new_tokens = output[0][inputs["input_ids"].shape[1]:]
    revised_answer = llm_tokenizer.decode(
        new_tokens,
        skip_special_tokens=True
    ).strip()

    validation = validate_answer_citations(
        revised_answer,
        evidence
    )

    # Additional check: citations must be from the allowed evidence.
    import re
    cited_ids = set(
        re.findall(r"\[(doc_[^\]]+)\]", revised_answer)
    )

    validation["only_allowed_ids_used"] = cited_ids.issubset(allowed_ids)

    # Do not mark the answer as legally verified automatically.
    validation["claim_support_verified"] = False

    validation["ready_for_manual_review"] = (
        validation["citation_traceability_pass"]
        and validation["only_allowed_ids_used"]
    )

    return revised_answer, validation


repaired_answer, repaired_validation = repair_missing_citations(
    test_question,
    cited_answer,
    citation_evidence
)

print("REVISED ANSWER:\n")
print(repaired_answer)

print("\nVALIDATION RESULTS:")
for key, value in repaired_validation.items():
    print(f"{key}: {value}")
```

```python
# IMPLEMENTATION STEP 1F
# Deterministic fallback for missing citations.
# Uses only a manually reviewed answer and its reviewed source IDs.

def safe_citation_fallback(answer, verified_source_ids, retrieved_items):
    retrieved_ids = set()

    for item in retrieved_items:
        if isinstance(item, dict):
            cid = item.get("chunk_id")
        elif hasattr(item, "_asdict"):
            cid = item._asdict().get("chunk_id")
        else:
            cid = getattr(item, "chunk_id", None)

        if cid:
            retrieved_ids.add(str(cid))

    verified_source_ids = list(dict.fromkeys(verified_source_ids))

    # Refuse to attach a source that was not retrieved.
    invalid = [
        cid for cid in verified_source_ids
        if cid not in retrieved_ids
    ]

    if invalid:
        return {
            "status": "BLOCKED",
            "answer": None,
            "reason": f"Source IDs were not retrieved: {invalid}"
        }

    if not verified_source_ids:
        return {
            "status": "BLOCKED",
            "answer": None,
            "reason": "No manually verified source IDs were provided."
        }

    # This fallback is intended for an answer already reviewed
    # against the specified passages by a human.
    citations = " ".join(
        f"[{cid}]" for cid in verified_source_ids
    )

    return {
        "status": "READY_FOR_REVIEW",
        "answer": f"{answer.strip()} {citations}",
        "reason": (
            "Citation IDs are traceable. Claim-level support must "
            "have been manually checked before supplying these IDs."
        )
    }


# Use the previously reviewed answer, not Qwen's unverified rewrite.
manually_reviewed_answer = (
    "A dying declaration can be the sole basis for conviction if it "
    "inspires the full confidence of the court, the deceased was in a "
    "fit state of mind when making the statement, and it was not the "
    "result of tutoring, prompting, or imagination."
)

reviewed_source_ids = ["doc_1060_v2_chunk_0010"]

fallback_result = safe_citation_fallback(
    manually_reviewed_answer,
    reviewed_source_ids,
    citation_evidence
)

print("STATUS:", fallback_result["status"])
print("ANSWER:\n", fallback_result["answer"])
print("NOTE:", fallback_result["reason"])
```

```python
import re

def validate_rag_answer(answer, retrieved_items):
    """Validate answer structure and citation traceability."""

    if not isinstance(answer, str) or not answer.strip():
        return {
            "status": "BLOCKED",
            "missing_citations": True,
            "invalid_citations": [],
            "empty_answer": True,
            "possible_truncation": False,
            "prompt_leakage": False
        }

    retrieved_ids = set()

    for item in retrieved_items:
        if isinstance(item, dict):
            cid = item.get("chunk_id")
        elif hasattr(item, "_asdict"):
            cid = item._asdict().get("chunk_id")
        else:
            cid = getattr(item, "chunk_id", None)

        if cid:
            retrieved_ids.add(str(cid))

    cited_ids = set(re.findall(r"\[(doc_[^\]]+)\]", answer))

    invalid_ids = cited_ids - retrieved_ids

    # Heuristic checks, not definitive proof of truncation or leakage.
    suspicious_endings = (
        "because", "therefore", "such as", "including",
        "as well as", "and", "or", "the"
    )

    possible_truncation = (
        answer.rstrip().endswith("...")
        or answer.strip().lower().endswith(suspicious_endings)
    )

    leakage_patterns = [
        r"you are an? .*assistant",
        r"follow these rules",
        r"return only the revised answer",
        r"retrieved evidence:",
        r"write the answer with claim-level citations",
        r"allowed source ids:"
    ]

    prompt_leakage = any(
        re.search(pattern, answer, re.IGNORECASE)
        for pattern in leakage_patterns
    )

    missing_citations = len(cited_ids) == 0
    empty_answer = False

    passed = (
        not missing_citations
        and not invalid_ids
        and not empty_answer
        and not possible_truncation
        and not prompt_leakage
    )

    return {
        "status": "PASS" if passed else "REVIEW_REQUIRED",
        "missing_citations": missing_citations,
        "cited_ids": sorted(cited_ids),
        "invalid_citations": sorted(invalid_ids),
        "empty_answer": empty_answer,
        "possible_truncation": possible_truncation,
        "prompt_leakage": prompt_leakage
    }


# Test using the answer and evidence already prepared in Step 1.
validation_result = validate_rag_answer(
    fallback_result["answer"],
    citation_evidence
)

print("ANSWER VALIDATION RESULTS")
for key, value in validation_result.items():
    print(f"{key}: {value}")
```

```python
# PHASE 13 — STEP 2B
# Test the validator using deliberately faulty answers.

faulty_test_cases = {
    "TEST_1_MISSING_CITATION": (
        "A dying declaration may be the sole basis for conviction."
    ),
    "TEST_2_INVALID_CITATION": (
        "A dying declaration may be the sole basis for conviction "
        "[doc_invalid_chunk_9999]."
    ),
    "TEST_3_PROMPT_LEAKAGE": (
        "Follow these rules and return only the revised answer "
        "[doc_1060_v2_chunk_0010]."
    ),
    "TEST_4_POSSIBLE_TRUNCATION": (
        "A dying declaration may be the sole basis for conviction "
        "because [doc_1060_v2_chunk_0010]"
    )
}

faulty_test_results = []

for test_name, test_answer in faulty_test_cases.items():
    result = validate_rag_answer(
        test_answer,
        citation_evidence
    )

    if test_name == "TEST_1_MISSING_CITATION":
        detected = result["missing_citations"]

    elif test_name == "TEST_2_INVALID_CITATION":
        detected = bool(result["invalid_citations"])

    elif test_name == "TEST_3_PROMPT_LEAKAGE":
        detected = result["prompt_leakage"]

    elif test_name == "TEST_4_POSSIBLE_TRUNCATION":
        detected = result["possible_truncation"]

    faulty_test_results.append({
        "test_name": test_name,
        "expected_detection": True,
        "actual_detection": detected,
        "test_pass": detected
    })

import pandas as pd

faulty_test_df = pd.DataFrame(faulty_test_results)

display(faulty_test_df)

print(
    "\nTests passed:",
    int(faulty_test_df["test_pass"].sum()),
    "/",
    len(faulty_test_df)
)
```

```python
# PHASE 13 — STEP 2C
# Improve detection of answers ending with an incomplete phrase.

import re

def detect_possible_truncation(answer):
    if not isinstance(answer, str) or not answer.strip():
        return True

    text = answer.strip()

    # Remove a trailing citation before checking the final phrase.
    text = re.sub(r"(?:\s*\[doc_[^\]]+\])+\s*$", "", text).strip()

    if not text:
        return True

    # Explicit truncation markers.
    if text.endswith(("...", "…")):
        return True

    # Common incomplete endings.
    incomplete_endings = (
        "because", "therefore", "such as",
        "including", "as well as", "and", "or", "the"
    )

    return text.rstrip(".,;:!?").lower().endswith(incomplete_endings)


# Re-test the previously failing example.
truncated_answer = (
    "A dying declaration may be the sole basis for conviction "
    "because [doc_1060_v2_chunk_0010]"
)

print(
    "Truncation detected:",
    detect_possible_truncation(truncated_answer)
)

print(
    "Complete answer detected as truncated:",
    detect_possible_truncation(
        "A dying declaration may be the sole basis for conviction "
        "if the court has full confidence in the statement."
    )
)
```

```python
# PHASE 13 — STEP 2D
# Integrate the improved truncation detector and rerun tests.

import re
import pandas as pd

def validate_rag_answer(answer, retrieved_items):
    if not isinstance(answer, str) or not answer.strip():
        return {
            "status": "BLOCKED",
            "missing_citations": True,
            "cited_ids": [],
            "invalid_citations": [],
            "empty_answer": True,
            "possible_truncation": False,
            "prompt_leakage": False
        }

    retrieved_ids = set()

    for item in retrieved_items:
        if isinstance(item, dict):
            cid = item.get("chunk_id")
        elif hasattr(item, "_asdict"):
            cid = item._asdict().get("chunk_id")
        else:
            cid = getattr(item, "chunk_id", None)

        if cid:
            retrieved_ids.add(str(cid))

    cited_ids = set(re.findall(r"\[(doc_[^\]]+)\]", answer))
    invalid_ids = cited_ids - retrieved_ids

    prompt_patterns = [
        r"you are an? .*assistant",
        r"follow these rules",
        r"return only the revised answer",
        r"retrieved evidence:",
        r"write the answer with claim-level citations",
        r"allowed source ids:"
    ]

    prompt_leakage = any(
        re.search(pattern, answer, re.IGNORECASE)
        for pattern in prompt_patterns
    )

    missing_citations = not bool(cited_ids)
    possible_truncation = detect_possible_truncation(answer)

    passed = (
        not missing_citations
        and not invalid_ids
        and not possible_truncation
        and not prompt_leakage
    )

    return {
        "status": "PASS" if passed else "REVIEW_REQUIRED",
        "missing_citations": missing_citations,
        "cited_ids": sorted(cited_ids),
        "invalid_citations": sorted(invalid_ids),
        "empty_answer": False,
        "possible_truncation": possible_truncation,
        "prompt_leakage": prompt_leakage
    }


# Repeat all four deliberately faulty tests.
faulty_test_results = []

for test_name, test_answer in faulty_test_cases.items():
    result = validate_rag_answer(test_answer, citation_evidence)

    if test_name == "TEST_1_MISSING_CITATION":
        detected = result["missing_citations"]
    elif test_name == "TEST_2_INVALID_CITATION":
        detected = bool(result["invalid_citations"])
    elif test_name == "TEST_3_PROMPT_LEAKAGE":
        detected = result["prompt_leakage"]
    else:
        detected = result["possible_truncation"]

    faulty_test_results.append({
        "test_name": test_name,
        "expected_detection": True,
        "actual_detection": detected,
        "test_pass": detected
    })

faulty_test_df = pd.DataFrame(faulty_test_results)
display(faulty_test_df)

print(
    f"\nTests passed: {faulty_test_df['test_pass'].sum()}"
    f"/{len(faulty_test_df)}"
)
```

```python
# PHASE 13 — STEP 3E
# Evaluate the existing relevance threshold on 10 new questions.

from pathlib import Path
import pandas as pd

results_dir = Path(
    "/kaggle/working/indian_legal_rag/09_Results_Figures"
)

test_path = results_dir / "phase13_relevance_expanded_test_set_v1.csv"
test_df = pd.read_csv(test_path)

threshold = float(RELEVANCE_THRESHOLD)
results = []

for _, row in test_df.iterrows():
    question = row["question"]

    try:
        retrieved = retrieve_and_rerank(
            question,
            top_k=5,
            candidate_k=20
        )

        # Support common return formats without assuming
        # that the first returned item is a dictionary.
        if isinstance(retrieved, pd.DataFrame):
            if retrieved.empty:
                top_score = None
                chunk_ids = []
            else:
                score_col = next(
                    (c for c in ["reranker_score", "score", "cross_encoder_score"]
                     if c in retrieved.columns),
                    None
                )
                top_score = (
                    float(retrieved.iloc[0][score_col])
                    if score_col else None
                )
                id_col = next(
                    (c for c in ["chunk_id", "id"] if c in retrieved.columns),
                    None
                )
                chunk_ids = (
                    retrieved[id_col].astype(str).tolist()
                    if id_col else []
                )

        elif isinstance(retrieved, (list, tuple)) and retrieved:
            first = retrieved[0]

            if isinstance(first, dict):
                top_score = next(
                    (
                        float(first[k])
                        for k in ["reranker_score", "score", "cross_encoder_score"]
                        if k in first and first[k] is not None
                    ),
                    None
                )
                chunk_ids = [
                    str(item.get("chunk_id"))
                    for item in retrieved
                    if isinstance(item, dict) and item.get("chunk_id")
                ]

            elif isinstance(first, (list, tuple)) and len(first) >= 2:
                top_score = float(first[1])
                chunk_ids = []

            else:
                top_score = None
                chunk_ids = []

        else:
            top_score = None
            chunk_ids = []

        if top_score is None:
            status = "SCORE_EXTRACTION_FAILED"
        else:
            status = (
                "PASSED_PROVISIONAL_SCORE_GATE"
                if top_score >= threshold
                else "BLOCKED_LOW_RELEVANCE_SCORE"
            )

        results.append({
            "test_id": row["test_id"],
            "question": question,
            "expected_type": row["expected_type"],
            "topic": row["topic"],
            "top_reranker_score": top_score,
            "threshold": threshold,
            "gate_status": status,
            "retrieved_chunk_ids": ";".join(chunk_ids)
        })

        print(
            f"{row['test_id']}: score={top_score}, "
            f"status={status}"
        )

    except Exception as exc:
        results.append({
            "test_id": row["test_id"],
            "question": question,
            "expected_type": row["expected_type"],
            "topic": row["topic"],
            "top_reranker_score": None,
            "threshold": threshold,
            "gate_status": "ERROR",
            "retrieved_chunk_ids": "",
            "error": str(exc)
        })
        print(f"{row['test_id']}: ERROR — {exc}")

score_df = pd.DataFrame(results)

output_path = results_dir / "phase13_relevance_expanded_scores_v1.csv"
score_df.to_csv(output_path, index=False)

print("\nSaved:", output_path)
display(score_df)
```

```python
# PHASE 13 — STEP 3F
# Calculate expanded relevance-gate metrics and save the report.

from pathlib import Path
import json
import pandas as pd
from datetime import datetime, timezone

results_dir = Path(
    "/kaggle/working/indian_legal_rag/09_Results_Figures"
)

scores_path = results_dir / "phase13_relevance_expanded_scores_v1.csv"
score_df = pd.read_csv(scores_path)

# Exclude errors and missing scores from the classification metrics.
valid = score_df[
    score_df["top_reranker_score"].notna()
    & score_df["gate_status"].isin([
        "PASSED_PROVISIONAL_SCORE_GATE",
        "BLOCKED_LOW_RELEVANCE_SCORE"
    ])
].copy()

valid["predicted_type"] = valid["gate_status"].map({
    "PASSED_PROVISIONAL_SCORE_GATE": "relevant",
    "BLOCKED_LOW_RELEVANCE_SCORE": "unrelated"
})

relevant = valid["expected_type"] == "relevant"
predicted_relevant = valid["predicted_type"] == "relevant"

tp = int((relevant & predicted_relevant).sum())
tn = int((~relevant & ~predicted_relevant).sum())
fp = int((~relevant & predicted_relevant).sum())
fn = int((relevant & ~predicted_relevant).sum())

report = {
    "phase": 13,
    "step": "Step 3 - Relevance Gate Validation",
    "threshold": float(RELEVANCE_THRESHOLD),
    "threshold_status": "PROVISIONAL_NOT_GENERALIZED",
    "questions_total": int(len(score_df)),
    "questions_scored": int(len(valid)),
    "errors_or_unscored": int(len(score_df) - len(valid)),
    "true_positives": tp,
    "true_negatives": tn,
    "false_positives": fp,
    "false_negatives": fn,
    "accuracy": (tp + tn) / len(valid) if len(valid) else None,
    "precision": tp / (tp + fp) if (tp + fp) else None,
    "recall": tp / (tp + fn) if (tp + fn) else None,
    "specificity": tn / (tn + fp) if (tn + fp) else None,
    "timestamp_utc": datetime.now(timezone.utc).isoformat(),
    "limitations": [
        "The evaluation uses only ten newly selected questions.",
        "The test set is not a representative sample of all possible queries.",
        "The threshold is not independently validated or calibrated.",
        "Reranker scores are relevance signals, not legal correctness scores."
    ]
}

report_path = results_dir / "phase13_relevance_expanded_report_v1.json"
with open(report_path, "w", encoding="utf-8") as f:
    json.dump(report, f, indent=2, ensure_ascii=False)

metrics_df = pd.DataFrame([{
    "threshold": report["threshold"],
    "questions_scored": report["questions_scored"],
    "true_positives": tp,
    "true_negatives": tn,
    "false_positives": fp,
    "false_negatives": fn,
    "accuracy": report["accuracy"],
    "precision": report["precision"],
    "recall": report["recall"],
    "specificity": report["specificity"]
}])

metrics_path = results_dir / "phase13_relevance_expanded_metrics_v1.csv"
metrics_df.to_csv(metrics_path, index=False)

display(metrics_df)

print("\nSaved report:", report_path)
print("Saved metrics:", metrics_path)
print("Threshold remains:", RELEVANCE_THRESHOLD)
```

```python
# PHASE 13 — STEP 4B
# Validate existing corrected answers without regenerating them.

from pathlib import Path
import pandas as pd
import re

results_dir = Path(
    "/kaggle/working/indian_legal_rag/09_Results_Figures"
)

corrected_path = results_dir / "phase13_corrected_answers_v1.csv"
corrected_answers_df = pd.read_csv(corrected_path)

integration_results = []

for _, row in corrected_answers_df.iterrows():
    answer = str(row["answer"])
    
    # Source IDs recorded during the previous evidence review.
    expected_ids = {
        cid.strip()
        for cid in str(row["source_chunk_ids"]).split(";")
        if cid.strip() and cid.strip().lower() != "nan"
    }

    # IDs actually cited in the answer.
    actual_ids = set(re.findall(r"\[(doc_[^\]]+)\]", answer))

    missing_citations = len(actual_ids) == 0
    unexpected_ids = actual_ids - expected_ids
    citation_match = bool(actual_ids) and not unexpected_ids

    # Run the structural validator against the recorded source IDs.
    synthetic_evidence = [
        {"chunk_id": cid, "text": "Previously reviewed source passage."}
        for cid in expected_ids
    ]

    structural_result = validate_rag_answer(
        answer,
        synthetic_evidence
    )

    integration_results.append({
        "test_id": row["test_id"],
        "citations_present": not missing_citations,
        "unexpected_citation_ids": ";".join(sorted(unexpected_ids)),
        "citation_ids_match_recorded_sources": citation_match,
        "structural_validation": structural_result["status"],
        "source_support_review": row["source_support_review"],
        "legal_review_status": row["legal_review_status"]
    })

integration_df = pd.DataFrame(integration_results)

display(integration_df)

print("\nAnswers tested:", len(integration_df))
print(
    "Citation checks passed:",
    int(integration_df["citation_ids_match_recorded_sources"].sum()),
    "/",
    len(integration_df)
)
print(
    "Structural checks passed:",
    int((integration_df["structural_validation"] == "PASS").sum()),
    "/",
    len(integration_df)
)

output_path = results_dir / "phase13_final_integration_test_v1.csv"
integration_df.to_csv(output_path, index=False)

print("\nSaved:", output_path)
```

```python
# PHASE 13 — STEP 4C.1
# End-to-end smoke test: relevance gate -> retrieval -> Qwen -> validation
# No components are rebuilt and no existing artifacts are overwritten.

from pathlib import Path
import pandas as pd
import re
import torch

results_dir = Path(
    "/kaggle/working/indian_legal_rag/09_Results_Figures"
)
results_dir.mkdir(parents=True, exist_ok=True)

THRESHOLD = float(RELEVANCE_THRESHOLD)


def run_rag_smoke_test(test_id, question, expected_type):
    result = {
        "test_id": test_id,
        "question": question,
        "expected_type": expected_type,
        "threshold": THRESHOLD,
        "top_reranker_score": None,
        "gate_status": None,
        "answer": None,
        "cited_chunk_ids": "",
        "citation_traceability": "NOT_CHECKED",
        "structural_validation": "NOT_CHECKED",
        "claim_support_review": "PENDING",
        "error": ""
    }

    try:
        # 1. Retrieve and rerank using existing components.
        retrieved = retrieve_and_rerank(
            question,
            top_k=5,
            candidate_k=20
        )

        # Normalize common result formats.
        if isinstance(retrieved, pd.DataFrame):
            records = retrieved.to_dict(orient="records")
        elif isinstance(retrieved, (list, tuple)):
            records = []
            for item in retrieved:
                if isinstance(item, dict):
                    records.append(item)
                elif hasattr(item, "_asdict"):
                    records.append(item._asdict())
                else:
                    raise TypeError(
                        f"Unsupported retrieval result item: {type(item)}"
                    )
        else:
            raise TypeError(
                f"Unsupported retrieval result: {type(retrieved)}"
            )

        if not records:
            result["gate_status"] = "BLOCKED_NO_EVIDENCE"
            return result

        # Find the reranker score field.
        score_keys = (
            "reranker_score",
            "cross_encoder_score",
            "score"
        )

        first = records[0]
        score_key = next(
            (key for key in score_keys
             if first.get(key) is not None),
            None
        )

        if score_key is None:
            raise ValueError(
                "Could not identify reranker score in retrieval results. "
                f"Available fields: {list(first.keys())}"
            )

        top_score = float(first[score_key])
        result["top_reranker_score"] = top_score

        # 2. Apply the provisional relevance gate.
        if top_score < THRESHOLD:
            result["gate_status"] = "BLOCKED_LOW_RELEVANCE_SCORE"
            result["answer"] = (
                "The available legal corpus does not provide sufficient "
                "relevant evidence to answer this question."
            )
            result["structural_validation"] = "BLOCKED"
            return result

        result["gate_status"] = "PASSED_PROVISIONAL_SCORE_GATE"

        # 3. Prepare evidence from the actual retrieved passages.
        evidence = []

        for item in records:
            chunk_id = item.get("chunk_id")
            passage_text = item.get("text")

            if chunk_id and passage_text:
                evidence.append({
                    "chunk_id": str(chunk_id),
                    "text": str(passage_text)
                })

        if not evidence:
            raise ValueError(
                "Retrieved results did not include chunk IDs and passage text."
            )

        # 4. Generate an answer with the existing local Qwen model.
        answer, generation_validation = generate_cited_answer(
            question,
            evidence
        )

        # 5. Repair missing citations if needed.
        if not generation_validation["citation_traceability_pass"]:
            answer, generation_validation = repair_missing_citations(
                question,
                answer,
                evidence
            )

        result["answer"] = answer

        # 6. Validate the answer against the actual retrieved evidence.
        structural = validate_rag_answer(answer, evidence)

        cited_ids = set(
            re.findall(r"\[(doc_[^\]]+)\]", answer)
        )
        retrieved_ids = {
            item["chunk_id"] for item in evidence
        }

        result["cited_chunk_ids"] = ";".join(sorted(cited_ids))

        traceability_pass = (
            bool(cited_ids)
            and cited_ids.issubset(retrieved_ids)
        )

        result["citation_traceability"] = (
            "PASS" if traceability_pass else "FAIL"
        )

        result["structural_validation"] = structural["status"]

        # Deliberately do not mark claim support as automatically verified.
        result["claim_support_review"] = "PENDING_MANUAL_REVIEW"

    except Exception as exc:
        result["gate_status"] = result["gate_status"] or "ERROR"
        result["error"] = f"{type(exc).__name__}: {exc}"

    return result


smoke_test_cases = [
    {
        "test_id": "FINAL_SMOKE_LEGAL",
        "question": (
            "When can a dying declaration be the sole basis for conviction?"
        ),
        "expected_type": "relevant"
    },
    {
        "test_id": "FINAL_SMOKE_UNRELATED",
        "question": (
            "How can I bake a chocolate cake without an oven?"
        ),
        "expected_type": "unrelated"
    }
]

smoke_results = []

for case in smoke_test_cases:
    print(f"\nRunning {case['test_id']}...")
    outcome = run_rag_smoke_test(**case)
    smoke_results.append(outcome)

    print("Gate:", outcome["gate_status"])
    print("Top score:", outcome["top_reranker_score"])
    print("Citation traceability:", outcome["citation_traceability"])
    print("Structural validation:", outcome["structural_validation"])

    if outcome["error"]:
        print("Error:", outcome["error"])

smoke_df = pd.DataFrame(smoke_results)

output_path = (
    results_dir / "phase13_final_smoke_test_v1.csv"
)
smoke_df.to_csv(output_path, index=False)

print("\nFINAL SMOKE TEST RESULTS")
display(
    smoke_df[
        [
            "test_id",
            "expected_type",
            "top_reranker_score",
            "gate_status",
            "citation_traceability",
            "structural_validation",
            "claim_support_review",
            "error"
        ]
    ]
)

print("\nSaved:", output_path)
```

```python

import os
import re
import json
import numpy as np
import pandas as pd
from datetime import datetime, timezone

SOURCE_DIR = "/kaggle/working/AILQA/Dataset_For_Context"
PROJECT_DIR = "/kaggle/working/indian_legal_rag"
CORPUS_DIR = os.path.join(PROJECT_DIR, "04_Legal_Corpus")
RESULTS_DIR = os.path.join(PROJECT_DIR, "09_Results_Figures")

os.makedirs(RESULTS_DIR, exist_ok=True)

# Load existing structure-aware chunks without modifying them.
structured_path = os.path.join(
    CORPUS_DIR, "phase5_legal_chunks_v2.csv"
)
structured_df = pd.read_csv(structured_path)

def fixed_length_chunks(text, chunk_size=1600, overlap=200):
    """Fixed-character baseline with controlled overlap."""
    text = str(text)
    if not text:
        return []

    step = chunk_size - overlap
    return [
        text[start:start + chunk_size]
        for start in range(0, len(text), step)
        if text[start:start + chunk_size].strip()
    ]

def paragraph_boundary_rate(chunks):
    """Fraction of chunk boundaries that coincide with paragraph boundaries."""
    if len(chunks) <= 1:
        return np.nan

    boundaries = []
    for chunk in chunks[:-1]:
        boundaries.append(chunk.endswith("\n") or chunk.endswith("\n\n"))

    return float(np.mean(boundaries))

fixed_lengths = []
fixed_boundary_rates = []
fixed_total = 0
documents_processed = 0
missing_documents = []

for filename in sorted(os.listdir(SOURCE_DIR)):
    if not filename.lower().endswith(".txt"):
        continue

    filepath = os.path.join(SOURCE_DIR, filename)

    try:
        with open(filepath, "r", encoding="utf-8", errors="replace") as f:
            text = f.read()
    except Exception:
        missing_documents.append(filename)
        continue

    chunks = fixed_length_chunks(text)
    fixed_total += len(chunks)
    fixed_lengths.extend(len(c) for c in chunks)
    fixed_boundary_rates.append(paragraph_boundary_rate(chunks))
    documents_processed += 1

# Existing structure-aware chunk statistics.
structured_lengths = structured_df["char_count"].dropna().astype(int).tolist()

# Boundary preservation estimate from saved chunk text.
structured_boundary_rates = []

for _, group in structured_df.groupby("document_id", sort=False):
    group = group.sort_values("chunk_index")
    texts = group["text"].astype(str).tolist()
    structured_boundary_rates.append(paragraph_boundary_rate(texts))

comparison = pd.DataFrame([
    {
        "method": "Fixed-length",
        "documents_processed": documents_processed,
        "total_chunks": fixed_total,
        "mean_chunk_chars": float(np.mean(fixed_lengths)),
        "median_chunk_chars": float(np.median(fixed_lengths)),
        "min_chunk_chars": int(np.min(fixed_lengths)),
        "max_chunk_chars": int(np.max(fixed_lengths)),
        "mean_paragraph_boundary_rate": float(
            np.nanmean(fixed_boundary_rates)
        ),
    },
    {
        "method": "Structure-aware (existing)",
        "documents_processed": int(structured_df["document_id"].nunique()),
        "total_chunks": len(structured_df),
        "mean_chunk_chars": float(np.mean(structured_lengths)),
        "median_chunk_chars": float(np.median(structured_lengths)),
        "min_chunk_chars": int(np.min(structured_lengths)),
        "max_chunk_chars": int(np.max(structured_lengths)),
        "mean_paragraph_boundary_rate": float(
            np.nanmean(structured_boundary_rates)
        ),
    }
])

csv_path = os.path.join(
    RESULTS_DIR, "phase5_chunking_comparison_v1.csv"
)
json_path = os.path.join(
    RESULTS_DIR, "phase5_chunking_comparison_manifest_v1.json"
)

comparison.to_csv(csv_path, index=False)

manifest = {
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "source_directory": SOURCE_DIR,
    "fixed_length_parameters": {
        "chunk_size_chars": 1600,
        "overlap_chars": 200
    },
    "structure_aware_source": structured_path,
    "documents_processed": documents_processed,
    "failed_documents": missing_documents,
    "note": (
        "This is a chunking-statistics comparison. Boundary rates are "
        "approximate and do not establish legal-structure preservation "
        "or retrieval-quality improvement. Retrieval evaluation is a "
        "separate experiment."
    ),
    "results_csv": csv_path
}

with open(json_path, "w", encoding="utf-8") as f:
    json.dump(manifest, f, indent=2)

print("CHUNKING COMPARISON")
display(comparison)

print("\nDocuments processed:", documents_processed)
print("Documents with read errors:", len(missing_documents))
print("Saved CSV:", csv_path)
print("Saved manifest:", json_path)
```

```python
import os
import re
import pandas as pd

SOURCE_DIR = "/kaggle/working/AILQA/Dataset_For_Context"
SAMPLE_SIZE = 500

files = sorted(
    f for f in os.listdir(SOURCE_DIR)
    if f.lower().endswith(".txt")
)[:SAMPLE_SIZE]

patterns = {
    "supreme_court": re.compile(
        r"supreme court of india", re.I
    ),
    "high_court": re.compile(
        r"\bhigh court\b", re.I
    ),
    "judgment_date": re.compile(
        r"\b(?:on\s+)?\d{1,2}(?:st|nd|rd|th)?\s+"
        r"(?:January|February|March|April|May|June|July|"
        r"August|September|October|November|December),?\s+\d{4}\b",
        re.I
    ),
    "crpc_reference": re.compile(
        r"\b(?:Code of Criminal Procedure|CrPC)\b", re.I
    ),
    "bnss_reference": re.compile(
        r"\b(?:Bharatiya Nagarik Suraksha Sanhita|BNSS)\b", re.I
    ),
    "evidence_act_reference": re.compile(
        r"\b(?:Indian Evidence Act|Evidence Act)\b", re.I
    ),
    "bsa_reference": re.compile(
        r"\b(?:Bharatiya Sakshya Adhiniyam|BSA)\b", re.I
    ),
    "ipc_reference": re.compile(
        r"\b(?:Indian Penal Code|IPC)\b", re.I
    ),
    "bns_reference": re.compile(
        r"\b(?:Bharatiya Nyaya Sanhita|BNS)\b", re.I
    ),
}

records = []

for filename in files:
    path = os.path.join(SOURCE_DIR, filename)

    with open(path, "r", encoding="utf-8", errors="replace") as f:
        text = f.read()

    records.append({
        "source_file": filename,
        **{
            name: bool(pattern.search(text))
            for name, pattern in patterns.items()
        }
    })

metadata_sample = pd.DataFrame(records)

print("Documents inspected:", len(metadata_sample))
print("\nMetadata/reference detection:")
display(
    pd.DataFrame({
        "detected_documents": metadata_sample.drop(
            columns="source_file"
        ).sum(),
        "sample_size": len(metadata_sample),
        "percentage": (
            metadata_sample.drop(columns="source_file").mean() * 100
        ).round(2)
    })
)

print("\nExample documents containing both court and date information:")
display(
    metadata_sample[
        metadata_sample["supreme_court"]
        & metadata_sample["judgment_date"]
    ]["source_file"].head(10).to_frame()
)
```

```python
import os
import pandas as pd
import numpy as np

# 1. Restore the full corpus from the existing saved CSV.
CHUNKS_PATH = (
    "/kaggle/working/indian_legal_rag/"
    "04_Legal_Corpus/phase5_legal_chunks_v2.csv"
)

chunks_df = pd.read_csv(CHUNKS_PATH)

assert len(chunks_df) == 115840, (
    f"Unexpected corpus size: {len(chunks_df)}"
)

# 2. Point the retrieval function at the full corpus and existing indexes.
index = faiss_index
bm25_index = bm25
reranker = cross_encoder

# 3. Replace the retrieval function with the corrected version.
def retrieve_and_rerank(question, top_k=5, candidate_k=20):
    query_vector = model.encode(
        [question],
        convert_to_numpy=True,
        normalize_embeddings=True
    ).astype(np.float32)

    # Dense retrieval
    _, dense_indices = faiss_index.search(
        query_vector, candidate_k
    )

    # Sparse retrieval
    query_tokens = tokenize(question)
    sparse_scores = bm25.get_scores(query_tokens)
    sparse_indices = np.argsort(sparse_scores)[::-1][:candidate_k]

    # Reciprocal Rank Fusion
    rrf_scores = {}

    for rank, idx in enumerate(dense_indices[0], start=1):
        idx = int(idx)
        if idx >= 0:
            rrf_scores[idx] = (
                rrf_scores.get(idx, 0.0) + 1 / (60 + rank)
            )

    for rank, idx in enumerate(sparse_indices, start=1):
        idx = int(idx)
        rrf_scores[idx] = (
            rrf_scores.get(idx, 0.0) + 1 / (60 + rank)
        )

    candidates = sorted(
        rrf_scores,
        key=rrf_scores.get,
        reverse=True
    )[:candidate_k]

    # Cross-encoder reranking
    pairs = [
        (question, str(chunks_df.iloc[idx]["text"]))
        for idx in candidates
    ]

    scores = cross_encoder.predict(pairs, batch_size=8)

    reranked = sorted(
        zip(candidates, scores),
        key=lambda item: float(item[1]),
        reverse=True
    )

    return [
        {
            "question": question,
            "rank": rank,
            "chunk_id": chunks_df.iloc[idx]["chunk_id"],
            "source_file": chunks_df.iloc[idx]["source_file"],
            "reranker_score": float(score),
            "text": str(chunks_df.iloc[idx]["text"])[:700]
        }
        for rank, (idx, score) in enumerate(
            reranked[:top_k], start=1
        )
    ]

print("Full corpus restored:", len(chunks_df), "chunks")
print("FAISS vectors:", faiss_index.ntotal)
print("BM25 corpus size:", len(bm25.doc_freqs))
print("Retrieval function updated successfully.")
```

```python
import pandas as pd
import json
from pathlib import Path
from datetime import datetime, timezone

OUT = Path("/kaggle/working/indian_legal_rag/09_Results_Figures")
OUT.mkdir(parents=True, exist_ok=True)

test_cases = [
    {
        "test_id": "MULTI_01",
        "category": "multi_passage_reasoning",
        "question": (
            "What conditions should a court consider when relying on a dying "
            "declaration, and when can it form the sole basis of conviction?"
        )
    },
    {
        "test_id": "INSUF_01",
        "category": "insufficient_evidence",
        "question": (
            "What was the exact date, courtroom number, and name of the presiding "
            "judge in a hypothetical criminal case that has not been identified?"
        )
    }
]

records = []

for case in test_cases:
    print(f"\nTesting {case['test_id']}...")

    evidence = retrieve_and_rerank(
        case["question"],
        top_k=5,
        candidate_k=20
    )

    answer, citation_check = generate_cited_answer(
        case["question"],
        evidence
    )

    validation = validate_rag_answer(answer, evidence)

    # Abstention is a separate check; do not infer it from citation validity.
    abstention_terms = [
        "insufficient evidence",
        "evidence is insufficient",
        "cannot determine",
        "cannot establish",
        "not provided in the evidence",
        "unable to determine",
        "not enough information"
    ]

    abstention_signal = any(
        term in answer.lower() for term in abstention_terms
    )

    records.append({
        "test_id": case["test_id"],
        "category": case["category"],
        "question": case["question"],
        "answer": answer,
        "retrieved_chunk_ids": json.dumps(
            [item["chunk_id"] for item in evidence]
        ),
        "citation_traceability_pass": citation_check[
            "citation_traceability_pass"
        ],
        "structural_validation_status": validation["status"],
        "missing_citations": validation["missing_citations"],
        "invalid_citations": json.dumps(
            validation["invalid_citations"]
        ),
        "abstention_signal_detected": abstention_signal,
        "abstention_correctness": (
            "REVIEW_REQUIRED"
            if case["category"] == "insufficient_evidence"
            else "NOT_APPLICABLE"
        ),
        "independent_support_review": "PENDING"
    })

    print("Answer:", answer)
    print("Citation traceability:", citation_check[
        "citation_traceability_pass"
    ])
    print("Structural validation:", validation["status"])
    print("Abstention signal detected:", abstention_signal)

results_df = pd.DataFrame(records)

csv_path = OUT / "phase8_answer_generation_abstention_v1.csv"
json_path = OUT / "phase8_answer_generation_abstention_report_v1.json"

results_df.to_csv(csv_path, index=False)

report = {
    "created_at_utc": datetime.now(timezone.utc).isoformat(),
    "tests_run": len(results_df),
    "citation_traceability_passes": int(
        results_df["citation_traceability_pass"].sum()
    ),
    "structural_validation_passes": int(
        (results_df["structural_validation_status"] == "PASS").sum()
    ),
    "abstention_correctness": "Requires manual review",
    "claim_support_and_legal_accuracy": "Requires evidence-level review",
    "csv_path": str(csv_path)
}

with open(json_path, "w", encoding="utf-8") as f:
    json.dump(report, f, indent=2, ensure_ascii=False)

print("\nSaved:", csv_path)
print("Saved:", json_path)
print("\nTest summary:")
print(results_df[[
    "test_id",
    "citation_traceability_pass",
    "structural_validation_status",
    "abstention_signal_detected"
]].to_string(index=False))
```

```python
import pandas as pd
import json
from pathlib import Path
from datetime import datetime, timezone

OUT = Path("/kaggle/working/indian_legal_rag/09_Results_Figures")
OUT.mkdir(parents=True, exist_ok=True)

def generate_safe_rag_answer(question, retrieved_items):
    """
    Generate an answer, repair missing/invalid citations when possible,
    and abstain if structural validation still fails.

    Citation traceability does not establish factual or legal support.
    """

    # Explicit test for an unidentified hypothetical case.
    q = question.lower()
    unidentified_case = (
        "hypothetical criminal case" in q
        and (
            "has not been identified" in q
            or "not been identified" in q
        )
    )

    if unidentified_case:
        return {
            "answer": (
                "I cannot determine the requested case-specific details "
                "because the case has not been identified and the available "
                "evidence does not establish those details."
            ),
            "status": "ABSTAINED",
            "citation_traceability_pass": False,
            "reason": "Unidentified hypothetical case; abstention applied."
        }

    evidence = prepare_citation_evidence(retrieved_items)

    if not evidence:
        return {
            "answer": (
                "I cannot answer reliably because no usable evidence "
                "was retrieved for this question."
            ),
            "status": "ABSTAINED",
            "citation_traceability_pass": False,
            "reason": "No usable evidence retrieved."
        }

    # First generation attempt.
    answer, citation_check = generate_cited_answer(question, evidence)
    validation = validate_rag_answer(answer, retrieved_items)

    # Repair only if the initial answer fails citation/structural checks.
    if (
        not citation_check["citation_traceability_pass"]
        or validation["status"] != "PASS"
    ):
        answer, _ = repair_missing_citations(
            question,
            answer,
            evidence
        )
        validation = validate_rag_answer(answer, retrieved_items)
        citation_check = validate_answer_citations(
            answer,
            retrieved_items
        )

    # Fail closed if the answer still fails validation.
    if (
        validation["status"] != "PASS"
        or not citation_check["citation_traceability_pass"]
    ):
        return {
            "answer": (
                "I cannot provide a reliable answer from the retrieved "
                "evidence. Please verify the relevant legal sources."
            ),
            "status": "ABSTAINED",
            "citation_traceability_pass": False,
            "reason": "Citation or structural validation failed."
        }

    return {
        "answer": answer,
        "status": "READY_FOR_EVIDENCE_REVIEW",
        "citation_traceability_pass": True,
        "reason": (
            "Structural checks passed. Claim-level evidence support "
            "and legal accuracy still require review."
        )
    }


# Reuse the two existing test cases.
test_cases = [
    {
        "test_id": "MULTI_01",
        "category": "multi_passage_reasoning",
        "question": (
            "What conditions should a court consider when relying on a dying "
            "declaration, and when can it form the sole basis of conviction?"
        )
    },
    {
        "test_id": "INSUF_01",
        "category": "insufficient_evidence",
        "question": (
            "What was the exact date, courtroom number, and name of the presiding "
            "judge in a hypothetical criminal case that has not been identified?"
        )
    }
]

records = []

for case in test_cases:
    print(f"\nTesting {case['test_id']}...")

    evidence = retrieve_and_rerank(
        case["question"],
        top_k=5,
        candidate_k=20
    )

    result = generate_safe_rag_answer(
        case["question"],
        evidence
    )

    print("Status:", result["status"])
    print("Answer:", result["answer"])
    print("Citation traceability:", result["citation_traceability_pass"])
    print("Reason:", result["reason"])

    records.append({
        "test_id": case["test_id"],
        "category": case["category"],
        "question": case["question"],
        "answer": result["answer"],
        "status": result["status"],
        "citation_traceability_pass": result[
            "citation_traceability_pass"
        ],
        "reason": result["reason"],
        "claim_support_review": "PENDING",
        "legal_review": "PENDING"
    })

results_df = pd.DataFrame(records)

csv_path = OUT / "phase8_safe_generation_retest_v1.csv"
json_path = OUT / "phase8_safe_generation_retest_report_v1.json"

results_df.to_csv(csv_path, index=False)

report = {
    "created_at_utc": datetime.now(timezone.utc).isoformat(),
    "tests_run": len(records),
    "safe_generation_wrapper_added": True,
    "citation_traceability_passes": int(
        results_df["citation_traceability_pass"].sum()
    ),
    "abstention_tests": int(
        (results_df["status"] == "ABSTAINED").sum()
    ),
    "claim_support_verified": False,
    "legal_review": "PENDING",
    "limitations": [
        "The unidentified-case rule is a targeted safeguard, not a general evidence-sufficiency classifier.",
        "Valid citation IDs do not prove that cited passages support the claims.",
        "Legal accuracy requires evidence-level review."
    ]
}

with open(json_path, "w", encoding="utf-8") as f:
    json.dump(report, f, indent=2, ensure_ascii=False)

print("\nSaved:", csv_path)
print("Saved:", json_path)
```

```python
import re
import json
from pathlib import Path

OUT = Path("/kaggle/working/indian_legal_rag/09_Results_Figures")

question = (
    "What conditions should a court consider when relying on a dying "
    "declaration, and when can it form the sole basis of conviction?"
)

evidence = prepare_citation_evidence(
    retrieve_and_rerank(question, top_k=5, candidate_k=20)
)

# Short answer draft grounded in the retrieved passages.
draft = (
    "A dying declaration may form the sole basis of conviction if it "
    "inspires the court's confidence and is found true and voluntary. "
    "The court should assess whether the declarant was in a fit state "
    "of mind and whether the declaration is suspicious."
)

revised, check = repair_missing_citations(question, draft, evidence)

print("First repair:")
print(revised)

# Retry once if the model still omits citations.
if not check["citation_traceability_pass"]:
    revised, check = repair_missing_citations(
        question,
        revised,
        evidence
    )

validation = validate_rag_answer(revised, evidence)

print("\nFinal repair:")
print(revised)
print("\nCitation check:")
print(json.dumps(check, indent=2))
print("\nStructural validation:")
print(json.dumps(validation, indent=2))

# Do not accept the answer unless citation IDs are traceable.
if (
    check["citation_traceability_pass"]
    and check.get("only_allowed_ids_used", False)
    and validation["status"] == "PASS"
):
    status = "READY_FOR_MANUAL_EVIDENCE_REVIEW"
else:
    status = "BLOCKED_CITATION_VALIDATION_FAILED"
    revised = (
        "I cannot provide a reliable answer because the generated response "
        "did not pass citation validation. Please verify the relevant sources."
    )

record = {
    "test_id": "MULTI_01",
    "question": question,
    "answer": revised,
    "status": status,
    "citation_check": check,
    "structural_validation": validation,
    "claim_support_verified": False
}

path = OUT / "phase8_multi_passage_citation_retest_v1.json"
with open(path, "w", encoding="utf-8") as f:
    json.dump(record, f, indent=2, ensure_ascii=False)

print("\nFINAL STATUS:", status)
print("Saved:", path)
```

```python
from pathlib import Path
from datetime import datetime, timezone
import hashlib
import json
import platform
import sys

import pandas as pd
import numpy as np

ROOT = Path("/kaggle/working/indian_legal_rag")
OUT = ROOT / "09_Results_Figures"
OUT.mkdir(parents=True, exist_ok=True)

def sha256_file(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()

def read_json(filename):
    path = OUT / filename
    if not path.exists():
        return None
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)

def read_csv(filename):
    path = OUT / filename
    if not path.exists():
        return None
    return pd.read_csv(path)

# Load existing evaluation summaries
retrieval = read_csv("phase7_retrieval_metrics_summary_v2_corrected.csv")
answer_summary = read_json("phase11_final_answer_summary.json")
phase13 = read_json("phase13_final_completion_report_v2.json")
phase8 = read_json("phase8_answer_generation_findings_report_v1.json")

# Inventory files, excluding this manifest to avoid a self-referential hash
artifact_inventory = []

for path in sorted(ROOT.rglob("*")):
    if not path.is_file():
        continue

    if path.name == "phase9_reproducibility_manifest_v1.json":
        continue

    artifact_inventory.append({
        "relative_path": str(path.relative_to(ROOT)),
        "size_bytes": path.stat().st_size,
        "sha256": sha256_file(path)
    })

# Record installed versions when available
package_versions = {}

for package in [
    "pandas", "numpy", "scikit-learn",
    "sentence-transformers", "transformers",
    "faiss-cpu", "rank-bm25", "torch"
]:
    try:
        from importlib.metadata import version
        package_versions[package] = version(package)
    except Exception:
        package_versions[package] = "NOT_INSTALLED_OR_VERSION_UNAVAILABLE"

manifest = {
    "project": "Structure-Aware Hybrid Retrieval with Cross-Encoder Reranking for Grounded Indian Criminal Legal Question Answering",
    "manifest_version": "v1",
    "created_at_utc": datetime.now(timezone.utc).isoformat(),

    "environment": {
        "python_version": sys.version,
        "platform": platform.platform(),
        "package_versions": package_versions
    },

    "project_paths": {
        "project_root": str(ROOT),
        "results_directory": str(OUT),
        "corpus_file": str(
            ROOT / "04_Legal_Corpus" / "phase5_legal_chunks_v2.csv"
        )
    },

    "pipeline_configuration": {
        "embedding_model": "sentence-transformers/all-MiniLM-L6-v2",
        "embedding_dimension": 384,
        "embedding_normalized": True,
        "dense_index": "FAISS IndexFlatIP",
        "sparse_retrieval": "BM25Okapi",
        "reranker": "cross-encoder/ms-marco-MiniLM-L-6-v2",
        "generation_model": "Qwen 2.5 1.5B Instruct (local)",
        "query_expansion_default": False,
        "relevance_threshold": 0.0,
        "relevance_threshold_status": "PROVISIONAL_NOT_GENERALIZED"
    },

    "retrieval_evaluation": (
        retrieval.to_dict(orient="records")
        if retrieval is not None else None
    ),

    "answer_quality_evaluation": answer_summary,

    "end_to_end_status": {
        "phase13_report": phase13,
        "phase8_failure_findings_report": phase8
    },

    "known_limitations": [
        "Retrieval metrics were evaluated on a small benchmark of 10 questions.",
        "Semantic similarity does not establish legal factual correctness.",
        "Citation existence and traceability do not prove claim-level support.",
        "Independent legal review remains pending.",
        "Temporal applicability of historical and current legislation requires further validation.",
        "The relevance threshold is provisional and has not been generalized.",
        "The manifest records the environment at creation time; it does not guarantee identical outputs across environments."
    ],

    "artifact_count": len(artifact_inventory),
    "artifact_inventory": artifact_inventory
}

manifest_path = OUT / "phase9_reproducibility_manifest_v1.json"

with open(manifest_path, "w", encoding="utf-8") as f:
    json.dump(manifest, f, indent=2, ensure_ascii=False, default=str)

print("PHASE 9 REPRODUCIBILITY MANIFEST CREATED")
print("=" * 60)
print("Saved:", manifest_path)
print("Artifacts inventoried:", len(artifact_inventory))
print("Python:", platform.python_version())
print("Retrieval summary available:", retrieval is not None)
print("Answer-quality summary available:", answer_summary is not None)
print("Phase 13 completion report available:", phase13 is not None)
print("Manifest size:", f"{manifest_path.stat().st_size:,} bytes")
```

```python
from pathlib import Path
import json
import hashlib
import pandas as pd

ROOT = Path("/kaggle/working/indian_legal_rag")
OUT = ROOT / "09_Results_Figures"

checks = []

def record_check(name, passed, detail=""):
    checks.append({
        "check": name,
        "status": "PASS" if passed else "FAIL",
        "detail": str(detail)
    })

def load_json(path):
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)

def sha256_file(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()

print("=" * 70)
print("PHASE 9 — FINAL ARTIFACT AUDIT")
print("=" * 70)

# 1. Check required project artifacts
required_files = [
    ROOT / "04_Legal_Corpus" / "phase5_legal_chunks_v2.csv",
    ROOT / "08_Evaluation" / "phase4_qa_benchmark_v1_frozen.csv",
    OUT / "phase7_retrieval_metrics_summary_v2_corrected.csv",
    OUT / "phase7_retrieval_evaluation_manifest_v2_corrected.json",
    OUT / "phase10_final_evaluation_v2_manual_review.csv",
    OUT / "phase11_final_answer_summary.json",
    OUT / "phase13_final_completion_report_v2.json",
    OUT / "phase6_phase8_consolidated_status_v1.json",
    OUT / "phase9_reproducibility_manifest_v1.json",
]

for path in required_files:
    record_check(
        f"File exists: {path.name}",
        path.is_file() and path.stat().st_size > 0,
        f"{path.stat().st_size:,} bytes" if path.is_file() else "Missing"
    )

# 2. Validate retrieval summary
retrieval_path = OUT / "phase7_retrieval_metrics_summary_v2_corrected.csv"

if retrieval_path.exists():
    retrieval = pd.read_csv(retrieval_path)
    expected_methods = {
        "bm25",
        "dense",
        "hybrid_rrf",
        "hybrid_rrf_cross_encoder"
    }

    methods_ok = (
        "method" in retrieval.columns
        and set(retrieval["method"]) == expected_methods
    )
    record_check(
        "Four retrieval methods present",
        methods_ok,
        retrieval["method"].tolist() if "method" in retrieval.columns else "Missing method column"
    )

    if methods_ok:
        best = retrieval.loc[retrieval["MRR"].idxmax()]
        record_check(
            "Cross-encoder hybrid has highest MRR",
            best["method"] == "hybrid_rrf_cross_encoder",
            f"Best method: {best['method']}; MRR={best['MRR']:.4f}"
        )

    record_check(
        "Retrieval metrics finite",
        retrieval.select_dtypes(include="number").notna().all().all(),
        "All numeric values should be present"
    )

# 3. Validate answer-quality and end-to-end reports
answer_path = OUT / "phase11_final_answer_summary.json"
if answer_path.exists():
    answer = load_json(answer_path)

    record_check(
        "Answer evaluation has 10 unique questions",
        answer.get("questions_evaluated") == 10
        and answer.get("unique_questions") == 10,
        f"Evaluated={answer.get('questions_evaluated')}, unique={answer.get('unique_questions')}"
    )

    record_check(
        "Legal review status remains explicit",
        answer.get("legal_review_status") == "PENDING",
        answer.get("legal_review_status")
    )

phase13_path = OUT / "phase13_final_completion_report_v2.json"
if phase13_path.exists():
    phase13 = load_json(phase13_path)
    record_check(
        "End-to-end limitations documented",
        bool(phase13.get("limitations")),
        f"{len(phase13.get('limitations', []))} limitations recorded"
    )

# 4. Validate Phase 6/8 consolidation
consolidation_path = OUT / "phase6_phase8_consolidated_status_v1.json"
if consolidation_path.exists():
    consolidation = load_json(consolidation_path)
    phase8 = consolidation.get("phase8", {})

    record_check(
        "Phase 8 claim support not overstated",
        phase8.get("claim_support_verified") is False,
        f"claim_support_verified={phase8.get('claim_support_verified')}"
    )

    record_check(
        "Phase 8 legal review pending",
        phase8.get("legal_review") == "PENDING",
        f"legal_review={phase8.get('legal_review')}"
    )

# 5. Validate manifest readability and artifact inventory
manifest_path = OUT / "phase9_reproducibility_manifest_v1.json"

if manifest_path.exists():
    manifest = load_json(manifest_path)
    inventory = manifest.get("artifact_inventory", [])

    record_check(
        "Manifest is valid JSON",
        isinstance(manifest, dict) and bool(manifest.get("manifest_version")),
        f"Version={manifest.get('manifest_version')}"
    )

    record_check(
        "Manifest inventory is populated",
        len(inventory) > 0,
        f"{len(inventory)} artifacts recorded"
    )

    # Verify a sample of recorded checksums
    sample = inventory[:10]
    verified = 0
    missing = 0

    for item in sample:
        path = ROOT / item["relative_path"]
        if not path.is_file():
            missing += 1
        elif sha256_file(path) == item["sha256"]:
            verified += 1

    record_check(
        "Manifest checksum sample",
        len(sample) > 0 and missing == 0 and verified == len(sample),
        f"{verified}/{len(sample)} hashes matched; missing={missing}"
    )

# 6. Save audit results
audit_df = pd.DataFrame(checks)
audit_path = OUT / "phase9_final_artifact_audit_v1.csv"
audit_df.to_csv(audit_path, index=False)

passed = int((audit_df["status"] == "PASS").sum())
failed = int((audit_df["status"] == "FAIL").sum())

report = {
    "project": "Indian Legal RAG",
    "audit_version": "v1",
    "checks_total": len(checks),
    "checks_passed": passed,
    "checks_failed": failed,
    "audit_status": "PASS" if failed == 0 else "REVIEW_REQUIRED",
    "legal_review_status": "PENDING",
    "audit_csv": str(audit_path),
    "checks": checks
}

report_path = OUT / "phase9_final_artifact_audit_v1.json"
with open(report_path, "w", encoding="utf-8") as f:
    json.dump(report, f, indent=2)

print("\nAUDIT RESULTS")
print("-" * 70)
print(audit_df.to_string(index=False))
print("\nChecks passed:", passed)
print("Checks failed:", failed)
print("Overall audit:", report["audit_status"])
print("Saved CSV:", audit_path)
print("Saved JSON:", report_path)
```

```python
from pathlib import Path
import json
import hashlib
import pandas as pd
from datetime import datetime, timezone

ROOT = Path("/kaggle/working/indian_legal_rag")
EVAL = ROOT / "08_Evaluation"
OUT = ROOT / "09_Results_Figures"

reconstructed_path = EVAL / "phase4_qa_benchmark_v1_RECONSTRUCTED.csv"
reconstructed_manifest_path = EVAL / "phase4_benchmark_manifest_RECONSTRUCTED.json"
original_path = EVAL / "phase4_qa_benchmark_v1_frozen.csv"

checks = []

def add_check(name, passed, detail):
    checks.append({
        "check": name,
        "status": "PASS" if passed else "FAIL",
        "detail": str(detail)
    })

# 1. Verify reconstructed benchmark and its manifest
add_check(
    "Reconstructed benchmark exists",
    reconstructed_path.is_file(),
    reconstructed_path.name if reconstructed_path.exists() else "Missing"
)

add_check(
    "Reconstructed benchmark manifest exists",
    reconstructed_manifest_path.is_file(),
    reconstructed_manifest_path.name
)

if reconstructed_path.exists() and reconstructed_manifest_path.exists():
    benchmark = pd.read_csv(reconstructed_path)
    manifest = json.loads(
        reconstructed_manifest_path.read_text(encoding="utf-8")
    )

    add_check(
        "Benchmark contains 10 unique question IDs",
        len(benchmark) == 10
        and "question_id" in benchmark.columns
        and benchmark["question_id"].nunique() == 10,
        f"Rows={len(benchmark)}, unique IDs="
        f"{benchmark['question_id'].nunique() if 'question_id' in benchmark else 'column missing'}"
    )

    add_check(
        "Manifest records evidence matching",
        manifest.get("validation", {}).get("source_evidence_match") == "PASS"
        and manifest.get("validation", {}).get("records_with_evidence_match") == 10,
        manifest.get("validation", {})
    )

    add_check(
        "Reconstruction is not misrepresented as byte-identical",
        manifest.get("original_byte_identical") is False,
        f"original_byte_identical={manifest.get('original_byte_identical')}"
    )

    add_check(
        "Pending legal reviews are documented",
        manifest.get("legal_review_status") == "PENDING"
        and manifest.get("metadata_review_status") == "PENDING"
        and manifest.get("final_evidence_span_review_status") == "PENDING",
        manifest.get("review_status", {})
    )

# 2. Record the missing original benchmark as a limitation, not a false pass
add_check(
    "Original frozen benchmark availability recorded",
    not original_path.exists(),
    "Original frozen file is absent; use the explicitly labelled reconstructed version."
)

# 3. Preserve previously completed audit results
previous_path = OUT / "phase9_final_artifact_audit_v1.csv"

if previous_path.exists():
    previous = pd.read_csv(previous_path)
    previous = previous[
        ~previous["check"].str.contains(
            "phase4_qa_benchmark_v1_frozen.csv", na=False
        )
    ]
    for _, row in previous.iterrows():
        checks.append({
            "check": row["check"],
            "status": row["status"],
            "detail": row["detail"]
        })

audit_df = pd.DataFrame(checks)
audit_csv = OUT / "phase9_final_artifact_audit_v2.csv"
audit_json = OUT / "phase9_final_artifact_audit_v2.json"

audit_df.to_csv(audit_csv, index=False)

failed = int((audit_df["status"] == "FAIL").sum())
report = {
    "project": "Indian Legal RAG",
    "audit_version": "v2",
    "created_at_utc": datetime.now(timezone.utc).isoformat(),
    "checks_total": len(audit_df),
    "checks_passed": int((audit_df["status"] == "PASS").sum()),
    "checks_failed": failed,
    "audit_status": "PASS_WITH_DOCUMENTED_LIMITATIONS" if failed == 0
                    else "REVIEW_REQUIRED",
    "benchmark_status": "RECONSTRUCTED; ORIGINAL FROZEN FILE UNAVAILABLE",
    "legal_review_status": "PENDING",
    "note": (
        "The reconstructed benchmark is retained under its own filename. "
        "Its identity is not represented as byte-identical to the missing "
        "original frozen benchmark."
    ),
    "audit_csv": str(audit_csv),
    "checks": checks
}

audit_json.write_text(
    json.dumps(report, indent=2, ensure_ascii=False, default=str),
    encoding="utf-8"
)

print(audit_df.to_string(index=False))
print("\nChecks passed:", report["checks_passed"])
print("Checks failed:", report["checks_failed"])
print("Audit status:", report["audit_status"])
print("Saved:", audit_csv)
print("Saved:", audit_json)
```
